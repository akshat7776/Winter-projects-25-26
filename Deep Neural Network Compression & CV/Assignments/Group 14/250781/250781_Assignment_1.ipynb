````python
rows, cols = fshift.shape
crow, ccol = rows // 2, cols // 2 #as the circle should be in the mid point
````

````python
fshift.shape
````

````python
mask_lpf_10=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_10, (ccol,crow),10,1,-1)
mask_lpf_30=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_30, (ccol,crow),30,1,-1)
mask_lpf_80=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_80, (ccol,crow),80,1,-1)
````

````python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))

ax[0].imshow(mask_lpf_10, cmap='gray')
ax[0].set_title("R=10")
ax[0].axis("off")

ax[1].imshow(mask_lpf_30, cmap='gray')
ax[1].set_title("R=30")
ax[1].axis("off")

ax[2].imshow(mask_lpf_80, cmap='gray')
ax[2].set_title("R=80")
ax[2].axis("off")

plt.show()
````

````python
mask10 = mask_lpf_10.astype(np.float32)
mask30 = mask_lpf_30.astype(np.float32)
mask80 = mask_lpf_80.astype(np.float32)

masks = [(mask10, 10), (mask30, 30), (mask80, 80)]
outputs = []
````

````python
for mask, R in masks:
  f_lpf = fshift * mask
  lpf_ishift = np.fft.ifftshift(f_lpf)
  img_back_lpf = np.fft.ifft2(lpf_ishift)
  img_back_lpf = np.abs(img_back_lpf)
  outputs.append((img_back_lpf, R))
````

````python
fig, ax = plt.subplots(3, 1, figsize=(25, 10))

ax[0].imshow(outputs[0][0], cmap='gray')
ax[0].set_title("R=10")
ax[0].axis("off")

ax[1].imshow(outputs[1][0], cmap='gray')
ax[1].set_title("R=30")
ax[1].axis("off")

ax[2].imshow(outputs[2][0], cmap='gray')
ax[2].set_title("R=80")
ax[2].axis("off")

plt.show()

#we can see, the lesser the radius, more blurred the image. as the lesser the radius, higher frequencies are removed more
````

````python
hpfmasks=[(1 - i,r) for i,r in masks] #formula from slides
````

````python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))

ax[0].imshow(hpfmasks[0][0], cmap='gray')
ax[0].set_title("R=10")
ax[0].axis("off")

ax[1].imshow(hpfmasks[1][0], cmap='gray')
ax[1].set_title("R=30")
ax[1].axis("off")

ax[2].imshow(hpfmasks[2][0], cmap='gray')
ax[2].set_title("R=80")
ax[2].axis("off")

plt.show()
````

````python
outputs_hpf=[]
for mask, R in hpfmasks:
  f_hpf = fshift * mask
  hpf_ishift = np.fft.ifftshift(f_hpf)
  img_back_hpf = np.fft.ifft2(hpf_ishift)
  img_back_hpf = np.abs(img_back_hpf)
  outputs_hpf.append((img_back_hpf, R))
````

````python
fig, ax = plt.subplots(3, 1, figsize=(25, 10))

ax[0].imshow(outputs_hpf[0][0], cmap='gray')
ax[0].set_title("R=10")
ax[0].axis("off")

ax[1].imshow(outputs_hpf[1][0], cmap='gray')
ax[1].set_title("R=30")
ax[1].axis("off")

ax[2].imshow(outputs_hpf[2][0], cmap='gray')
ax[2].set_title("R=80")
ax[2].axis("off")

plt.show()

#we can see, the lesser the radius, lesser low frequencies are removed and the output resembles original image more than the outline
````

````python
import base64
import numpy as np
import cv2
import matplotlib.pyplot as plt

img_bytes2 = base64.b64decode(imgstring2)
img_array2 = np.frombuffer(img_bytes2, dtype=np.uint8)
img2 = cv2.imdecode(img_array2, cv2.IMREAD_COLOR)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

img_bytes1 = base64.b64decode(imgstring)
img_array1 = np.frombuffer(img_bytes1, dtype=np.uint8)
img1 = cv2.imdecode(img_array1, cv2.IMREAD_COLOR)
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

plt.imshow(img2)
plt.axis("off")
````

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Resize second image to match first (required for FFT swap)
if img1.shape != img2.shape:
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

gray1=cv2.cvtColor(img1,cv2.COLOR_RGB2GRAY)
gray2=cv2.cvtColor(img2,cv2.COLOR_RGB2GRAY)

F1 = np.fft.fft2(gray1)
F2 = np.fft.fft2(gray2)

F1_shift = np.fft.fftshift(F1)
F2_shift = np.fft.fftshift(F2)

mag1 = np.abs(F1_shift)
mag2 = np.abs(F2_shift)

phase1 = np.angle(F1_shift)
phase2 = np.angle(F2_shift)

swap1 = mag1 * np.exp(1j * phase2)
swap2 = mag2 * np.exp(1j * phase1)

new1 = np.fft.ifft2(np.fft.ifftshift(swap1))
new2 = np.fft.ifft2(np.fft.ifftshift(swap2))

new1 = np.abs(new1)
new2 = np.abs(new2)

new1 = new1 / new1.max()
new2 = new2 / new2.max()

plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.title("Image 1")
plt.imshow(gray1, cmap='gray')
plt.axis("off")

plt.subplot(2,3,2)
plt.title("Image 2")
plt.imshow(gray2, cmap='gray')
plt.axis("off")

plt.subplot(2,3,4)
plt.title("Magnitude 1 + Phase 2")
plt.imshow(new1, cmap='gray', vmin=0, vmax=1)
plt.axis("off")

plt.subplot(2,3,5)
plt.title("Magnitude 2 + Phase 1")
plt.imshow(new2, cmap='gray', vmin=0, vmax=1)
plt.axis("off")

plt.tight_layout()
plt.show()

#Yes, after this i realise that the phase is actually what controls the structure and determines whether we can see the image or not
#The magnitude seems to control the intensity/brightness of the image as a whole
# we can see in the result
# one image is modern art (as suggested by host in a meeting "picasso art") with a lot of contrast
# and the other image is natural with very less contrast
````
