````python
import numpy as np
import cv2
#making gray_histogram function
def gray_histogram(gray_img):
  histogram=np.zeros(256)
  row,col=gray_img.shape

  for i in range(row):
    for j in range(col):
      histogram[gray_img[i,j]]+=1

  return histogram

img_1=np.array(img_1)
gray_img=cv2.cvtColor(img_1,cv2.COLOR_RGB2GRAY)
hist=gray_histogram(gray_img)

plt.plot(hist)
plt.title("Grayscale Histogram")
plt.xlabel("Intensity Value")
plt.ylabel("Pixel Count")
plt.show()
````

````python
#making rgb_h function
def rgb_h(rgb_img):

  r_h=np.zeros(256)
  g_h=np.zeros(256)
  b_h=np.zeros(256)

  row,col,_=rgb_img.shape

  for i in range(row):
    for j in range(col):
      r_h[rgb_img[i,j,0]]+=1
      g_h[rgb_img[i,j,1]]+=1
      b_h[rgb_img[i,j,2]]+=1

  return r_h,g_h,b_h


img_1=np.array(img_1)
r_hist,g_hist,b_hist=rgb_h(img_1)

plt.plot(r_hist,color='red')
plt.plot(g_hist,color='green')
plt.plot(b_hist,color='blue')
plt.title("RGB Histogram")
plt.xlabel("Intensity Value")
plt.ylabel("Pixel Count")
plt.show()
````

````python
#Code for hsv conversion
import cv2
def rgb_to_hsv(img):
  img=np.array(img)
  h,w,_=img.shape
  hsv=np.zeros((h,w,3),dtype=np.uint8)

  for i in range(h):
    for j in range(w):

      #First we need to normalise r g and b channels
      r,g,b=img[i,j]
      r/=255.0
      g/=255.0
      b/=255.0

      max_color=max(r,g,b)
      min_color=min(r,g,b)

      if (max_color-min_color)==0:
        H=0
      elif max_color==r:
        H=60*(g-b)/(max_color-min_color)
      elif max_color==g:
        H=120+60*(b-r)/(max_color-min_color)
      else:
        H=240+60*(r-g)/(max_color-min_color)

      H=H%360  #If out of bounds we need to bring it back to 0-2pi. This can happen when H<0

      if max_color==0:
        S=0
      else:
        S=(max_color-min_color)/max_color

      V=max_color

      #We now have our required H,S,V channels
      hsv[i,j,0]=int(H/2)  #Scaling factor since cv2 takes values from 0-179
      hsv[i,j,1]=int(S*255)
      hsv[i,j,2]=int(V*255)

  return hsv


#Now lets display original rgb image and rgb image made from hsv using cv2.cvtColor

original_img=img
hsv_img=rgb_to_hsv(original_img)
converted_img=cv2.cvtColor(hsv_img,cv2.COLOR_HSV2RGB)

plt.subplot(1,2,1)
plt.imshow(original_img)
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(converted_img)
plt.axis("off")
````

````python
#Filter-1: First I tried making a cyberpunk inspired filter, but couldn't pull it off int time, hence I decided to turn the attempt into a neon type filter.

#I Chose following modifications: CONTRAST BOOSTING, SATURATION CHANGE, SPLIT-TONING.
#Contrast Boosting: Makes the lights and shadows stand out in the night scenery, giving image a punchier look

#Saturation Change: Enhances vivid neon colours and make signs pop more clearly

#Split-toning: Add cyan tones to shadows and magenta tones to highlights for the neon effect (reminiscent of my attempt at a cyberpunk filter)

import cv2
def neon_night(img):

  img=np.array(img)

  #First boosting contrast around midpoint
  out=img.copy()
  mean_val=out.mean()
  out=(out-mean_val)*1.5+mean_val
  out=np.clip(out,0,255).astype(np.uint8)

  #Now applying saturation boost by converting image to HSV using function made in PART B
  out_hsv = cv2.cvtColor(out, cv2.COLOR_RGB2HSV).astype(np.float32)
  out_hsv[:,:,1]=out_hsv[:,:,1]*1.3
  out_hsv[:,:,1]=np.clip(out_hsv[:,:,1],0,255)
  out_hsv=out_hsv.astype(np.uint8)

  out=cv2.cvtColor(out_hsv,cv2.COLOR_HSV2RGB)

  #Now doing split-toning to make shadows more cyan and add magenta highlights
  gray_out=cv2.cvtColor(out,cv2.COLOR_RGB2GRAY)

  shadow_mask=gray_out<128
  highlight_mask=gray_out>128

  out_int=out.astype(np.int16)
  out_int[shadow_mask]+=np.array([0,40,40],dtype=np.int16) #Since cyan=green+blue, hence we increase the green and blue channels equally
  out_int[highlight_mask]+=np.array([40,0,40],dtype=np.int16) #since magenta=red+blue
  out=np.clip(out_int,0,255).astype(np.uint8)

  return out

img_filter_1=neon_night(img_1)

plt.figure(figsize=(15,8))
plt.subplot(1,2,1)
plt.imshow(img_1)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_filter_1)
plt.title("Neon Night")
plt.axis("off")
````

````python
#Filter-2: In minecraft I find myself wishing that caves looked brighter. Hence I used a gamma correction filter to solve the problem of low visibility in cave systems
#modifications used were gamma correction and brightness adjustment.

#Gamma correction: Brings out hidden cave details without blowing out brighter areas
#Brightness adjusment: Adjusts overall exposure so final environment becomes easier to see

def cave_brightner(img):
    img=np.array(img)
    img=img.astype(np.float32)

    #Gamma correction to increase visibility in image
    gamma=0.55
    img=255*((img/255)**gamma)

    #Brightness increase
    img=img+5
    img=np.clip(img,0,255).astype(np.uint8)

    return img


img_filter_1=cave_brightner(img_2)
plt.figure(figsize=(15,8))
plt.subplot(1,2,1)
plt.imshow(img_2)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_filter_1)
plt.title("Filtered")
plt.axis("off")
````

````python
# @title
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np

# PASTE BASE64 STRING HERE
base64_image = """

"""

img_data = base64.b64decode(base64_image)
img_3 = Image.open(io.BytesIO(img_data))

plt.imshow(img_3)
plt.axis("off")
````

````python
def white_patch_balance(img_rgb):
  img=np.array(img_rgb).astype(np.float32)

  max_val=np.max(img)
  scale=255.0/max_val

  out=img*scale
  out=np.clip(out,0,255).astype(np.uint8)
  return out


balanced_img=white_patch_balance(img_3)

plt.figure(figsize=(15,8))
plt.subplot(1,2,1)
plt.imshow(img_3)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(balanced_img)
plt.title("White Patch Balanced")
plt.axis("off")
plt.show()
````
