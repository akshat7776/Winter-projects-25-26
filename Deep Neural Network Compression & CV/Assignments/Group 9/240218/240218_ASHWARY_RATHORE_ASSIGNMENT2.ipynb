````python
import numpy as np # importing all the libraries
import matplotlib.pyplot as plt
from PIL import Image
````

````python

def grayhist(gray_img): # this function computes gray histogram from 0-255 and return the histogram of 256 size with the help of pure numpy loops
  hist=np.zeros(256,dtype=int)
  for pixel in gray_img.flat: #the iteration is done pixel by pixel
      hist[pixel]+=1
  return hist
````

````python




def rgbhist(rgb_img):# this function will compute the R,G,B channels histogram using pure numPy loops
#returning the color combinations red green and blue
    hist_R=np.zeros(256,dtype=int)
    hist_G=np.zeros(256,dtype=int)
    hist_B=np.zeros(256,dtype=int)

    H,W,_ = rgb_img.shape

    for i in range(H):          # iterate rows and column both pixel by pixel
        for j in range(W):
            r,g,b=rgb_img[i,j]
            hist_R[r]+=1
            hist_G[g]+=1
            hist_B[b]+=1

    return hist_R,hist_G,  hist_B


#now we will load the image
img=np.array(Image.open("my_image1.jpg").convert("RGB"))

#next converting into the gray manually
gray=(0.299*img[...,0]+0.587*img[...,1]+0.114*img[...,2]).astype(np.uint8)

# computing the histograms
hist_gray=grayhist(gray)
hist_R, hist_G, hist_B=rgbhist(img)






````

````python
plt.figure(figsize=(7,4))#  ploting the gray histogram
plt.plot(hist_gray, color='black')
plt.title("grayscale histogram")
plt.xlabel("intensity (0–255)")
plt.ylabel("pixel count")
plt.grid(alpha=0.3)
plt.show()

````

````python
plt.figure(figsize=(7,4)) #plotting the RGB histogram
plt.plot(hist_R, color='red', label='Red')
plt.plot(hist_G, color='green', label='Green')
plt.plot(hist_B, color='blue', label='Blue')
plt.title("RGB channel histograms")
plt.xlabel("intensity (0–255)")
plt.ylabel("pixel count")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

````python
import cv2
````

````python
def rgbtohsv(rgb):
    rgb=rgb.astype(np.float32)/255.0 #using an RGB image this function determines the hue, saturation


    r,g,b=rgb[...,0],rgb[...,1],rgb[...,2]
    mx,mn=np.max(rgb,2),np.min(rgb,2)
    d=mx-mn

    h=np.zeros_like(mx)
    mask=d!=0
    h[mask & (mx==r)]=((g-b)[mask&(mx==r)]/d[mask& (mx==r)])%6
    h[mask & (mx==g)]=((b-r)[mask&(mx==g)]/d[mask &(mx==g)])+2     #IMPORTANT NOTE AI is used here but i have understood whta is happening in this particualr part of the code
    h[mask & (mx==b)]=((r-g)[mask&(mx==b)]/d[mask&(mx==b)])+4
    h=(h*60)%360

    s=np.where(mx==0,0,d/mx)
    v=mx

    H=(h/360*179).astype(np.uint8)# here we are finding the hsv values and then coverting to openCV usable format
    S=(s*255).astype(np.uint8)
    V=(v*255).astype(np.uint8)

    return np.stack([H,S,V],-1)
````

````python
rgb = np.array(Image.open("my_image1.jpg").convert("RGB"))
hsv = rgbtohsv(rgb)


rgb_back = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)
plt.figure(figsize=(10,4))
#plotting the original iamge
plt.subplot(1,2,1)
plt.imshow(rgb)
plt.title("original")
plt.axis("off")



#plotting the reconverter image as a subplot
plt.subplot(1,2,2)
plt.imshow(rgb_back)
plt.title("reconverted")
plt.axis("off")

plt.show()



print("max pixel difference:", # this is to determine the maximum difference in the pixels
      np.max(np.abs(rgb.astype(int) - rgb_back.astype(int))))
````

````python
import cv2
import numpy as np

def clamp(x):
    return np.clip(x, 0, 255).astype(np.uint8)
````

````python

#filter 1 SUPERAURA999
#SuperAura999 adds warm colors and gentle sharpness to make the image brighter, clearer, and richer in detail


def SuperAura999(img):
    img = img.astype(np.float32)

      # contrast + brightness + warm tint
    img=img*1.1+8
    img[...,0]*= 1.97
    img[...,1]*= 0.85
    img[...,2]*= 0.95



  #saturation boost of the filter
    hsv = cv2.cvtColor(clamp(img), cv2.COLOR_RGB2HSV).astype(np.float32)
    hsv[...,1] *= 0.95
    img = cv2.cvtColor(clamp(hsv), cv2.COLOR_HSV2RGB).astype(np.float32)

       #light sharpening
    blur=cv2.GaussianBlur(clamp(img),(0,0),3)
    img=img+(img-blur)*0.5

    return clamp(img)
````

````python
import cv2
img = cv2.imread("your_image3.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

out =SuperAura999(img)

cv2.imwrite("your_image3_tropical.jpg", cv2.cvtColor(out, cv2.COLOR_RGB2BGR))
````

#### White Balance Function
Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image. Use Numpy only to keep track of the brightest pixel and then scale the entire image with the factor you find. Display the original and output image.

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
````

````python
import numpy as np
import matplotlib.pyplot as plt

def whitepatch(img_rgb): #this function actually finds the brightest patch in the whole image and then scale that pixel and scale it to whole image
    img=img_rgb.astype(np.float32)
    maxval=np.max(img)
    scale=255.0/maxval

    balance=img*scale
    balance=np.clip(balance,0,255).astype(np.uint8)

    return balance
````

````python
import cv2
img = plt.imread("your_image6.jpg")

out=whitepatch(img)

cv2.imwrite("image6 balanced.jpg", out)
````

````python
# assume img is loaded as an RGB image using matplotlib or other method

output = whitepatch(img)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(output)
plt.title("White Patch Balanced")
plt.axis('off')

plt.show()
````

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def whitefunction(img_rgb,value):
    img=img_rgb.astype(np.float32)
    hotfilter=np.array([1.5,1.0,0.5])
    coldfilter=np.array([0.5,1.0,1.5])
    if value<0.5:
        t=value*2
        scale=coldfilter*(1-t)+np.ones(3)*t
    else:
        t=(value-0.5)*2
        scale=hotfilter*(t)+np.ones(3)*(1-t)

    img*=scale
    return np.clip(img,0,255).astype(np.uint8)
````

````python
import cv2
import numpy as np
img=cv2.imread("your_image3.jpg")
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

val=0.8   # try 0, 0.3, 0.5, 0.8, 1.0
wb_img=whitefunction(img, val)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(wb_img)
plt.title(f"White Balanced (value={val})")
plt.axis("off")

plt.show()
````
