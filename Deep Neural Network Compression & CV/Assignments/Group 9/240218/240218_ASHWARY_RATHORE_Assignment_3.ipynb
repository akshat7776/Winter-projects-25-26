````python
def convolve(img,kernel):
    img=img.astype(np.float32)
    kernel=np.flipud(np.fliplr(kernel))
    kh,kw=kernel.shape
    pad_h,pad_w=kh//2,kw//2
    padded=np.pad(img,((pad_h,pad_h),(pad_w,pad_w)),mode='reflect')
    output=np.zeros_like(img,dtype=np.float32)
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region=padded[i:i+kh,j:j+kw]
            output[i,j]=np.sum(region*kernel)
    return output
#this code slides a flipped kernel over the image, multiplies it with local regions, sums the result
# and builds a new image that is filtered according to the kernel you give
````

````python
def average_blur_gray(img,ksize):
    kernel=np.ones((ksize,ksize),dtype=np.float32)
    kernel/=kernel.sum()
    return convolve(img,kernel)
#this function makes an average filter by filling the kernel with equal values and normalizing it.
#it blurs the grayscale image by averaging nearby pixels using the convolve function.
````

````python
def average_blur_rgb(img,ksize):
    channels=[]
    for c in range(3):
        channels.append(average_blur_gray(img[:,:,c],ksize))
    return np.stack(channels,axis=2)
#this function applies average blur separately to each of the three rgb channels using the grayscale blur function.
#after blurring all channels, it stacks them back together to form the final blurred color image
````

````python
def gaussian_kernel(ksize,sigma):
    ax=np.arange(-ksize//2+1.,ksize//2+1.)
    xx,yy=np.meshgrid(ax,ax)
    kernel=np.exp(-(xx**2+yy**2)/(2*sigma**2))
    kernel/=kernel.sum()
    return kernel
#this function creates a gaussian kernel using the given size and sigma, where center values are higher and edges are lower.
#the kernel is then normalized so all values add up to 1, keeping image brightness unchanged.
````

````python
def gaussian_blur_gray(img,ksize,sigma):
    kernel=gaussian_kernel(ksize,sigma)
    return convolve(img,kernel)
#this function does the blur on the grayscale immage
````

````python
def gaussian_blur_rgb(img,ksize,sigma):
    channels=[]
    for c in range(3):
        channels.append(gaussian_blur_gray(img[:,:,c],ksize,sigma))
    return np.stack(channels,axis=2)
#same doing gaussian blur on rgb image
````

````python
import cv2
img=cv2.imread("my_image.jpg")
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)

avg_gray=average_blur_gray(gray,5)
avg_rgb=average_blur_rgb(img,5)

gauss_gray=gaussian_blur_gray(gray,5,1.0)
gauss_rgb=gaussian_blur_rgb(img,5,1.0)

plt.figure(figsize=(10,8))
plt.subplot(2,2,1);plt.imshow(img);plt.title("Original");plt.axis("off")
plt.subplot(2,2,2);plt.imshow(avg_rgb.astype(np.uint8));plt.title("Average Blur");plt.axis("off")
plt.subplot(2,2,3);plt.imshow(gauss_rgb.astype(np.uint8));plt.title("Gaussian Blur");plt.axis("off")
plt.subplot(2,2,4);plt.imshow(gauss_gray,cmap='gray');plt.title("Gaussian Gray");plt.axis("off")
plt.show()
````
