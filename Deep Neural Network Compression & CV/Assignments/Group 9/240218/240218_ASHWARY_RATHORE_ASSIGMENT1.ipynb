````python
import numpy as np
import matplotlib.pyplot as plt



rows, cols = gray_np.shape
crow, ccol = rows // 2, cols // 2
````

````python
# radii for circular low-pass filters
radii = [10, 20, 100]

lpf_results = []

for r in radii:
    mask = np.zeros((rows, cols), np.uint8)

    y, x = np.ogrid[:rows, :cols]
    distance = np.sqrt((y - crow)**2 + (x - ccol)**2)
    mask[distance <= r] = 1

    fshift_lpf = fshift * mask

    f_ishift = np.fft.ifftshift(fshift_lpf)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    lpf_results.append((r, img_back))

    plt.figure(figsize=(5, 5))
    plt.imshow(img_back, cmap='gray')
    plt.title(f"lpf radius = {r}")
    plt.axis('off')
    plt.show()
````

````python
#we pick different radii to control how much low-frequency content passes

#we make a zero mask and draw a circle of 1s at the center

#this circle keeps low frequencies and blocks high ones

#we apply the mask, then do inverse fft to get the filtered image back

#small radius → very blurry image

#big radius → less blurry image
````

````python
r_hpf = 30

hpf_mask = np.ones((rows, cols), np.uint8)

y, x = np.ogrid[:rows, :cols]
distance = np.sqrt((y - crow)**2 + (x - ccol)**2)
hpf_mask[distance <= r_hpf] = 0

fshift_hpf = fshift * hpf_mask

f_ishift_hpf = np.fft.ifftshift(fshift_hpf)
img_hpf = np.fft.ifft2(f_ishift_hpf)
img_hpf = np.abs(img_hpf)

plt.figure(figsize=(5, 5))
plt.imshow(img_hpf, cmap='gray')
plt.title(f"hpf, radius = {r_hpf}")
plt.axis('off')
plt.show()
````

````python
#a high pass filler removes smooth areas and keep sharp detaiils, so the photo  looks cleerer but can get a bit noisy. the radius controls how much smooth content is removed.
````

````python
side = 50
half = side // 2

mask = np.zeros((rows, cols), np.uint8)
mask[crow-half : crow+half, ccol-half : ccol+half] = 1

img = np.abs(np.fft.ifft2(np.fft.ifftshift(fshift * mask)))

plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()
# this makes a small square in the center that keeps only low frequency which blurs the photo.
````

````python
import numpy as np, cv2, matplotlib.pyplot as plt

I1 = cv2.imread("your_image.jpg",0)
I2 = cv2.imread("your_image2.jpg",0)
if I1 is None or I2 is None: raise ValueError("Images not found.")

# same size
I2 = cv2.resize(I2,(I1.shape[1],I1.shape[0]))


````

````python
#we load two images and make sure they’re the same size so stuff doesnt break.
````

````python
# FFTs
F1, F2 = np.fft.fftshift(np.fft.fft2(I1)), np.fft.fftshift(np.fft.fft2(I2))
mag1, ang1 = np.abs(F1), np.angle(F1)
mag2, ang2 = np.abs(F2), np.angle(F2)
````

````python
#we take the fft of both and split them into magnitude (strenght) and phase (shape).
````

````python
# swapped spectra
S1 = mag1 * np.exp(1j*ang2)
S2 = mag2 * np.exp(1j*ang1)
````

````python
#then we swap them: image1’s magnitude + image2’s phase, and the other way around
````

````python

# inverse
rec1 = np.abs(np.fft.ifft2(np.fft.ifftshift(S1)))
rec2 = np.abs(np.fft.ifft2(np.fft.ifftshift(S2)))
````

````python
#we do inverse fft to turn the mixed data back into normal pics
````

````python
# normalize for display
rec1 = (255*(rec1/rec1.max())).astype(np.uint8)
rec2 = (255*(rec2/rec2.max())).astype(np.uint8)
````
