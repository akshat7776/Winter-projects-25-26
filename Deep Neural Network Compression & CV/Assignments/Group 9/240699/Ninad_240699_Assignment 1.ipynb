### C3. Create a square LPF mask
Keep side of square as 50. The square will be at the center of the filter.

````python
rows, cols = gray_manual.shape
crow, ccol = rows//2, cols//2
#LPF
radii =[80,30,10]
plt.figure(figsize=(15,5))
for i, radius in enumerate(radii):
  #creating the mask(white circle in the center)
  y,x = np.ogrid[:rows, :cols]
  mask_area = (x- ccol)**2 +(y-crow)**2 <= radius**2
  mask = np.zeros((rows, cols), np.uint8)
  mask[mask_area]=1
  #applying mask to fft
  fshift_masked = fshift*mask
  #recontruction of to pixels(Inverse fft)
  f_ishift = np.fft.ifftshift(fshift_masked)#reshift the center to back to corners
  img_back = np.abs(np.fft.ifft2(f_ishift)) #magnitude of inverse fft

  plt.subplot(1,3, i+1)
  plt.imshow(img_back, cmap='gray')
  plt.title(f'LPF radius : {radius}')
  plt.axis('off')

plt.suptitle("Low pass filters")
plt.show()
````

````python
#c2
radius_hpf =30
#creating mask
mask_hpf = np.ones((rows,cols), np.uint8)
#creating circular area
y,x = np.ogrid[:rows, :cols]
center_circle = (x-ccol)**2 + (y-crow)**2 <=radius_hpf**2
mask_hpf[center_circle] =0

#applying mask
fshift_hpf = fshift* mask_hpf
#inverse fft
f_ishift_hpf = np.fft.ifftshift(fshift_hpf)
img_back_hpf = np.abs(np.fft.ifft2(f_ishift_hpf))

plt.plot(12,6)
plt.imshow(img_back_hpf, cmap ='gray')
plt.axis('off')
plt.show()
````

````python
sidelength = 50
half = sidelength//2
#creating mask
mask_square = np.zeros((rows, cols), np.uint8)

#square region
r_start = crow - half
r_end = crow + half
c_start = ccol - half
c_end = ccol + half
#setting center to low pass
mask_square[r_start:r_end, c_start:c_end]=1
#apply mask
fshift_sq = fshift*mask_square
f_ishift_sq = np.fft.ifftshift(fshift_sq)
img_back_sq = np.abs(np.fft.ifft2(f_ishift_sq))

plt.plot(12,6)
plt.imshow(img_back_sq, cmap='gray')
plt.axis('off')
plt.show
````
