````python
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np
import cv2

# PASTE BASE64 STRING HERE
base64_image = """

"""  # long base64 string

img_data = base64.b64decode(base64_image)
img = Image.open(io.BytesIO(img_data))

plt.imshow(img)
plt.axis("off")
````

````python
def convolve(image, kernel):
    #flipping the kernel(mathematical convolution requires flipping)
    kernel = np.flipud(np.fliplr(kernel))
    #dimensions
    img_h, img_w = image.shape
    k_h,k_w = kernel.shape
    #padding size
    pad_h = k_h//2
    pad_w =k_w//2
    #padding image using reflect mode
    padded_img = np.pad(image, ((pad_h,pad_h), (pad_w,pad_w)), mode='reflect')
    output =np.zeros_like(image)
    #sliding window
    for i in range(img_h):
        for j in range(img_w):
            #selecting region of image as same size of kernel
            region = padded_img[i:i+k_h, j:j+k_w]
            #applying kernel
            output[i,j]= np.sum(region*kernel)

    return output
````

````python
def average_blur_gray(image, ksize):
    kernel_values = np.ones((ksize,ksize))
    normalization_factor = ksize*ksize
    box_kernel = kernel_values/ normalization_factor

    return convolve(image, box_kernel)

def average_blur_rgb(image, ksize):
    #spliting channels
    r,g,b = cv2.split(image)

    #applying blurring to each channel
    r_blurred = average_blur_gray(r, ksize)
    g_blurred = average_blur_gray(g,ksize)
    b_blurred = average_blur_gray(b, ksize)

    #merging back
    return cv2.merge([r_blurred, g_blurred, b_blurred])
````

````python
def gaussian_kernel(ksize, sigma):
    kernel = np.zeros((ksize,ksize))
    center = ksize//2

    for x in range(ksize):
        for y in range(ksize):
            #to make distribution symmetric
            x_dist =x- center
            y_dist =y- center

            #applying gaussian
            expo =-(x_dist**2+y_dist**2)/(2*sigma**2)
            kernel[x,y]= (1/(2*np.pi * sigma**2))* np.exp(expo)

    return kernel/ np.sum(kernel)

````

````python
np_data = np.frombuffer(img_data, np.uint8)
img_bgr = cv2.imdecode(np_data, cv2.IMREAD_COLOR)
gray_image = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
rgb_image = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)

#GENERATING KERNEL
g_kernel = gaussian_kernel(ksize=51, sigma=8)
#convolving with gray
gray_blur = convolve(gray_image, g_kernel)
#convolving with rgb image
r,g,b = cv2.split(rgb_image)
rgb_blur = cv2.merge([convolve(r, g_kernel), convolve(g, g_kernel), convolve(b,g_kernel)])

plt.figure(figsize=(20,10))
plt.subplot(1,2,1)
plt.imshow(gray_image, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(gray_blur, cmap='gray')
plt.show()
````

#### Task - Sobel Edge Detection (NumPy Only)
Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

Compute:
1. Gx
2. Gy
3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python
def sobel_edge_detection(image):
    kx = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
    ky = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])

    gx = convolve(image, kx)
    gy = convolve(image, ky)

    #computing gradient magnitude
    magnitude = np.sqrt(gx**2 + gy**2)
    #normalise to 0-255
    min_val = magnitude.min()
    max_val = magnitude.max()

    if max_val- min_val ==0:
        norm_magnitude = np.zeros_like(magnitude)
    else:
        norm_magnitude = (magnitude - min_val) / (max_val - min_val)*255

    edge_map = norm_magnitude.astype(np.uint8)
    return gx, gy, edge_map

gx,gy,edges = sobel_edge_detection(gray_image)

plt.figure(figsize=(30,15))
plt.subplot(3,1,1)
plt.imshow(gx, cmap='gray')
plt.title('Sobel X')
plt.subplot(3,1,2)
plt.imshow(gy, cmap='gray')
plt.title('Sobel Y')
plt.subplot(3,1,3)
plt.imshow(edges, cmap='gray')
plt.title('magnitude')
plt.show()

````

````python
def get_magnitude_spectrum(image):
    f = np.fft.fft2(image)
    fshift = np.fft.fftshift(f)
    magnitude_spectrum = 20*np.log(np.abs(fshift)+1)
    return magnitude_spectrum
````
