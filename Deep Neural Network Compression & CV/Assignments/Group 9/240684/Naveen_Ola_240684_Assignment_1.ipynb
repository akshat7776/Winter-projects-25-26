````python
# Your PART C codes follow here. You can create more cells under this. Write clean and commented code.
rows,cols = img_grayscale.shape
crow,ccol=rows//2,cols//2
radius_1=10
radius_2=30
radius_3=80
mask_lpf=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf,(ccol,crow),radius_1,1,-1)
f_lpf = fshift * mask_lpf
ishift=np.fft.ifftshift(f_lpf)
img_lpf=np.fft.ifft2(ishift)
img_lpf=np.abs(img_lpf)
plt.figure(figsize=(6,6))
plt.imshow(img_lpf,cmap='gray')
plt.title("lpf_image")
plt.axis("off")
plt.show()

mask_lpf=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf,(ccol,crow),radius_2,1,-1)
f_lpf = fshift * mask_lpf
ishift=np.fft.ifftshift(f_lpf)
img_lpf=np.fft.ifft2(ishift)
img_lpf=np.abs(img_lpf)
plt.figure(figsize=(6,6))
plt.imshow(img_lpf,cmap='gray')
plt.title("lpf_image")
plt.axis("off")
plt.show()

mask_lpf=np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf,(ccol,crow),radius_3,1,-1)
f_lpf = fshift * mask_lpf
ishift=np.fft.ifftshift(f_lpf)
img_lpf=np.fft.ifft2(ishift)
img_lpf=np.abs(img_lpf)
plt.figure(figsize=(6,6))
plt.imshow(img_lpf,cmap='gray')
plt.title("lpf_image")
plt.axis("off")
plt.show()
# less radius of low pass filter will lead to pass only very low frequency to pass which will lead to remove majority of edges leading to more blurness.
# so as radius increases some high frequencies enter and edges become visible lead to less blurrness.
````

````python
mask_hpf=1-mask_lpf
f_hpf=fshift * mask_hpf
ishift1=np.fft.ifftshift(f_hpf)
img_hpf=np.fft.ifft2(ishift1)
img_hpf=np.abs(img_hpf)
plt.figure(figsize=(6,6))
plt.imshow(img_hpf,cmap='gray')
plt.title("hpf_image")
plt.axis("off")
plt.show()
````

````python
s_mask=np.zeros((rows,cols),np.uint8)
cv2.rectangle(s_mask,(ccol-25,crow-25),(ccol+25,crow+25),1,-1)
s_lpf = s_mask * fshift
ishift2=np.fft.ifftshift(s_lpf)
img_lpf1=np.fft.ifft2(ishift2)
img_lpf1=np.abs(img_lpf1)
plt.figure(figsize=(6,6))
plt.imshow(img_lpf1,cmap='gray')
plt.title("lpf1_image")
plt.axis("off")
plt.show()
````

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
phase_only = fshift / np.abs(fshift)
img_phase= np.fft.ifftshift(phase_only)
img_phase= np.fft.ifft2(img_phase)
img_phase=np.abs(img_phase)
plt.figure(figsize=(6,6))
plt.imshow(img_phase,cmap='gray')
plt.title("phase_only_image")
plt.axis("off")
plt.show()
````

````python
phase2_img = np.angle(fshift)
img_phase1= np.fft.ifftshift(phase2_img)
img_phase1= np.fft.ifft2(img_phase1)
img_phase1=np.abs(img_phase1)
plt.figure(figsize=(6,6))
plt.imshow(img_phase1,cmap='gray')
plt.title("phase2_image (np.angle(fshift))")
plt.axis("off")
plt.show()
# the difference between np.angle(fshift) and phase_only = fshift/abs(fshift) is that np.angle will give us the value of phase(real) while in phase_only it will give complaex number
# which can also represent direction so image using this will be better than np.agle
````

````python
mag_only = np.abs(fshift)
img_mag= np.fft.ifftshift(mag_only)
img_mag= np.fft.ifft2(img_mag)
img_mag=np.abs(img_mag)
plt.figure(figsize=(6,6))
plt.imshow(img_mag,cmap='gray')
plt.title("magnitude_only_image")
plt.axis("off")
plt.show()
# frequency ranges are really large fom some hertz to mega hertz so log reduces the scale so they will become easy to study.
# Looking different :
# The spectrum visualization used a logarithm to make faint details visible,
# whereas this reconstruction uses raw values where the massive central peak
# makes everything else scale down to black
````
