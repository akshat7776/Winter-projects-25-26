````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def convolve(img, kernel):

    if len(img.shape) == 3:
        h, w, c = img.shape

        output = np.zeros_like(img, dtype=np.float32)

        for i in range(c):
            output[:, :, i] = convolve_2d_single(img[:, :, i], kernel)
        return output
    else:
        return convolve_2d_single(img, kernel)

def convolve_2d_single(img, kernel):

    h, w = img.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros((h, w))

    for i in range(h):
        for j in range(w):
            region = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize ** 2)
    return convolve(img, kernel)

def gaussian_kernel(ksize, sigma):
    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2. * sigma**2))
    return kernel / np.sum(kernel)

def apply_gaussian_blur(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)


image_path = 'Fiddle-Leaf.jpg'
original_img = cv2.imread(image_path)
original_img = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
gray_img = cv2.cvtColor(original_img, cv2.COLOR_RGB2GRAY)
avg_blur_result_gray = average_blur_gray(gray_img, ksize=5)
avg_blur_result_rgb = average_blur_gray(original_img, ksize=5)

ksize_g = 7
sigma_g = 1.5
g_kernel = gaussian_kernel(ksize_g, sigma_g)

gauss_blur_result_gray = apply_gaussian_blur(gray_img, ksize_g, sigma_g)
gauss_blur_result_rgb = apply_gaussian_blur(original_img, ksize_g, sigma_g)

plt.figure(figsize=(12, 10))

plt.subplot(3, 2, 1)
plt.title("Original RGB")
plt.imshow(original_img)
plt.axis('off')

plt.subplot(3, 2, 2)
plt.title("Average Blur (Gray)")
plt.imshow(avg_blur_result_gray, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 3)
plt.title("Average Blur (RGB)")
plt.imshow(avg_blur_result_rgb.astype(int))
plt.axis('off')

plt.subplot(3, 2, 4)
plt.title(f"Gaussian Blur (Gray)\nk={ksize_g}, sigma={sigma_g}")
plt.imshow(gauss_blur_result_gray, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 5)
plt.title(f"Gaussian Blur (RGB)")
plt.imshow(gauss_blur_result_rgb.astype(int))
plt.axis('off')

plt.subplot(3, 2, 6)
plt.title("Generated Gaussian Kernel Heatmap")
plt.imshow(g_kernel, cmap='viridis')
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def convolve_2d(image, kernel):
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2
    padded_img = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(image, dtype=np.float32)
    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

def sobel_edge_detection(img):
    sobel_x = np.array([[-1, 0, 1],
                        [-2, 0, 2],
                        [-1, 0, 1]])

    sobel_y = np.array([[-1, -2, -1],
                        [ 0,  0,  0],
                        [ 1,  2,  1]])

    print("Computing Gradient X")
    gx = convolve_2d(img, sobel_x)

    print("Computing Gradient Y")
    gy = convolve_2d(img, sobel_y)
    magnitude = np.sqrt(gx**2 + gy**2)
    if np.max(magnitude) != 0:
        norm_magnitude = (magnitude / np.max(magnitude)) * 255
    else:
        norm_magnitude = magnitude

    edge_map = norm_magnitude.astype(np.uint8)

    return gx, gy, edge_map

gx, gy, final_edges = sobel_edge_detection(gray_img)
_, thresholded_edges = cv2.threshold(final_edges, 50, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title("Original Grayscale")
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title("Sobel X (Vertical Edges)")
plt.imshow(np.abs(gx), cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title("Sobel Y (Horizontal Edges)")
plt.imshow(np.abs(gy), cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.title("Final Edge Magnitude (Normalized)")
plt.imshow(final_edges, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure()
plt.title("Optional: Thresholded Edges (Binary)")
plt.imshow(thresholded_edges, cmap='gray')
plt.axis('off')
plt.show()
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def convolve_2d(image, kernel):
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(image, dtype=np.float32)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

def get_gaussian_kernel(ksize, sigma):
    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2. * sigma**2))
    return kernel / np.sum(kernel)

def normalize_image(img):
    return np.clip(img, 0, 255).astype(np.uint8)

def laplacian_sharpening(img, alpha=1.0):

    lap_kernel = np.array([[ 0,  1,  0],
                           [ 1, -4,  1],
                           [ 0,  1,  0]])

    lap = convolve_2d(img, lap_kernel)
    sharp = img - (lap * alpha)
    return normalize_image(sharp), lap

def unsharp_masking(img, alpha=1.5):

    blur_kernel = get_gaussian_kernel(ksize=5, sigma=1.0)
    blurred = convolve_2d(img, blur_kernel)
    mask = img - blurred
    sharp = img + (alpha * mask)
    return normalize_image(sharp), normalize_image(mask), normalize_image(blurred)

gray_float = gray_img.astype(np.float32)

print("Running Laplacian Sharpening")
sharp_lap, lap_edges = laplacian_sharpening(gray_float, alpha=1.0)

lap_kernel_cv2 = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
cv2_lap = cv2.filter2D(gray_float, -1, lap_kernel_cv2)
cv2_sharp = normalize_image(gray_float - cv2_lap)

difference = np.abs(sharp_lap.astype(float) - cv2_sharp.astype(float))
is_identical = np.all(difference < 1e-5)
print(f"Match between NumPy and CV2 implementation: {is_identical}")
print("Running Unsharp Masking")

sharp_usm, mask_usm, blurred_img = unsharp_masking(gray_float, alpha=2.0)

blur_effect_usm, _, _ = unsharp_masking(gray_float, alpha=-0.8)

plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.title("Original Grayscale")
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title("Laplacian Edges")
plt.imshow(lap_edges, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title("Laplacian Sharpened (alpha=1)")
plt.imshow(sharp_lap, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title("Blurred Image (Gaussian)")
plt.imshow(blurred_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title("Unsharp Mask (Detail)")
plt.imshow(mask_usm, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title("Unsharp Masking Output (alpha=2)")
plt.imshow(sharp_usm, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure()
plt.title("Negative Alpha (Unsharp Masking) to Blurring")
plt.imshow(blur_effect_usm, cmap='gray')
plt.axis('off')
plt.show()

````

Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def get_magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)

    return magnitude_spectrum

def generate_images(img_gray):
    blurred = cv2.GaussianBlur(img_gray, (9, 9), 2.0)
    alpha = 2.0
    mask = img_gray.astype(float) - blurred.astype(float)
    sharpened = img_gray.astype(float) + (alpha * mask)
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

    return blurred, sharpened

blurred_img, sharpened_img = generate_images(gray_img)

spec_original = get_magnitude_spectrum(gray_img)
spec_blur = get_magnitude_spectrum(blurred_img)
spec_sharp = get_magnitude_spectrum(sharpened_img)
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.title("Original Image")
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title("Gaussian Blurred (LPF)")
plt.imshow(blurred_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title("Unsharp Masking (Sharpened)")
plt.imshow(sharpened_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title("Spectrum: Original")
plt.imshow(spec_original, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title("Spectrum: Blurred (LPF Effect)")
plt.imshow(spec_blur, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title("Spectrum: Sharpened (HPF Boost)")
plt.imshow(spec_sharp, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show


   #OBSERVATIONS & EXPLANATION:
   #
   #1. Interpreting the Magnitude Spectrum:
   #   - The CENTER of the spectrum represents Low Frequencies (smooth areas, general brightness).
   #   - The OUTER regions represent High Frequencies (edges, noise, rapid changes).
   #   - Bright areas mean those frequencies are present; dark areas mean they are absent.

   #2. Gaussian Blur (Low Pass Filter - LPF):
   #   - Observation: In the 'Spectrum: Blurred' plot, the center remains bright, but the
   #     outer corners become significantly darker compared to the original spectrum.
   #   - Why: Gaussian blur smoothes out fine details (edges). Since edges are high-frequency
   #     components, the blur kernel attenuates (filters out) these high frequencies.
   #     This confirms that convolution with a Gaussian kernel acts as a Low Pass Filter.

   #3. Unsharp Masking (High Pass Boosting):
   #   - Observation: In the 'Spectrum: Sharpened', the outer regions are brighter
   #     compared to the blurred spectrum, and potentially brighter than the original.
   #   - Why: Sharpening works by adding the 'difference' (edges) back into the image.
   #     This amplifies the high-frequency components. By increasing 'alpha', we are
   #     literally boosting the magnitude of these high frequencies in the frequency domain.
````

````python
import numpy as np
import matplotlib.pyplot as plt

def convolve_2d(image, kernel):
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(image, dtype=np.float32)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

r_channel = img_part5[:, :, 0].astype(float)
g_channel = img_part5[:, :, 1].astype(float)

diff_map = r_channel - g_channel

kernel = np.array([[-1, -1, -1],
                   [-1,  8, -1],
                   [-1, -1, -1]])

print("Convolving kernel with (Red - Green) map")
edge_response = convolve_2d(diff_map, kernel)

edge_magnitude = np.abs(edge_response)

norm_edges = (edge_magnitude / np.max(edge_magnitude)) * 255
norm_edges = norm_edges.astype(np.uint8)
_, clean_outline = cv2.threshold(norm_edges, 40, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.title("Feature Map (Red - Green)")
plt.imshow(diff_map, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Raw Edge Response")
plt.imshow(norm_edges, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Final Clean Outline")
plt.imshow(clean_outline, cmap='gray')
plt.axis('off')

plt.show()
````
