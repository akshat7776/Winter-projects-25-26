````python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Task 1: Convolution

def convolve(image, kernel):

    img_h, img_w = image.shape
    k_h, k_w = kernel.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    output = np.zeros_like(image, dtype=np.float32)

    # We iterate over every single pixel of the original image dimensions.
    # For each pixel, we take a 3x3(size of kernel) region from the padded image
    for i in range(img_h):
        for j in range(img_w):

            roi = padded_img[i : i + k_h, j : j + k_w]

            output[i, j] = np.sum(roi * kernel)

    return output


# Task 2: Average Blur

def average_blur_gray(img, ksize):

    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)

    blurred = convolve(img, kernel)

    return np.clip(blurred, 0, 255).astype(np.uint8)

def apply_to_rgb(img_rgb, func, **kwargs):

    r = func(img_rgb[:,:,0], **kwargs)
    g = func(img_rgb[:,:,1], **kwargs)
    b = func(img_rgb[:,:,2], **kwargs)
    return cv2.merge([r, g, b])

# Task 3: Gaussian Blur

def gaussian_kernel(ksize, sigma):

    # Ensuring odd kernel size
    if ksize % 2 == 0: ksize += 1
    center = ksize // 2

    kernel = np.zeros((ksize, ksize), dtype=np.float32)

    for x in range(ksize):
        for y in range(ksize):
            # Distance from center
            dx = x - center
            dy = y - center

            exponent = -(dx**2 + dy**2) / (2 * sigma**2)
            kernel[x, y] = (1 / (2 * np.pi * sigma**2)) * np.exp(exponent)

    return kernel / np.sum(kernel)

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    blurred = convolve(img, kernel)
    return np.clip(blurred, 0, 255).astype(np.uint8)


if len(img_rgb.shape) == 3:
    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
else:
    gray = img_rgb

# Running average blur
avg_img = apply_to_rgb(img_rgb, average_blur_gray, ksize=7)

# Running gaussian blur
def gaussian_wrapper(img_channel, ksize, sigma):
    return gaussian_blur_gray(img_channel, ksize, sigma)

gauss_img = apply_to_rgb(img_rgb, gaussian_wrapper, ksize=7, sigma=1.5)

# Displaying the results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(avg_img)
plt.title("Average Blur")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(gauss_img)
plt.title("Gaussian Blur")
plt.axis('off')

plt.show()
````

````python
def sobel_edge_detection(img_gray):

    # Gx
    Gx_kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)

    # Gy
    Gy_kernel = np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ], dtype=np.float32)

    # Convolving
    gx = convolve(img_gray, Gx_kernel)
    gy = convolve(img_gray, Gy_kernel)

    # Computing magnitude
    magnitude = np.sqrt(gx**2 + gy**2)

    # Normalizing

    # Gx
    gx_abs = np.abs(gx)
    gx_norm = (gx_abs / (np.max(gx_abs) + 1e-6)) * 255

    # Gy
    gy_abs = np.abs(gy)
    gy_norm = (gy_abs / (np.max(gy_abs) + 1e-6)) * 255

    # Magnitude
    mag_norm = (magnitude / (np.max(magnitude) + 1e-6)) * 255

    return gx_norm.astype(np.uint8), gy_norm.astype(np.uint8), mag_norm.astype(np.uint8)

# Grayscale
if len(img_rgb.shape) == 3:
    gray_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
else:
    gray_img = img_rgb

# Running the edge detection function
gx_img, gy_img, mag_img = sobel_edge_detection(gray_img)

# Thresholding
threshold = 50
binary_edges = np.where(mag_img > threshold, 255, 0).astype(np.uint8)

# Displaying the results
plt.figure(figsize=(12, 12))


plt.subplot(2, 2, 3)
plt.imshow(mag_img, cmap='gray')
plt.title("Gradient Magnitude")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(binary_edges, cmap='gray')
plt.title(f"Thresholded Edges")
plt.axis('off')

plt.show()
````

````python
# Task 1: Laplacian Sharpening

def laplacian_sharpening(img_gray, alpha=1.0):

    lap_kernel = np.array([
        [ 0, -1,  0],
        [-1,  4, -1],
        [ 0, -1,  0]
    ], dtype=np.float32)

    lap = convolve(img_gray, lap_kernel)

    img_float = img_gray.astype(np.float32)
    sharpened = img_float - (alpha * lap)

    return np.clip(sharpened, 0, 255).astype(np.uint8)

# Task 2: Unsharp Masking

def unsharp_masking(img_gray, alpha=1.5, sigma=1.0):

    img_float = img_gray.astype(np.float32)

    blurred = gaussian_blur_gray(img_float, ksize=5, sigma=sigma)

    mask = img_float - blurred

    sharpened = img_float + (alpha * mask)

    return np.clip(sharpened, 0, 255).astype(np.uint8)

def verify(img_gray):

    manual_sharp = laplacian_sharpening(img_gray, alpha=1.0)
    lap_cv = cv2.Laplacian(img_gray, cv2.CV_32F)
    sharp_cv = np.clip(img_gray.astype(np.float32) + lap_cv, 0, 255).astype(np.uint8)

    # Comparing the results
    difference = np.abs(manual_sharp.astype(int) - sharp_cv.astype(int))
    if np.sum(difference) == 0:
        print("Verified")
    else:
        print(f"Verification failed: Total difference is {np.sum(difference)}")

# Grayscale
if len(img_rgb.shape) == 3:
    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
else:
    gray = img_rgb

verify(gray)

# Sharpening
lap_sharp_img = laplacian_sharpening(gray, alpha=0.8)
unsharp_img = unsharp_masking(gray, alpha=2.0)
blur_test_img = unsharp_masking(gray, alpha=-0.5)

# Displaying the results
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(lap_sharp_img, cmap='gray')
plt.title("Laplacian Sharpening (alpha=0.8)")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(unsharp_img, cmap='gray')
plt.title("Unsharp Masking (alpha=2.0))")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(blur_test_img, cmap='gray')
plt.title("Unsharp Masking (alpha=-0.5)")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
def get_magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)
    return magnitude_spectrum

img_orig = gray

# Guassian Blur
img_blur = gaussian_blur_gray(img_orig.astype(np.float32), ksize=9, sigma=2.0)

# Unsharp Mask
img_sharp = unsharp_masking(img_orig, alpha=3.0, sigma=1.0)

# Computing the magnitude spectrums
spec_orig = get_magnitude_spectrum(img_orig)
spec_blur = get_magnitude_spectrum(img_blur)
spec_sharp = get_magnitude_spectrum(img_sharp)

# Displaying the results
plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)
plt.imshow(img_orig, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(img_blur, cmap='gray')
plt.title("Blurred Image (LPF)")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(img_sharp, cmap='gray')
plt.title("Sharpened Image (HPF Boost)")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(spec_orig, cmap='gray')
plt.title("Spectrum: Original")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(spec_blur, cmap='gray')
plt.title("Spectrum: Blurred\n(Outer corners are DARKER)")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(spec_sharp, cmap='gray')
plt.title("Spectrum: Sharpened\n(Outer corners are BRIGHTER)")
plt.axis('off')

plt.tight_layout()
plt.show()

# Observations:
# 1. Blurred Spectrum: The center remains bright, but the outer corners
#                      become significantly darker compared to the original.
# 2. Sharpened Spectrum: The outer corners are brighter
#                        compared to the original spectrum.
````
