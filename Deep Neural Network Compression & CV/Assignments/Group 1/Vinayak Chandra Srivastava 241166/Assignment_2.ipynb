Write a function <code>rgb_to_hsv</code> that takes a 0-255 ( not normalised ) RGB image and returns an HSV image ( also not normalised ).

````python
def rgb_to_hsv(img_rgb):
  R = img_rgb[:, :, 0]
  G = img_rgb[:, :, 1]
  B = img_rgb[:, :, 2]

  R1 = R/255.0
  G1 = G/255.0
  B1 = B/255.0

  Cmax = np.maximum(np.maximum(R1, G1), B1)
  Cmin = np.minimum(np.minimum(R1, G1), B1)
  delta = Cmax - Cmin

  H1 = np.zeros_like(R1)
  S1 = np.zeros_like(R1)
  V1 = Cmax

  H1[delta == 0] = 0

  mask_R = (Cmax == R1) & (delta != 0)
  H1[mask_R] = 60 * np.mod(((G1[mask_R] - B1[mask_R]) / delta[mask_R]), 6)

  mask_G = (Cmax == G1) & (delta != 0)
  H1[mask_G] = 60 * (((B1[mask_G] - R1[mask_G]) / delta[mask_G]) + 2)

  mask_B = (Cmax == B1) & (delta != 0)
  H1[mask_B] = 60 * (((R1[mask_B] - G1[mask_B]) / delta[mask_B]) + 4)

  S1[Cmax == 0] = 0
  S1[Cmax != 0] = delta[Cmax != 0] / Cmax[Cmax != 0]

  H = (H1 /2).astype(np.uint8)
  S = (S1 * 255).astype(np.uint8)
  V = (V1 * 255).astype(np.uint8)
  hsv_image = np.stack((H, S, V), axis=-1)
  return hsv_image
````

Filter 1: Sleepy 8am lectures

````python
import cv2
import numpy as np

def sleepy_8am_lectures_filter(img):
    img = img.astype(np.float32) / 255.0

    # Reduce saturation (because lecture = no color in life)
    hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
    hsv[...,1] *= 0.55        # -45% saturation
    hsv[...,1] = np.clip(hsv[...,1], 0, 1)
    desat = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

    # Lower brightness (dim mood)
    dark = desat * 0.92       # -8% brightness

    # Low contrast (brain is foggy)
    contrast = 0.88
    low_con = (dark - 0.5) * contrast + 0.5
    low_con = np.clip(low_con, 0, 1)

    # Soft blur (sleepy eyes)
    blur = cv2.GaussianBlur(low_con, (0,0), sigmaX=2)

    # Cool tint (AC is too cold in LHC)
    tint = np.array([0.94, 0.96, 1.08])  # slightly bluish + cold
    final = blur * tint
    final = np.clip(final, 0, 1)

    return (final * 255).astype(np.uint8)
````

I chose saturation reduction, blur, low contrast, and a cold tint to simulate the tired, washed-out vibe of 8 am lectures. These modifications capture the low-energy ambience of lecture halls, where colors feel dull and vision feels slightly foggy. This filter fits classroom photos and dim indoor images perfectly.

Filter 2: Golden Glow

````python
def golden_glow_filter(img):
    img = img.astype(np.float32) / 255.0

    # Brightness lift
    bright = img + 0.08
    bright = np.clip(bright, 0, 1)

    # Warm tint (golden look)
    warm = bright * np.array([1.10, 1.03, 0.90], dtype=np.float32)
    warm = np.clip(warm, 0, 1).astype(np.float32)

    # Vibrance boost (selective saturation)
    hsv = cv2.cvtColor(warm, cv2.COLOR_RGB2HSV)
    mean_sat = hsv[...,1].mean()
    hsv[...,1] = hsv[...,1] + (hsv[...,1] < mean_sat) * 0.25
    hsv[...,1] = np.clip(hsv[...,1], 0, 1)
    vibrance = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

    # Highlight glow shiner
    blur = cv2.GaussianBlur(vibrance, (0,0), 4)
    glow = vibrance * 0.85 + blur * 0.15

    return (glow * 255).astype(np.uint8)
````

I used a brightness lift, warm tint, vibrance boost, and a soft golden glow to create a radiant, sun-kissed aesthetic. These modifications enhance warm tones, highlights, and overall vibrancy, giving images a lively and inviting feel. This filter works best for festivals, outdoor shots, and any scene that benefits from a warm, glowing atmosphere.

#### White Balance Function
Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image. Use Numpy only to keep track of the brightest pixel and then scale the entire image with the factor you find. Display the original and output image.

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def white_patch_balance(img_rgb):
    img = img_rgb.astype(np.float32)

    r_max = np.max(img[...,0])
    g_max = np.max(img[...,1])
    b_max = np.max(img[...,2])

    r_scale = 255.0 / r_max
    g_scale = 255.0 / g_max
    b_scale = 255.0 / b_max

    img_balanced = np.zeros_like(img)
    img_balanced[...,0] = img[...,0] * r_scale
    img_balanced[...,1] = img[...,1] * g_scale
    img_balanced[...,2] = img[...,2] * b_scale

    wb_img = np.clip(img_balanced, 0, 255).astype(np.uint8)

    return wb_img
````

````python
img3 = cv2.imread("Assignment Image 4.jpg")
img3 = cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)
img3 = img3.astype(np.uint8)
white_balanced_img = white_patch_balance(img3)
````

````python
plt.figure(figsize=(18,6))
plt.subplot(1,3,1)
plt.imshow(img3)
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(white_balanced_img)
plt.title("White Patch Balanced Image")
plt.axis('off')
````

It takes the img_rgb ( 0-255 RGB image ) and value ( 0-1 decimal number ).

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def white_balance_slider(img_rgb, value):
    value = np.clip(value, 0, 1)

    img = img_rgb.astype(np.float32)

    r_avg = np.mean(img[..., 0])
    g_avg = np.mean(img[..., 1])
    b_avg = np.mean(img[..., 2])

    k = (r_avg + g_avg + b_avg) / 3.0

    neutral = np.zeros_like(img)
    neutral[..., 0] = img[..., 0] * (k / r_avg)
    neutral[..., 1] = img[..., 1] * (k / g_avg)
    neutral[..., 2] = img[..., 2] * (k / b_avg)

    # Warm: boost reds, reduce blues slightly
    warm = neutral.copy()
    warm[..., 0] *= 1.15
    warm[..., 2] *= 0.90

    # Cold: boost blues, reduce reds slightly
    cold = neutral.copy()
    cold[..., 2] *= 1.15
    cold[..., 0] *= 0.90

    if value < 0.5:
        # 0: cold, 0.5: original
        t = value / 0.5
        out = cold * (1 - t) + neutral * t
    else:
        # 0.5: original, 1: warm
        t = (value - 0.5) / 0.5
        out = neutral * (1 - t) + warm * t

    final_image = np.clip(out, 0, 255).astype(np.uint8)

    return final_image
````

````python
cold = white_balance_slider(img3, 0.0)
neutral = white_balance_slider(img3, 0.5)
warm = white_balance_slider(img3, 1.0)
slider_val = white_balance_slider(img3, 0.34)

plt.figure(figsize=(14,10))

plt.subplot(2,2,1)
plt.imshow(cold)
plt.title("Cold (value = 0.0)")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(neutral)
plt.title("Neutral (value = 0.5)")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(warm)
plt.title("Warm (value = 1.0)")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(slider_val)
plt.title("Slider (value = 0.34)")
plt.axis("off")

plt.tight_layout()
plt.show()
````
