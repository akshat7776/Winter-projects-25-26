````python
#lpf for radius 10
rows, cols = gray_float.shape
crow,ccol = rows//2 , cols//2
mask_lpf_10 = np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_10,(crow,ccol),10,1,-1)
f_shift_lpf_10=fshift*mask_lpf_10
f_ishift_lpf_10=np.fft.ifftshift(f_shift_lpf_10)
img_lpf_10=np.fft.ifft2(f_ishift_lpf_10)
img_lpf_10=np.abs(img_lpf_10)
plt.imshow(img_lpf_10)
plt.axis("off")
````

````python
#lpf for radius 30
mask_lpf_30 = np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_30,(crow,ccol),30,1,-1)
f_shift_lpf_30=fshift*mask_lpf_30
f_ishift_lpf_30=np.fft.ifftshift(f_shift_lpf_30)
img_lpf_30=np.fft.ifft2(f_ishift_lpf_30)
img_lpf_30=np.abs(img_lpf_30)
plt.imshow(img_lpf_30)
plt.axis("off")
````

````python
#lpf for radius 80
mask_lpf_80 = np.zeros((rows,cols),np.uint8)
cv2.circle(mask_lpf_80,(crow,ccol),80,1,-1)
f_shift_lpf_80=fshift*mask_lpf_80
f_ishift_lpf_80=np.fft.ifftshift(f_shift_lpf_80)
img_lpf_80=np.fft.ifft2(f_ishift_lpf_80)
img_lpf_80=np.abs(img_lpf_80)
plt.imshow(img_lpf_80)
plt.axis("off")
````

````python
phase_only = fshift / np.abs(fshift)
f_ishift_phase_only=np.fft.ifftshift(phase_only)
img_phase_only=np.fft.ifft2(f_ishift_phase_only)
img_phase_only=np.abs(img_phase_only)
plt.imshow(img_phase_only)
plt.axis("off")
````

````python
phase_using_angle = np.angle(fshift)
f_ishift_phase_using_angle=np.fft.ifftshift(phase_using_angle)
img_phase_using_angle=np.fft.ifft2(f_ishift_phase_using_angle)
img_phase_using_angle=np.abs(img_phase_using_angle)
plt.imshow(img_phase_using_angle)
````

In phase_only, we have phase having the complex exponential. However, in phase_using_angle, we are taking absolute value so phase is only the real part. Hence, the complex part is lost and we don't get as clear an image as for complex exponential.

````python
mag_only = np.abs(fshift)
f_ishift_mag_only=np.fft.ifftshift(mag_only)
img_mag_only=np.fft.ifft2(f_ishift_mag_only)
img_mag_only=np.abs(img_mag_only)
plt.imshow(img_mag_only)
plt.axis("off")
````

````python
mag_using_log = 20*np.log(np.abs(fshift))
plt.imshow(mag_using_log)
plt.axis("off")
````

Values in magnitude spectrum vary in orders of magnitude, so logarithm is required, otherwise we lose the image characteristics.
