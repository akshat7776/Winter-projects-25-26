````python

# Slower convol via loops in O(HWk^2)
def convolve(img,kernel):
  kernel=np.flip(kernel) # for true convolution

  h,w=img.shape
  r,c=kernel.shape
  pr,pc=r//2,c//2
  img_pad=np.pad(img,((pr,pr),(pc,pc)),mode='reflect') # maintains continuity
  out=np.zeros_like(img,dtype=float)
  for i in range(h):
    for j in range(w):
# Pixel (i, j) in img corresponds to pixel (i+pr, j+pc) in img_pad, hence this slicing
      patch=img_pad[i:i+r,j:j+c]
      out[i,j]=np.sum(patch*kernel)
  return out

# faster convolution via fft in O(HWlog(HW))
def convolve_fft(img, kernel):
    H, W = img.shape
    r, c = kernel.shape
    pr, pc = r // 2, c // 2

    # true convolution
    kernel = np.flip(kernel)

    # reflect padding in spatial domain
    img_pad = np.pad(img, ((pr, pr), (pc, pc)), mode='reflect')

    # size for linear convolution
    out_shape = (img_pad.shape[0] + r - 1,
                 img_pad.shape[1] + c - 1)

    # FFTs
    F_img = np.fft.fft2(img_pad, out_shape)
    F_ker = np.fft.fft2(kernel, out_shape)

    # multiply + inverse FFT
    out_full = np.fft.ifft2(F_img * F_ker).real

    # crop back to original image size
    return out_full[pr:pr+H, pc:pc+W]
````

````python
def average_blur_gray(img, ksize):
  kernel=np.ones((ksize,ksize),dtype=float)
  kernel/=ksize*ksize # normalize
  out=convolve_fft(img,kernel)
  return np.clip(out,0,255).astype(np.uint8) # for displaying via open-cv
plt.imshow(average_blur_gray(img_gray,8),cmap='gray')
````

````python
def average_blur_rgb(img, ksize):
  r,g,b=img[:,:,0],img[:,:,1],img[:,:,2]
  r=average_blur_gray(r,ksize)
  g=average_blur_gray(g,ksize)
  b=average_blur_gray(b,ksize)
  return np.stack((r,g,b),axis=2)
plt.imshow(img_rgb)
plt.imshow(average_blur_rgb(img_rgb,8))
````

````python
def gaussian_kernel(ksize, sigma):
  kernel=np.zeros((ksize,ksize))
  r=ksize//2 # center at (0,0) with this radius
  # 1-D coordinate axis
  ax=np.arange(-r,r+1) # [-r,..0..,r]
  xx,yy=np.meshgrid(ax,ax) # now (xx[i,j],yy[i,j]) are coordinates of point (i,j)
  kernel=np.exp(-(xx**2+yy**2)/(2*sigma**2)) # apply formula
  kernel/=kernel.sum() # normalize
  return kernel
````

#### Task - Sobel Edge Detection (NumPy Only)
Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

Compute:
1. Gx
2. Gy
3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python
# uses fixed gx and gy kernels of 3X3
def sobelEdgeDetect(img,T):
  # compute gradients
  kx=np.array([[-1,0,1],
               [-2,0,2],
               [-1,0,1]])
  ky=np.array([[-1,-2,-1],
               [0,0,0],
               [1,2,1]])
  gx=convolve_fft(img,kx)
  gy=convolve_fft(img,ky)
  mag=np.hypot(gx,gy)
  # normalize(don't clip first :))
  mag=mag/mag.max() * 255
  mag=mag.astype(np.uint8)

  # thresholding
  mag[mag>=T]=255
  mag[mag<T]=0

  return mag
plt.imshow(sobelEdgeDetect(img_gray,40),cmap='gray')
````

````python
# Original
f_orig=np.fft.fft2(img_gray.astype(np.float32))
f_orig=np.fft.fftshift(f_orig)
mag_orig=20*np.log(np.abs(f_orig)+1)

# Gaussian Blurred
"""
when sigma is increased, the spectrum becomes concentrated at the center,
indicating stronger blur and more suppression of high frequencies
hence its LPF
"""
gauss_gray=convolve_fft(img_gray,gaussian_kernel(3,1))
f_gauss=np.fft.fft2(gauss_gray.astype(np.float32))
f_gauss=np.fft.fftshift(f_gauss)
mag_gauss=20*np.log(np.abs(f_gauss)+1)

"""
when alpha is increased, the central spot diminishes more,
meaning the low frequency components die out more and hence it leads sharpening of the image
hence its HPF
"""
# Unsharp masking
unsharp_gray=task2(img_gray,1)
f_unsharp=np.fft.fft2(unsharp_gray.astype(np.float32))
f_unsharp=np.fft.fftshift(f_unsharp)
mag_unsharp=20*np.log(np.abs(f_unsharp)+1)
````
