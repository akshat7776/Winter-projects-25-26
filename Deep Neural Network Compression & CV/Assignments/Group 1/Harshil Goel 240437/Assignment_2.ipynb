````python
# Importing necessary libraries

from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2

import io
import base64

# Loading image and displaying it

with open("test.jpg", "rb") as image_file:
    encoded_string = base64.b64encode(image_file.read()).decode()
    print(encoded_string)

img_data = base64.b64decode(encoded_string)
image = Image.open(io.BytesIO(img_data))

plt.imshow(image)
plt.axis('off')  

img_array = np.array(image)
````

````python
# Converting to grayscale 
gray_image = np.zeros((img_array.shape[0], img_array.shape[1]), dtype=np.uint8)

for i in range(img_array.shape[0]):
    for j in range(img_array.shape[1]):
        r, g, b = img_array[i, j][:3]
        gray_value = int(0.299 * r + 0.587 * g + 0.114 * b)
        gray_image[i, j] = gray_value

plt.figure()
plt.imshow(gray_image, cmap='gray')
plt.axis('off')
````

````python
# Writting the required function to compute histogram
def grayscale_histogram(image):
    histogram = np.zeros(256, dtype=int)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            pixel_value = image[i, j]
            histogram[pixel_value] += 1

    return histogram

histogram = grayscale_histogram(gray_image)

# Plotting the histogram

plt.figure()
plt.hist(gray_image.ravel(), bins=256, range=(0, 256), color='darkblue')
plt.title('Grayscale Histogram')
````

````python
# Writting the required function to compute RGB histogram
def rgb_histogram(image):
    histogram = np.zeros((3, 256), dtype=int)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i, j][:3]
            histogram[0, r] += 1
            histogram[1, g] += 1
            histogram[2, b] += 1

    return histogram

rgb_hist = rgb_histogram(img_array)

# Plotting the RGB histogram

plt.figure()
plt.plot(rgb_hist[0], color='red', label='Red Channel')
plt.plot(rgb_hist[1], color='green', label='Green Channel')
plt.plot(rgb_hist[2], color='blue', label='Blue Channel')   
plt.title('RGB Histogram')
````

````python
def rgb_to_hsv (image):

    hsv_image = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

    for i in range (image.shape[0]):
        for j in range (image.shape[1]):
            r,g,b = image[i,j][:3]

            # Normalising the data
            rn = r/255
            gn = g/255
            bn = b/255

            # finding min/max components
            c_max = max(rn,gn,bn)
            c_min = min(rn,gn,bn)

            diff = c_max-c_min

            # Calculating hue
            hue = 0

            if diff == 0 :
                hue = 0
            elif c_max == rn :
                hue = 60 * (((gn-bn)/diff)%6)
            elif c_max == gn :
                hue = 60 * (((bn-rn)/diff)+2)
            elif c_max == bn :
                hue = 60 * (((rn-gn)/diff)+4)

            cv2_hue = hue//2

            # Calculating saturation
            saturation = 0

            if c_max == 0 :
                saturation = 0
            else :
                saturation = (diff/c_max)*255
                saturation = saturation//1

            # Calculating value
            value = c_max * 255
            value = value//1

            # Assigning HSV values to the new image

            hsv_image[i,j,0] = cv2_hue
            hsv_image[i,j,1] = saturation
            hsv_image[i,j,2] = value

    return hsv_image

hsv_image = rgb_to_hsv(img_array)

test_conversion = cv2.cvtColor(hsv_image, cv2.COLOR_HSV2RGB)

plt.figure()

plt.subplot(1,2,1)
plt.imshow(img_array)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(test_conversion)
plt.title("After RGB to HSV and back to RGB")
plt.axis('off')

error = np.mean((img_array - test_conversion) ** 2)
mean = np.mean(img_array)
print("Mean Squared Error between original and converted image:", error)
print ("Mean of original image:", mean)

print("Thus we can see that the conversion is quite accurate with very less mean squared error.")
````

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Helper: clamp and convert to uint8
def to_uint8(img):
    return np.clip(img, 0, 255).astype(np.uint8)

# Creating a filter to make the beaches and sunsets look more vibrant and warm

def sunny_saturday_holidays(rgb):
    img = rgb.astype(np.float32)
    
    # Contrast & brightness adjustment (alpha, beta)
    alpha = 1.15   
    beta = 8.0    
    img = img * alpha + beta 
    
    # Changing the gamma to make warm tones more dominant
    gamma = 0.95
    img = 255.0 * ((img / 255.0) ** (1.0 / gamma))  
    
    # Making the image more saturated to give it an mystical and vibrant look
    hsv = cv2.cvtColor(to_uint8(img), cv2.COLOR_RGB2HSV).astype(np.float32)
    hsv[..., 1] *= 1.35  
    hsv[..., 1] = np.clip(hsv[..., 1], 0, 255)

    img = cv2.cvtColor(to_uint8(hsv), cv2.COLOR_HSV2RGB).astype(np.float32)
    
    # Now to make the image more warm we will add backgrounds tints to the image

    # Generating low pass and high pass masks based on the Value in HSV format
    hsv = cv2.cvtColor(to_uint8(img), cv2.COLOR_RGB2HSV).astype(np.float32)
    V = hsv[..., 2]  # 0-255
    shadow_mask = (V < 128).astype(np.float32)[..., None]
    highlight_mask = (V >= 128).astype(np.float32)[..., None]
    
    # Creating tint colors
    teal_tint = np.array([6.0, 18.0, 35.0], dtype=np.float32)  
    warm_tint = np.array([28.0, 10.0, 0.0], dtype=np.float32)
    
    # Applying the tints (Split tonning)
    img += shadow_mask * teal_tint
    img += highlight_mask * warm_tint
    
    return to_uint8(img)



def mystic_lagoon(rgb):
    img = rgb.astype(np.float32)
    
    # Hue rotation (small shift toward cyan/teal)
    hsv = cv2.cvtColor(to_uint8(img), cv2.COLOR_RGB2HSV).astype(np.float32)
    hsv[..., 0] = (hsv[..., 0] + 15) % 180  
    
    img = cv2.cvtColor(to_uint8(hsv), cv2.COLOR_HSV2RGB).astype(np.float32)
    
    # Channel mixing to cool tones to reduce red a bit, favor green+blue
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]
    # New channels formed as linear mixes 
    newR = 0.86 * R + 0.06 * G + 0.08 * B
    newG = 0.04 * R + 0.92 * G + 0.04 * B
    newB = 0.06 * R + 0.18 * G + 0.76 * B
    img = np.stack([newR, newG, newB], axis=-1)
    
    # Changing gamma to make the cooler tones more dominant
    gamma = 1.08
    img = 255.0 * ((img / 255.0) ** (1.0 / gamma))  
    
    # Adding fog effect to give a misty look
    h, w = img.shape[:2]
    fog_color = np.array([220.0, 235.0, 245.0], dtype=np.float32)
    alpha = np.linspace(0.35, 0.0, h).reshape(h, 1, 1)
    img = img * (1.0 - alpha) + fog_color * alpha
    
    return to_uint8(img)


rgb_image = img_array  

# Apply filters
out1 = sunny_saturday_holidays(rgb_image)
out2 = mystic_lagoon(rgb_image)

# Display original and filtered outputs 
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(rgb_image)
axes[0].axis('off')
axes[0].set_title("Original (input)")

axes[1].imshow(out1)
axes[1].axis('off')
axes[1].set_title("Sunny Saturday Holidays\n(contrast, brightness, saturation, warm tint)")

axes[2].imshow(out2)
axes[2].axis('off')
axes[2].set_title("Mystic Lagoon (Cool Mist)\n(hue shift, fog)")

plt.show()
````

#### White Balance Function
Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image. Use Numpy only to keep track of the brightest pixel and then scale the entire image with the factor you find. Display the original and output image.

````python
def white_patch_balance(image):

    new_image = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

    r_max = np.max(image[:,:,0])
    g_max = np.max(image[:,:,1])
    b_max = np.max(image[:,:,2])

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i,j][:3]

            r_new = (r/r_max)*255
            g_new = (g/g_max)*255
            b_new = (b/b_max)*255

            r_new = r_new//1
            g_new = g_new//1
            b_new = b_new//1

            new_image[i,j,0] = r_new
            new_image[i,j,1] = g_new
            new_image[i,j,2] = b_new

    return new_image.astype(np.uint8)

balanced_image = white_patch_balance(img_array)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_array)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(balanced_image)
plt.title("After White Patch Balancing")
plt.axis('off')

print("Even though the images look similar, the white patch balanced image has enhanced colors making the image more vibrant and bright.")
            
````

````python
# Trying linear scaling of RGB channels to create cold/warm filter
def cold_warm_filter_using_white_balance(image, value = 0.5):

    r_gain = 0.5 + value
    g_gain = 1
    b_gain = 1.5 - value

    new_image = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i,j][:3]

            r_new = (r * r_gain)
            g_new = (g * g_gain)
            b_new = (b * b_gain)

            r_new = min(r_new, 255)
            g_new = min(g_new, 255)
            b_new = min(b_new, 255)

            new_image[i,j,0] = r_new
            new_image[i,j,1] = g_new
            new_image[i,j,2] = b_new

    return new_image

filtered_image_1 = cold_warm_filter_using_white_balance(img_array, value=0.2)
filtered_image_2 = cold_warm_filter_using_white_balance(img_array, value=0.5)
filtered_image_3 = cold_warm_filter_using_white_balance(img_array, value=0.8)

plt.figure(figsize=(10,5))

plt.imshow(img_array)
plt.title("Original Image")
plt.axis('off')

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(filtered_image_1)
plt.title("cold Filter with value=0.2")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(filtered_image_2)
plt.title("Neutral Filter with value=0.5")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(filtered_image_3)
plt.title("Warm Filter with value=0.8")
plt.axis('off')
````

````python
# Trying sigmoid scaling of RGB channels to create cold/warm filter for more dramatic effect

# Hyperparameter for sigmoid function
k = 5  # steepness parameter for sigmoid function

def cold_warm_filter_using_sigmoid(image, value = 0.5, k=10):

    def sigmoid(x):
        return 1 / (1 + np.exp(-x))

    r_gain = 0.5 + sigmoid((value - 0.5) * k)
    g_gain = 1
    b_gain = 0.5 + sigmoid((0.5 - value) * k)

    new_image = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i,j][:3]

            r_new = (r * r_gain)
            g_new = (g * g_gain)
            b_new = (b * b_gain)

            r_new = min(r_new, 255)
            g_new = min(g_new, 255)
            b_new = min(b_new, 255)

            new_image[i,j,0] = r_new
            new_image[i,j,1] = g_new
            new_image[i,j,2] = b_new

    return new_image

filtered_image_1 = cold_warm_filter_using_sigmoid(img_array, value=0.2, k=k)
filtered_image_2 = cold_warm_filter_using_sigmoid(img_array, value=0.5, k=k)
filtered_image_3 = cold_warm_filter_using_sigmoid(img_array, value=0.8, k=k)


plt.figure(figsize=(10,5))
plt.imshow(img_array)
plt.title("Original Image")
plt.axis('off')

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(filtered_image_1)
plt.title("cold Filter with value=0.2")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(filtered_image_2)
plt.title("Neutral Filter with value=0.5")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(filtered_image_3)
plt.title("Warm Filter with value=0.8")
plt.axis('off')
````
