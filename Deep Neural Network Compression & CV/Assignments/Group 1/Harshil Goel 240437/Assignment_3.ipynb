````python
# Task 1

def convolve(image, kernel):

    k = kernel.shape[0]
    pad_height = k // 2
    pad_width = k // 2

    output = np.zeros_like(image)

    image_padded = np.pad(image, ((pad_height, pad_height), (pad_width, pad_width)), mode='reflect')

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            region = image_padded[i:i + k, j:j + k]
            output[i, j] = np.sum(region * kernel)

    return output


# Task 2

def average_blur_gray(image, kernel_size):

    kernel = np.ones((kernel_size, kernel_size)) / (kernel_size * kernel_size)
    blurred_image = convolve(image, kernel)

    return blurred_image

def average_blur_color(image, kernel_size):

    blurred_image = np.zeros_like(image)

    for c in range(3):
        blurred_image[:, :, c] = average_blur_gray(image[:, :, c], kernel_size)

    return blurred_image

# Task 3

def gaussian_kernel_generator(kernel_size, sigma):

    kernel = np.zeros((kernel_size, kernel_size))
    center = kernel_size // 2

    for i in range(kernel_size):
        for j in range(kernel_size):
            radius = (i - center) ** 2 + (j - center) ** 2
            kernel[i, j] = (1/(2 * np.pi * sigma ** 2)) * np.exp(-radius / (2 * sigma ** 2))

    kernel /= np.sum(kernel)

    return kernel

def gaussian_blur_gray(image, kernel_size, sigma):

    kernel = gaussian_kernel_generator(kernel_size, sigma)
    blurred_image = convolve(image, kernel)

    return blurred_image

def gaussian_blur_color(image, kernel_size, sigma):

    blurred_image = np.zeros_like(image)

    for c in range(3):
        blurred_image[:, :, c] = gaussian_blur_gray(image[:, :, c], kernel_size, sigma)

    return blurred_image

average_blured_img = average_blur_color(image_arr, 15)
gaussian_blured_img = gaussian_blur_color(image_arr, 15, 6.0)

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.imshow(image_arr)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(average_blured_img)
plt.title("Average Blurred Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gaussian_blured_img)
plt.title("Gaussian Blurred Image")
plt.axis("off")
````

````python
def gx_kernel():
    return np.array([[-1, 0, 1],
                     [-2, 0, 2],
                     [-1, 0, 1]])

def gy_kernel():
    return np.array([[1, 2, 1],
                     [0, 0, 0],
                     [-1, -2, -1]])

def sobel_edge_detection(image):
    Gx = gx_kernel()
    Gy = gy_kernel()

    gradient_x = convolve(image, Gx)
    gradient_y = convolve(image, Gy)

    magnitude = np.sqrt(gradient_x ** 2 + gradient_y ** 2)
    magnitude = ((magnitude / np.max(magnitude)) * 255).astype(np.uint8)

    # No thresholding applied
    return magnitude

gray_image_arr = np.zeros((image_arr.shape[0], image_arr.shape[1]), dtype=np.uint8)
for i in range(image_arr.shape[0]):
    for j in range(image_arr.shape[1]):
        gray_image_arr[i, j] = int(0.2989 * image_arr[i, j, 0] + 0.5870 * image_arr[i, j, 1] + 0.1140 * image_arr[i, j, 2])


edges_img = sobel_edge_detection(gray_image_arr)
plt.figure(figsize=(8, 8))

plt.subplot(1, 2, 1)
plt.imshow(gray_image_arr, cmap='gray')
plt.title("Original Gray Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges_img, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
````

````python
def laplacian_kernel_1():
    return np.array([[0, 1, 0],
                     [1, -4, 1],
                     [0, 1, 0]])

def laplacian_kernel_2():
    return np.array([[1, 1, 1],
                     [1, -8, 1],
                     [1, 1, 1]])

def laplacian_edge_detection(image, kernel):
    edges = convolve(image, kernel).astype(np.float64)
    return edges

def sharpen_image_4_connected(image, alpha):
    smooth = gaussian_blur_gray(image.astype(np.float64),10,3.0)
    edges = laplacian_edge_detection(smooth, laplacian_kernel_1())
    sharpened = image.astype(np.float64) - edges*alpha
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    return sharpened

def sharpen_image_8_connected(image, alpha):
    smooth = gaussian_blur_gray(image.astype(np.float64),10,3.0)
    edges = laplacian_edge_detection(smooth, laplacian_kernel_2())
    sharpened = image.astype(np.float64) - edges*alpha
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    return sharpened


sharpened_img_4_connected = sharpen_image_4_connected(gray_image_arr, 0.8)
sharpened_img_8_connected = sharpen_image_8_connected(gray_image_arr, 0.8)

plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.imshow(gray_image_arr, cmap='gray')
plt.title("Original Gray Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(sharpened_img_4_connected, cmap='gray')
plt.title("Sharpened Image using Laplacian (4-connected)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened_img_8_connected, cmap='gray')
plt.title("Sharpened Image using Laplacian (8-connected)")
plt.axis("off")

plt.tight_layout()
plt.show()
````

````python
# Task 2

def unsharp_masking(image, alpha, kernel_size, sigma):
    blurred = gaussian_blur_gray(image.astype(np.float64), kernel_size, sigma)
    mask = image.astype(np.float64) - blurred
    sharpened = image.astype(np.float64) + alpha * mask
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    return sharpened

unsharp_masked_img = unsharp_masking(gray_image_arr, 1.5, 10, 3.0)

blurred_using_unsharp = unsharp_masking(unsharp_masked_img, -1.5, 10, 3.0)

plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.imshow(gray_image_arr, cmap='gray')
plt.title("Original Gray Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(unsharp_masked_img, cmap='gray')
plt.title("Unsharp Masked Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blurred_using_unsharp, cmap='gray')
plt.title("Blurred back from Unsharp Masked Image")
plt.axis("off")
````

````python
def magnitude_spectrum(image):
    f = np.fft.fft2(image.astype(np.float64))
    fshift = np.fft.fftshift(f)
    magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)
    return magnitude_spectrum

mag_original = magnitude_spectrum(gray_image_arr)
mag_gaussian = magnitude_spectrum(gaussian_blur_gray(gray_image_arr, 10, 3.0))
mag_unsharp = magnitude_spectrum(unsharp_masking(gray_image_arr, 1.5, 10, 3.0))

plt.figure(figsize=(15, 6))
plt.subplot(1, 3, 1)
plt.imshow(mag_original, cmap='gray')
plt.title("Magnitude Spectrum of Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(mag_gaussian, cmap='gray')
plt.title("Magnitude Spectrum after Gaussian Blur")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(mag_unsharp, cmap='gray')
plt.title("Magnitude Spectrum after Unsharp Masking")
plt.axis("off")

# Varying sigma and alpha
mag_gaussian_sigma2 = magnitude_spectrum(gaussian_blur_gray(gray_image_arr, 10, 6.0))
mag_unsharp_alpha2 = magnitude_spectrum(unsharp_masking(gray_image_arr, 3.0, 10, 3.0))

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(mag_gaussian_sigma2, cmap='gray')
plt.title("Magnitude Spectrum after Gaussian Blur (sigma=6.0)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(mag_unsharp_alpha2, cmap='gray')
plt.title("Magnitude Spectrum after Unsharp Masking (alpha=3.0)")
plt.axis("off")


````

The magnitude spectrum of the original image shows strong energy at the center, indicating dominance of low-frequency components, with gradually decreasing intensity toward the edges representing high-frequency details (edges and textures).

After Gaussian blurring, the spectrum becomes more concentrated and brighter near the center while the outer regions turn darker, confirming that Gaussian convolution acts as a low-pass filter by suppressing high frequencies. Increasing σ further tightens this bright central region, removing even more fine details.

In unsharp masking, the magnitude spectrum shows relatively enhanced brightness away from the center, indicating amplification of high-frequency components (edges).

Increasing α in unsharp masking strengthens this effect, making high-frequency regions brighter while slightly reducing low-frequency dominance.

Overall, the dark regions in the spectrum correspond to suppressed frequencies, and bright regions indicate preserved or amplified frequencies, verifying that convolution filters frequencies according to the kernel’s frequency response.

````python
gaussian_img = gaussian_blur_color(img_part5, 5, 1.0)

red_part   = gaussian_img[:, :, 0]
green_part = gaussian_img[:, :, 1]
blue_part  = gaussian_img[:, :, 2]

difference_rg = red_part.astype(np.float64) - green_part.astype(np.float64)

# Contrast stretching (IMPORTANT)
p2, p98 = np.percentile(difference_rg, (2, 98))
difference_rg = np.clip((difference_rg - p2) / (p98 - p2 + 1e-6), 0, 1)
difference_rg *= 255

# Blue masking
mask_blue = blue_part > 0.55 * np.minimum(red_part, green_part)

new_difference_rg = difference_rg.copy()
new_difference_rg[mask_blue] = 0

average_diff = 0.5 * (difference_rg + new_difference_rg)

# Advanced Sobel Edge Detection with 5x5 Kernels and Thresholding
def sobel_edge_detection_large_kernel(image):

    image = image.astype(np.float64)

    Gx = np.array([[-3, -10, 0, 10, 3],
                   [-6, -20, 0, 20, 6],
                   [-9, -30, 0, 30, 9],
                   [-6, -20, 0, 20, 6],
                   [-3, -10, 0, 10, 3]], dtype=np.float64)

    Gy = np.array([[3, 6, 9, 6, 3],
                   [10, 20, 30, 20, 10],
                   [0, 0, 0, 0, 0],
                   [-10, -20, -30, -20, -10],
                   [-3, -6, -9, -6, -3]], dtype=np.float64)

    gradient_x = convolve(image, Gx)
    gradient_y = convolve(image, Gy)

    magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

    # Normalize safely
    max_val = np.max(magnitude)
    if max_val > 0:
        magnitude = (magnitude / max_val) * 255

    magnitude = magnitude.astype(np.uint8)

    # Advanced Thresholding
    threshold = 35
    strong_threshold = 110

    magnitude[magnitude < threshold] = 0

    # Neighborhood consistency check
    enhanced = magnitude.copy()
    rows, cols = magnitude.shape

    for i in range(1, rows - 1):
        for j in range(1, cols - 1):
            if magnitude[i, j] < strong_threshold:
                strong_neighbors = np.sum(
                    magnitude[i-1:i+2, j-1:j+2] >= strong_threshold
                )
                if strong_neighbors < 2:
                    enhanced[i, j] = 0

    return enhanced

sobel_difference_rg = sobel_edge_detection_large_kernel(average_diff)

plt.figure(figsize=(14, 5))

plt.subplot(1, 4, 1)
plt.imshow(gaussian_img)
plt.title("Gaussian Blurred Image")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(difference_rg, cmap='gray')
plt.title("R-G Difference before Blue Masking")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(average_diff, cmap='gray')
plt.title("Processed R-G Difference after Blue Masking")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(sobel_difference_rg, cmap='gray')
plt.title("Final Yellow Leaf Edge")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Due to applying blue masking, there we a few extra edges being created inside the leaf areas. However, overall the edges of the yellow leaves were detected effectively.")
````

````python
def rgb_to_hsv(image):
    image = image.astype(np.float32) / 255.0
    hsv_image = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i, j]
            mx = max(r, g, b)
            mn = min(r, g, b)
            diff = mx - mn

            # Hue calculation
            if diff == 0:
                h = 0
            elif mx == r:
                h = (60 * ((g - b) / diff) + 360) % 360
            elif mx == g:
                h = (60 * ((b - r) / diff) + 120) % 360
            else:
                h = (60 * ((r - g) / diff) + 240) % 360

            # Saturation calculation
            s = 0 if mx == 0 else (diff / mx)

            # Value calculation
            v = mx

            hsv_image[i, j] = [h / 360.0, s, v]

    return hsv_image

# detecting yellow leaves in hsv space

hsv_image = rgb_to_hsv(img_part5)

hue = hsv_image[:, :, 0] * 360
saturation = hsv_image[:, :, 1]

yellow_mask_hsv = (hue >= 25) & (hue <= 60) & (saturation >= 0.1)

# Visualizing the yellow mask in HSV space
plt.figure(figsize=(8, 8))
plt.imshow(yellow_mask_hsv, cmap='gray')
plt.title("Yellow Leaf Mask in HSV Space")
plt.axis("off")

# applying gaussian blur to the mask
yellow_mask_hsv = yellow_mask_hsv.astype(np.float64)
yellow_mask_hsv = gaussian_blur_gray(yellow_mask_hsv, 15, 5.0)

# Apply Sobel edge detection on the yellow mask (convert to uint8 if needed)
yellow_mask_uint8 = yellow_mask_hsv.astype(np.uint8) * 255
sobel_edges = sobel_edge_detection_large_kernel(yellow_mask_uint8)

plt.figure(figsize=(8, 8))
plt.imshow(sobel_edges, cmap='gray')
plt.title("Sobel Edges on Yellow Mask")
plt.axis("off")
plt.show()


````

````python
# converting rgb to hsv 

def rgb_to_hsv(image):
    image = image.astype(np.float32) / 255.0
    hsv_image = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r, g, b = image[i, j]
            mx = max(r, g, b)
            mn = min(r, g, b)
            diff = mx - mn

            # Hue calculation
            if diff == 0:
                h = 0
            elif mx == r:
                h = (60 * ((g - b) / diff) + 360) % 360
            elif mx == g:
                h = (60 * ((b - r) / diff) + 120) % 360
            else:
                h = (60 * ((r - g) / diff) + 240) % 360

            # Saturation calculation
            s = 0 if mx == 0 else (diff / mx)

            # Value calculation
            v = mx

            hsv_image[i, j] = [h / 360.0, s, v]

    return hsv_image

# detecting yellow leaves in hsv space

hsv_image = rgb_to_hsv(img_part5)

hue = hsv_image[:, :, 0] * 360
saturation = hsv_image[:, :, 1]

yellow_mask_hsv = (hue >= 30) & (hue <= 60) & (saturation >= 0.2)

# Visualizing the yellow mask in HSV space
plt.figure(figsize=(8, 8))
plt.imshow(yellow_mask_hsv, cmap='gray')
plt.title("Yellow Leaf Mask in HSV Space")
plt.axis("off")

# Converting yellow to red in hsv space

hsv_modified = hsv_image.copy()
# Rotate hue by 55 degrees (in HSV, hue is in [0,1], so add 55/360)
hue_shift = -55.0 / 360.0
hsv_modified[yellow_mask_hsv, 0] = (hsv_modified[yellow_mask_hsv, 0] + hue_shift) % 1.0

def hsv_to_rgb(hsv_image):
    rgb_image = np.zeros_like(hsv_image)

    for i in range(hsv_image.shape[0]):
        for j in range(hsv_image.shape[1]):
            h, s, v = hsv_image[i, j]
            h *= 360
            c = v * s
            x = c * (1 - abs((h / 60) % 2 - 1))
            m = v - c

            if 0 <= h < 60:
                r1, g1, b1 = c, x, 0
            elif 60 <= h < 120:
                r1, g1, b1 = x, c, 0
            elif 120 <= h < 180:
                r1, g1, b1 = 0, c, x
            elif 180 <= h < 240:
                r1, g1, b1 = 0, x, c
            elif 240 <= h < 300:
                r1, g1, b1 = x, 0, c
            else:
                r1, g1, b1 = c, 0, x

            r = (r1 + m) * 255
            g = (g1 + m) * 255
            b = (b1 + m) * 255

            rgb_image[i, j] = [r, g, b]

    return rgb_image.astype(np.uint8)

rgb_modified = hsv_to_rgb(hsv_modified)

plt.figure(figsize=(8, 8))
plt.imshow(rgb_modified)
plt.title("Image with Yellow Leaves Converted to Red")
plt.axis("off")
````
