````python
base64_image = ''''''

img_data = base64.b64decode(base64_image)
img_rgb = Image.open(io.BytesIO(img_data))
img_rgb = np.array(img_rgb)

plt.imshow(img_rgb)
plt.axis("off")
````

````python
# Convert to grayscale (Weighted sum formula)
# R is channel 0, G is channel 1, B is channel 2
r, g, b = img_rgb[:,:,0], img_rgb[:,:,1], img_rgb[:,:,2]

# Calculate weighted sum
gray_numpy = 0.299 * r + 0.587 * g + 0.114 * b
gray_numpy = gray_numpy.astype(np.uint8)  # Result is float, so convert back to uint8

# Display the NumPy grayscale result
plt.figure(figsize=(6, 6))
plt.imshow(gray_numpy, cmap='gray')
plt.title("Grayscale via NumPy (Weighted Sum)")
plt.axis("off")
plt.show()
````

````python
# Using cvtColor function
gray_cv2 = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Compare results
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(gray_numpy, cmap='gray')
ax[0].set_title("NumPy Implementation")
ax[0].axis("off")

ax[1].imshow(gray_cv2, cmap='gray')
ax[1].set_title("cv2.cvtColor Function")
ax[1].axis("off")
plt.show()

# Check numerical difference (should be very close to 0)
diff = np.abs(gray_numpy.astype(int) - gray_cv2.astype(int))
print(f"Average pixel difference between methods: {np.mean(diff):.4f}")
````

## Part B
### B1. Compute the FFT of the grayscale image ( use NumPy )
Steps:
1. <code>np.fft.fft2</code>

### B2. Compute magnitude, phase and display Magnitude Spectrum

````python
# Compute the FFT
f = np.fft.fft2(gray_numpy)
fshift = np.fft.fftshift(f)  # Shift the zero-frequency component to the center of the spectrum

# Calculate Magnitude Spectrum
# We use log scale (decibels) because the dynamic range of the FFT is very large
# 20 * log10(|fshift|)
magnitude_spectrum = 20 * np.log(np.abs(fshift))

# Calculate Phase Spectrum
phase_spectrum = np.angle(fshift)
````

````python
# Display Results
plt.figure(figsize=(18, 6))

# Original Grayscale Image
plt.subplot(1, 3, 1)
plt.imshow(gray_numpy, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Magnitude Spectrum
plt.subplot(1, 3, 2)
plt.imshow(magnitude_spectrum, cmap='gray')
plt.title('Magnitude Spectrum')
plt.axis('off')

# Phase Spectrum
plt.subplot(1, 3, 3)
plt.imshow(phase_spectrum, cmap='gray')
plt.title('Phase Spectrum')
plt.axis('off')

plt.show()
````

````python
# Setup Center coordinates
rows, cols = gray_numpy.shape
crow, ccol = rows // 2, cols // 2

# Helper function to perform Inverse FFT and Magnitude calculation
def reconstruct_image(fshift_filtered):
    # Move the zero-frequency component back to the corner
    f_ishift = np.fft.ifftshift(fshift_filtered)
    # Perform Inverse FFT
    img_back = np.fft.ifft2(f_ishift)
    # Take the magnitude (absolute value) to get the image
    return np.abs(img_back)
````

````python
radii = [10, 30,80]  # Small radius (more blur) to Large radius (less blur)

plt.figure(figsize=(18, 6))

for i, r in enumerate(radii):
    # Create a mask of zeros
    mask = np.zeros((rows, cols), np.uint8)

    # Set the center circle to 1 (white) - Ideal LPF
    # Using np.ogrid to create a grid of coordinates
    y, x = np.ogrid[:rows, :cols]
    mask_area = (x - ccol)**2 + (y - crow)**2 <= r*r
    mask[mask_area] = 1

    # Apply mask to the shifted FFT (Element-wise multiplication)
    fshift_filtered = fshift * mask

    # Reconstruct
    img_lpf = reconstruct_image(fshift_filtered)

    # Display result
    plt.subplot(1, 3, i+1)
    plt.imshow(img_lpf, cmap='gray')
    plt.title(f'LPF Radius: {r}')
    plt.axis('off')

plt.show()

# REASON FOR BLUR:
# The center of the spectrum contains low frequencies (general intensity/smoothness).
# The outer regions contain high frequencies (edges/noise/details).
# As the radius decreases, we cut off more high-frequency information.
# Losing high frequencies results in the loss of sharp edges, creating a blur effect.
````

````python
# Phase-only reconstruction

# Method 1: Using the complex phasor (Magnitude = 1, Phase = original)
# Adding 1e-5 to the denominator to avoid division by zero errors
phase_only_spectrum = fshift / (np.abs(fshift) + 1e-5)

# Reconstruct using the helper function from part C
img_phase_only = reconstruct_image(phase_only_spectrum)

# Method 2: Trying to reconstruct directly from np.angle(fshift)
angle_values = np.angle(fshift)

# Attempting reconstruction directly from these scalar values
img_angle_raw = reconstruct_image(angle_values)

# Display Results
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img_phase_only, cmap='gray')
plt.title('Phase Only')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_angle_raw, cmap='gray')
plt.title('Reconstruction from Raw np.angle')
plt.axis('off')

plt.show()

# Difference between code lines:
# 1. 'fshift / np.abs(fshift)' creates a COMPLEX number with Magnitude 1 and the original phase.
#    Mathematically, this is e^(j*theta). The FFT expects complex inputs.
# 2. 'np.angle(fshift)' returns a REAL number (float) representing the angle theta in radians.
````

````python
# Explanation of Output:
# - The first image works (shows edges) because we provided valid complex phasors to the IFFT.
# - The second image is garbage/noise. This is because we fed raw scalar numbers (radians)
#   into the IFFT. The IFFT interpreted these radians as frequency magnitudes/coefficients,
#   which is mathematically meaningless for image reconstruction.
````

````python
# Magnitude-only reconstruction

# Create a spectrum with Original Magnitude but Zero Phase (Real numbers)
mag_only_spectrum = np.abs(fshift)

# Reconstruct
img_mag_only = reconstruct_image(mag_only_spectrum)

# Display
plt.figure(figsize=(6, 6))
plt.imshow(img_mag_only, cmap='gray')
plt.title('Magnitude Only')
plt.axis('off')
plt.show()

# Role of Logarithm:
# In Part B, we used '20 * log(abs(fshift))' to DISPLAY the spectrum.
# The dynamic range of FFT magnitudes is massive (e.g., 0 to 1,000,000).
# Without the log, we would only see the single white center pixel (DC component),
# and all other frequencies would look black. The log compresses this range so we can see details.

# Why does this reconstruction look different from the Magnitude Spectrum?
# 1. Domain Difference: The "Magnitude Spectrum" in Part B was a frequency domain PLOT.
#    The image above is the spatial domain RECONSTRUCTION (result of IFFT).
# 2. Content: The reconstruction looks like a blob because we threw away the Phase.
#    Without phase, all the sine waves align at the origin (center), creating a bright spot/blob,
#    but they don't interfere constructively to form the edges of the original object.
````

````python
base64_img2 = ''''''

img_data2 = base64.b64decode(base64_img2)
img_rgb2 = Image.open(io.BytesIO(img_data2))
img_rgb2 = np.array(img_rgb2)

plt.imshow(img_rgb2)
plt.axis("off")
````

````python
# Convert Image 2 to Grayscale using manual numpy conversion
gray_numpy2 = np.dot(img_rgb2[...,:3], [0.299, 0.587, 0.114])

# Resize Image 2 to match Image 1's shape exactly
# (This is required so their frequency spectrums are compatible)
target_shape = (gray_numpy.shape[1], gray_numpy.shape[0])
gray_numpy2 = cv2.resize(gray_numpy2, target_shape)

print(f"Image 1 Shape: {gray_numpy.shape}")
print(f"Image 2 Shape: {gray_numpy2.shape}")
````

````python
# Compute FFT for Image 1
f1 = np.fft.fft2(gray_numpy)
fshift1 = np.fft.fftshift(f1)

# Compute FFT for Image 2
f2 = np.fft.fft2(gray_numpy2)
fshift2 = np.fft.fftshift(f2)

# Extract Magnitude and Phase
# Image 1 Components
mag1 = np.abs(fshift1)
phase1 = np.angle(fshift1)

# Image 2 Components
mag2 = np.abs(fshift2)
phase2 = np.angle(fshift2)
````

````python
# Combine Magnitude of Image 1 with Phase of Image 2
# F_new1 = |F1| * e^(j * theta2)
fshift_swapped_1 = mag1 * np.exp(1j * phase2)

# Combine Magnitude of Image 2 with Phase of Image 1
# F_new2 = |F2| * e^(j * theta1)
fshift_swapped_2 = mag2 * np.exp(1j * phase1)
````

````python
# Reconstruct the swapped images
img_swapped1 = reconstruct_image(fshift_swapped_1) # Phase from 2
img_swapped2 = reconstruct_image(fshift_swapped_2) # Phase from 1

# Display results
plt.figure(figsize=(12, 10))

# Show Original Image 1
plt.subplot(2, 2, 1)
plt.imshow(gray_numpy, cmap='gray')
plt.title('Original Image 1')
plt.axis('off')

# Show Original Image 2
plt.subplot(2, 2, 2)
plt.imshow(gray_numpy2, cmap='gray')
plt.title('Original Image 2')
plt.axis('off')

# Show Result A (Mag 1 + Phase 2)
plt.subplot(2, 2, 3)
plt.imshow(img_swapped1, cmap='gray')
plt.title('Result A: Mag 1 + Phase 2')
plt.axis('off')

# Show Result B (Mag 2 + Phase 1)
plt.subplot(2, 2, 4)
plt.imshow(img_swapped2, cmap='gray')
plt.title('Result B: Mag 2 + Phase 1')
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
# We can see that Result A looks like Image 2, and Result B looks like Image 1.
# This proves that:
# 1. PHASE carries the structural information (edges, shapes, geometry).
#    Since Result A used Phase 2, it retained the shape of Image 2.
# 2. MAGNITUDE carries the texture/intensity information (brightness, contrast patterns).
#    Result A looks like a "ghostly" version of Image 2 because it has the
#    wrong intensity distribution (borrowed from Image 1).
````
