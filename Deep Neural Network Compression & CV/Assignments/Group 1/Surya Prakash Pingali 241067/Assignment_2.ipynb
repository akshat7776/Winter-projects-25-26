````python
def gray_histogram(gray_img):
    """
    Calculates the histogram of a grayscale image using NumPy loops.
    """
    # Initialize histogram bins (0 to 255) with zeros
    hist = np.zeros(256, dtype=int)

    # Get image dimensions
    height, width = gray_img.shape

    # Using NumPy loops to iterate over every pixel
    for i in range(height):
        for j in range(width):
            intensity = gray_img[i, j]  # Get the pixel intensity value
            hist[intensity] += 1  # Increment the corresponding bin count

    return hist

# Calculate the histogram
hist_values = gray_histogram(img_gray)

# Plot using matplotlib
plt.figure(figsize=(10, 6))
plt.bar(range(256), hist_values, color='gray', width=1.0)
plt.title('Grayscale Histogram')
plt.xlabel('Pixel Value (0-255)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.5)
plt.show()
````

````python
def rgb_histogram(rgb_img):
    """
    Calculates the histogram for R, G, and B channels using NumPy loops.
    """
    height, width, channels = rgb_img.shape

    # Initialize histograms for Red, Green, and Blue channels
    hist_r = np.zeros(256, dtype=int)
    hist_g = np.zeros(256, dtype=int)
    hist_b = np.zeros(256, dtype=int)

    # Loop through every pixel
    for i in range(height):
        for j in range(width):
            # Extract intensity values for each channel
            r = rgb_img[i, j, 0]
            g = rgb_img[i, j, 1]
            b = rgb_img[i, j, 2]

            # Increment the respective bins
            hist_r[r] += 1
            hist_g[g] += 1
            hist_b[b] += 1

    return hist_r, hist_g, hist_b

# Calculate RGB histograms
r_hist, g_hist, b_hist = rgb_histogram(img_rgb)

# Plotting the results on the same graph
plt.figure(figsize=(10, 6))
plt.plot(r_hist, color='red', label='Red Channel')
plt.plot(g_hist, color='green', label='Green Channel')
plt.plot(b_hist, color='blue', label='Blue Channel')

plt.title('RGB Histogram')
plt.xlabel('Pixel Value (0-255)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

Write a function <code>rgb_to_hsv</code> that takes a 0-255 ( not normalised ) RGB image and returns an HSV image ( also not normalised ).

````python
def rgb_to_hsv(img_rgb):
    """
    Converts an RGB image to HSV (using NumPy formulas).
    """
    # Separate channels
    r = img_float[..., 0]
    g = img_float[..., 1]
    b = img_float[..., 2]

    # Calculate Value (V), Max, Min, and Delta
    v = np.max(img_float, axis=2)      # Max of r, g, b
    v_min = np.min(img_float, axis=2)  # Min of r, g, b
    delta = v - v_min

    # Initialize S and H arrays with zeros
    s = np.zeros_like(v)
    h = np.zeros_like(v)

    # Calculate Saturation (S)
    non_zero_v = v > 0  # Avoid division by zero
    s[non_zero_v] = delta[non_zero_v] / v[non_zero_v]  # Formula: S = delta / V (if V != 0)

    # Calculate Hue (H)
    # Only calculate H where delta > 0 (otherwise H is 0, which is grayscale)
    non_zero_delta = delta > 0

    # Create boolean masks for which channel is the max
    # (Combine with non_zero_delta to avoid calculating for grayscale pixels)
    mask_r = (r == v) & non_zero_delta
    mask_g = (g == v) & non_zero_delta
    mask_b = (b == v) & non_zero_delta

    # Formula when Red is max: H = (G - B) / Delta
    h[mask_r] = (g[mask_r] - b[mask_r]) / delta[mask_r]
    h[mask_r] = np.mod(h[mask_r], 6)  # For when G < B

    # Formula when Green is max: H = (B - R) / Delta + 2
    h[mask_g] = ((b[mask_g] - r[mask_g]) / delta[mask_g]) + 2

    # Formula when Blue is max: H = (R - G) / Delta + 4
    h[mask_b] = ((r[mask_b] - g[mask_b]) / delta[mask_b]) + 4

    # Convert H to degrees
    h = h * 60

    # Scale for OpenCV 8-bit Compatibility
    h_opencv = h / 2
    s_opencv = s * 255
    v_opencv = v * 255

    # Merge channels and convert to uint8
    hsv_img = np.dstack((h_opencv, s_opencv, v_opencv))
    return np.round(hsv_img).astype(np.uint8)
````

````python
# Defining Helper Functions

def adjust_gamma(image, gamma=1.0):
    """
    Apply Gamma Correction.
    Gamma < 1: Image gets brighter (bleached).
    Gamma > 1: Image gets darker (richer colors).
    """
    inv_gamma = 1.0 / gamma
    # Create a lookup table for speed (0-255)
    table = np.array([((i / 255.0) ** inv_gamma) * 255
                      for i in np.arange(0, 256)]).astype("uint8")
    return cv2.LUT(image, table)

def clip_img(img):
    """Ensures values stay within 0-255 and converts to uint8."""
    return np.clip(img, 0, 255).astype(np.uint8)
````

````python
def filter_garmi(img_rgb):
    """
    Applies a warm 'Garmi' effect.
    1. Warm Tint (Red/Green boost).
    2. Gamma Correction (Richness).
    """

    filtered = img_rgb.copy().astype(np.float32)  # Creating a copy

    # Warming
    filtered[:, :, 0] = filtered[:, :, 0] * 1.3  # Boost Red channel by 30%
    filtered[:, :, 1] = filtered[:, :, 1] * 1.1  # Boost Green channel by 10% (Yellow = Red + Green)

    # Clip values to 0-255
    filtered = np.clip(filtered, 0, 255).astype(np.uint8)

    # Gamma Correction
    filtered = adjust_gamma(filtered, gamma=1.2)  # Makes shadows slightly darker/richer

    return filtered
````

````python
def snyder_ka_filter(img_rgb):
    """
    Applies a dark and desaturated effect, just like in snyder cut.
    The opposite of Garmi: desaturated and high contrast.
    1. Desaturation (S channel down).
    2. High Contrast (V channel spread).
    """
    # Convert to HSV
    hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV).astype(np.float32)

    # Desaturation
    # Multiply Saturation by 0.4 (Retain only 40% of color)
    # This makes it look "faded" or "metallic"
    hsv[:, :, 1] = hsv[:, :, 1] * 0.4

    # High Contrast
    # Formula: factor * (pixel - 128) + 128
    # factor > 1 increases contrast (darks get darker, brights get brighter)
    contrast_factor = 1.4
    hsv[:, :, 2] = (hsv[:, :, 2] - 128) * contrast_factor + 128

    # Clip values to stay within [0, 255]
    hsv = np.clip(hsv, 0, 255)

    # Convert back to RGB
    return cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB)
````

````python
def white_patch_balance(img_rgb):
    """
    Applies White Patch algorithm to white balance an image.
    Assumption: The brightest pixel in each channel should be white (255).
    """
    # Use float32 to prevent overflow during multiplication

    # Find the maximum value (brightest pixel) for each channel
    # (Use axis=(0, 1) to find the max across the height and width)
    r_max = np.max(img_float[:, :, 0])
    g_max = np.max(img_float[:, :, 1])
    b_max = np.max(img_float[:, :, 2])

    # Calculate scaling factors
    # Target is 255. Factor = 255 / Current_Max
    # (Add a small epsilon 1e-6 to avoid division by zero if an image is pure black)
    scale_r = 255.0 / (r_max + 1e-6)
    scale_g = 255.0 / (g_max + 1e-6)
    scale_b = 255.0 / (b_max + 1e-6)

    # Apply the scaling factors to the respective channels
    balanced_img = np.zeros_like(img_float)
    balanced_img[:, :, 0] = img_float[:, :, 0] * scale_r
    balanced_img[:, :, 1] = img_float[:, :, 1] * scale_g
    balanced_img[:, :, 2] = img_float[:, :, 2] * scale_b

    # Clip values to range [0, 255] and convert back to uint8
    balanced_img = np.clip(balanced_img, 0, 255).astype(np.uint8)

    return balanced_img
````

````python
def white_balance(img_rgb, value):
    """
    Adjusts the color temperature of an image.
    """

    # Create a local float version of the image, normalized to [0, 1]
    img_local_float = img_rgb.astype(np.float32) / 255.0

    # Define the maximum strength of the effect
    #   0.5 means the channels can change by up to +50% or -50%
    max_adjustment = 0.5

    # Calculate the shift factor from the input value
    #   Input: 0.0  ->  0.5  ->  1.0
    #   Shift: -1.0 ->  0.0  ->  +1.0
    shift = (value - 0.5) * 2

    # Calculate scaling factors for Red and Blue channels
    #   If shift > 0 (Warm): Red increases, Blue decreases
    #   If shift < 0 (Cold): Red decreases, Blue increases
    r_factor = 1.0 + (shift * max_adjustment)
    b_factor = 1.0 - (shift * max_adjustment)

    # Apply the factors
    #   Green channel is left untouched to anchor the brightness
    img_local_float[:, :, 0] = img_local_float[:, :, 0] * r_factor # Red
    img_local_float[:, :, 2] = img_local_float[:, :, 2] * b_factor # Blue

    # Clip values to valid range [0, 1], then scale to [0, 255] and convert to uint8
    return (np.clip(img_local_float, 0, 1) * 255.0).astype(np.uint8)
````
