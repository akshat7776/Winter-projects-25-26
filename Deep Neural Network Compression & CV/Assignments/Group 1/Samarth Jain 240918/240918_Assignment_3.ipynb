````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

original_image = Image.open('image.jpg')
img_rgb = np.array(original_image)

img_gray = np.dot(img_rgb[...,:3], [0.2989, 0.5870, 0.1140])

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original RGB")
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Grayscale Version")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.show()
````

````python
def convolve(image, kernel):

    kernel = np.flipud(np.fliplr(kernel))

    img_h, img_w = image.shape
    k_h, k_w = kernel.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_image = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    output = np.zeros_like(image)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_image[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output
````

````python
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize * ksize)
    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    r_blur = average_blur_gray(img[:,:,0], ksize)
    g_blur = average_blur_gray(img[:,:,1], ksize)
    b_blur = average_blur_gray(img[:,:,2], ksize)

    return np.dstack((r_blur, g_blur, b_blur)).astype(int)

blurred_gray = average_blur_gray(img_gray, 5)
blurred_rgb = average_blur_rgb(img_rgb, 15)
````

````python
def gaussian_kernel(ksize, sigma):
    axis = np.linspace(-(ksize // 2), ksize // 2, ksize)
    x, y = np.meshgrid(axis, axis)

    kernel = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    return kernel / np.sum(kernel)

g_kernel = gaussian_kernel(ksize=5, sigma=1.0)
gaussian_gray = convolve(img_gray, g_kernel)

gaussian_r = convolve(img_rgb[:,:,0], g_kernel)
gaussian_g = convolve(img_rgb[:,:,1], g_kernel)
gaussian_b = convolve(img_rgb[:,:,2], g_kernel)
gaussian_rgb = np.dstack((gaussian_r, gaussian_g, gaussian_b)).astype(int)
````

````python
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(blurred_rgb)
plt.title("Average Blur (RGB)")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(blurred_gray, cmap='gray')
plt.title("Average Blur (Gray)")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(g_kernel, cmap='viridis')
plt.title("Gaussian Kernel (Visualized)")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Blur (RGB)")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(gaussian_gray, cmap='gray')
plt.title("Gaussian Blur (Gray)")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
# Your PART B codes follow here. You can create more cells under this. Write clean and commented code.

import numpy as np

def rgb_to_hsv(image_rgb):

    img = image_rgb.astype(np.float32) / 255.0


    r = img[:, :, 0]
    g = img[:, :, 1]
    b = img[:, :, 2]

    c_max = np.max(img, axis=2)
    c_min = np.min(img, axis=2)
    delta = c_max - c_min

    v = c_max

    s = np.zeros_like(v)
    non_zero = c_max > 0
    s[non_zero] = delta[non_zero] / c_max[non_zero]

    h = np.zeros_like(v)

    mask_r = (c_max == r) & (delta > 0)
    h[mask_r] = (g[mask_r] - b[mask_r]) / delta[mask_r]

    mask_g = (c_max == g) & (delta > 0)
    h[mask_g] = 2.0 + (b[mask_g] - r[mask_g]) / delta[mask_g]

    mask_b = (c_max == b) & (delta > 0)
    h[mask_b] = 4.0 + (r[mask_b] - g[mask_b]) / delta[mask_b]

    h = h * 60.0
    h[h < 0] += 360.0

    final_h = (h / 2).astype(np.uint8)      # 0-360 -> 0-179
    final_s = (s * 255).astype(np.uint8)    # 0.0-1.0 -> 0-255
    final_v = (v * 255).astype(np.uint8)    # 0.0-1.0 -> 0-255

    hsv_image = cv2.merge([final_h, final_s, final_v])

    return hsv_image
````

````python
import cv2
img_bgr = cv2.imread("image.jpg")

if img_bgr is None:
    print("Error: Please upload 'image.jpg' to Colab.")
else:
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    print("Converting RGB to HSV...")
    my_hsv = rgb_to_hsv(img_rgb)

    reconverted_rgb = cv2.cvtColor(my_hsv, cv2.COLOR_HSV2RGB)

    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.title("Original RGB")
    plt.imshow(img_rgb)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.title("Re-converted (HSV -> RGB)\nShould match Original")
    plt.imshow(reconverted_rgb)
    plt.axis("off")

    plt.show()

    print("If the images look identical, your Part B code is correct!")
````

````python
# Your PART C codes follow here. You can create more cells under this. Write clean and commented code.

img_path = 'image.jpg'
img_gray = cv2.imread(img_path, 0)

if img_gray is None:
    print("Error: Please upload 'image.jpg' to Colab.")
else:
    print(f"Image loaded. Shape: {img_gray.shape}")

def convolve(image, kernel):
    """
    Slides a kernel over the image to apply a filter.
    """
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    output = np.zeros_like(image, dtype=np.float32)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(region * kernel)

    return output

def get_gaussian_kernel(ksize, sigma):
    """Creates a 2D Gaussian kernel."""
    ax = np.linspace(-(ksize - 1) / 2., (ksize - 1) / 2., ksize)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-0.5 * (np.square(xx) + np.square(yy)) / np.square(sigma))
    return kernel / np.sum(kernel)
````

````python

laplacian_kernel = np.array([
    [0,  1, 0],
    [1, -4, 1],
    [0,  1, 0]
])

print("Applying Laplacian Convolution...")
lap_response = convolve(img_gray, laplacian_kernel)

alpha = 0.5
img_sharpened = img_gray - (lap_response * alpha)

img_sharpened = np.clip(img_sharpened, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.title("Original Grayscale")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Laplacian Edges")
plt.imshow(lap_response, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title(f"Sharpened (Alpha={alpha})")
plt.imshow(img_sharpened, cmap='gray')
plt.axis('off')

plt.show()
````

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.

def get_magnitude_spectrum(img):
    """
    Computes the Frequency Magnitude Spectrum of an image using FFT.
    """
    f = np.fft.fft2(img)

    fshift = np.fft.fftshift(f)

    magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1e-6)

    return magnitude_spectrum

print("Spectrum function defined.")
````

````python
img = cv2.imread('image.jpg', 0)

if img is None:
    print("Error: Please upload 'image.jpng'.")
else:
    blurred = cv2.GaussianBlur(img, (15, 15), 0)

    mask = img.astype(float) - blurred.astype(float)
    alpha = 2.0
    sharpened = img.astype(float) + (alpha * mask)
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

    print("Images prepared: Original, Blurred, and Sharpened.")
````

````python
spec_original = get_magnitude_spectrum(img)
spec_blurred = get_magnitude_spectrum(blurred)
spec_sharpened = get_magnitude_spectrum(sharpened)

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title("Gaussian Blurred (LPF)")
plt.imshow(blurred, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title("Sharpened (Boosted HPF)")
plt.imshow(sharpened, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title("Original Spectrum")
plt.imshow(spec_original, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title("Blurred Spectrum\n(Outer regions darker)")
plt.imshow(spec_blurred, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title("Sharpened Spectrum\n(Outer regions brighter)")
plt.imshow(spec_sharpened, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()

"""OBSERVATIONS:
1. CENTER (Low Frequencies): All images have a bright center because
   all images still have the main shape/background info.
2. BLURRED SPECTRUM: The outer region is black/dark.
   This proves Convolution with a Gaussian Kernel is a LOW PASS FILTER.
   It blocked the high-frequency edges.
3. SHARPENED SPECTRUM: The outer region has more white noise/dots.
   This proves Sharpening amplifies HIGH FREQUENCIES (details/noise)."""
````

````python
def apply_custom_kernel(image_channel, kernel):

    rows, cols = image_channel.shape


    result = np.zeros((rows - 2, cols - 2))


    result += image_channel[0:-2, 0:-2] * kernel[0, 0]
    result += image_channel[0:-2, 1:-1] * kernel[0, 1]
    result += image_channel[0:-2, 2:  ] * kernel[0, 2]

    result += image_channel[1:-1, 0:-2] * kernel[1, 0]
    result += image_channel[1:-1, 1:-1] * kernel[1, 1]
    result += image_channel[1:-1, 2:  ] * kernel[1, 2]

    result += image_channel[2:  , 0:-2] * kernel[2, 0]
    result += image_channel[2:  , 1:-1] * kernel[2, 1]
    result += image_channel[2:  , 2:  ] * kernel[2, 2]

    return result
````

````python
red_channel = image_matrix[:, :, 0].astype(int)

custom_kernel = np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
])

edge_response = apply_custom_kernel(red_channel, custom_kernel)

edge_response = np.abs(edge_response)
edge_response = np.clip(edge_response, 0, 255)

plt.figure(figsize=(10, 5))

plt.imshow(edge_response, cmap='gray')
plt.title("Part E: Yellow Leaf Edges (Red Channel Gradient)")
plt.axis('off')
plt.show()
````
