````python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
img=cv.imread('image.jpg')
img=cv.resize(img,(600, int(img.shape[0] * 600 / img.shape[1])))
img=cv.cvtColor(img,cv.COLOR_BGR2RGB)
img_gray=cv.cvtColor(img,cv.COLOR_RGB2GRAY)
#Task 1 create function
def convolve(img, kernel):
    # 1. Flip the kernel
    kernel = np.flipud(np.fliplr(kernel))
    # 2. Get shapes
    img_h, img_w = img.shape
    kh, kw = kernel.shape
    # 3. Padding (As requested: mode='reflect')
    # We pad enough so the kernel can center on the edge pixels
    pad_h = kh // 2
    pad_w = kw // 2
    img_padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    # 4. Convolution (Shift & Add)
    # Instead of sliding a window (slow in Python),  shifting the entire image
    output = np.zeros_like(img, dtype=np.float64)
    for i in range(kh):
        for j in range(kw):
            # Extract the "shifted" version of the image
            # equivalent to sliding the kernel
            region = img_padded[i : i + img_h, j : j + img_w]
            # Multiply by the corresponding kernel weight and add
            output += region * kernel[i, j]
    return output
#Task 2-Part 1(gray)
def average_blur_gray(img,ksize): #here img is in grayscale means 1 matrix not 3
  kernel=np.ones((ksize,ksize))/ksize**2 # kernel values are 1/ksize^2
  return convolve(img,kernel) # we return convolved matrix which is average blur gray
#Task 2-Part 1(rgb)
def average_blur_rgb(img,ksize):# for rgb we just break it into three different matrices and apply the previous function
  return np.dstack((average_blur_gray(img[:,:,0],ksize),average_blur_gray(img[:,:,1],ksize),average_blur_gray(img[:,:,2],ksize)))
#Task 3-kernel formation
def gaussian_kernel(ksize,sigma):
  kernel=np.zeros((ksize,ksize)) #created square matrix of size ksize with all values set to zero
  for i in range(ksize):
    for j in range(ksize):
      kernel[i,j]=np.exp(-((i-ksize//2)**2+(j-ksize//2)**2)/(2*sigma**2))/(2*np.pi*sigma**2) #formula of guassian
  return kernel/np.sum(kernel) #normalising
#Task 3-guassian blur(convolve the image with kernel)(gray)
def gaussian_blur_gray(img,ksize,sigma):
  kernel=gaussian_kernel(ksize,sigma)
  return convolve(img,kernel)
#Task 3-guassian blur(convolve the image with kernel)(rgb)
def gaussian_blur_rgb(img,ksize,sigma):
  return np.dstack((gaussian_blur_gray(img[:,:,0],ksize,sigma),gaussian_blur_gray(img[:,:,1],ksize,sigma),gaussian_blur_gray(img[:,:,2],ksize,sigma)))
avg_blur_result_gray=average_blur_gray(img_gray,ksize=11)
avg_blur_result_rgb=average_blur_rgb(img,ksize=11)
gauss_blur_result_rgb=gaussian_blur_rgb(img,ksize=11,sigma=11.0)
gauss_blur_result_gray=gaussian_blur_gray(img_gray,ksize=11,sigma=11.0)
plt.figure(figsize=(15, 10))
# --- ROW 1: Grayscale Images ---
# Plot 1: Original Gray
plt.subplot(2, 3, 1)
plt.imshow(img_gray.astype(np.uint8), cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')
# Plot 2: Gaussian Blur Gray
plt.subplot(2, 3, 2)
plt.imshow(gauss_blur_result_gray.astype(np.uint8), cmap='gray')
plt.title("Gaussian Blur (Gray)")
plt.axis('off')
# Plot 3: Average Blur Gray
plt.subplot(2, 3, 3)
plt.imshow(avg_blur_result_gray.astype(np.uint8), cmap='gray')
plt.title("Average Blur (Gray)")
plt.axis('off')
# --- ROW 2: RGB Images ---
# Plot 4: Original RGB
plt.subplot(2, 3, 4)
plt.imshow(img.astype(np.uint8))
plt.title("Original RGB")
plt.axis('off')
# Plot 5: Gaussian Blur RGB
plt.subplot(2, 3, 5)
plt.imshow(gauss_blur_result_rgb.astype(np.uint8))
plt.title("Gaussian Blur (RGB)")
plt.axis('off')
# Plot 6: Average Blur RGB
plt.subplot(2, 3, 6)
plt.imshow(avg_blur_result_rgb.astype(np.uint8))
plt.title("Average Blur (RGB)")
plt.axis('off')
# Adjust layout to prevent titles from overlapping
plt.tight_layout()
plt.show()
````
