````python
import numpy as np

def convolve(img, kernel):
    kernel = np.flipud(np.fliplr(kernel))

    img_h, img_w = img.shape
    k_h, k_w = kernel.shape
    pad_h = k_h // 2
    pad_w = k_w // 2
    padded_img = np.pad(
        img,
        ((pad_h, pad_h), (pad_w, pad_w)),
        mode='reflect'
    )
    out = np.zeros_like(img, dtype=np.float64)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i:i+k_h, j:j+k_w]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
def average_blur_gray(img, ksize):

    kernel = np.ones((ksize, ksize), dtype=np.float64)
    kernel /= kernel.sum()

    pad = ksize // 2
    padded_img = np.pad(img, ((pad, pad), (pad, pad)), mode='reflect')

    h, w = img.shape
    out = np.zeros_like(img, dtype=np.float64)
    for i in range(h):
        for j in range(w):
            region = padded_img[i:i+ksize, j:j+ksize]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
def average_blur_rgb(img_rgb, ksize):

    out = np.zeros_like(img_rgb, dtype=np.float64)

    for c in range(3):  # R, G, B
        out[:, :, c] = average_blur_gray(img_rgb[:, :, c], ksize)

    return out
````

````python
def gaussian_kernel(ksize, sigma):
    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= kernel.sum()   # normalization

    return kernel
````

````python
def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)

    pad = ksize // 2
    padded_img = np.pad(img, ((pad, pad), (pad, pad)), mode='reflect')

    h, w = img.shape
    out = np.zeros_like(img, dtype=np.float64)

    for i in range(h):
        for j in range(w):
            region = padded_img[i:i+ksize, j:j+ksize]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
def gaussian_blur_rgb(img_rgb, ksize, sigma):
    out = np.zeros_like(img_rgb, dtype=np.float64)

    for c in range(3):
        out[:, :, c] = gaussian_blur_gray(img_rgb[:, :, c], ksize, sigma)

    return out
````

````python
def display_gray_results(img_gray, ksize, sigma):
    avg_blur = average_blur_gray(img_gray, ksize)
    gauss_blur = gaussian_blur_gray(img_gray, ksize, sigma)

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(img_gray, cmap='gray')
    plt.title("Original Grayscale")
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(avg_blur, cmap='gray')
    plt.title(f"Average Blur (ksize={ksize})")
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(gauss_blur, cmap='gray')
    plt.title(f"Gaussian Blur (σ={sigma})")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
````

#### Task - Sobel Edge Detection (NumPy Only)
Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

Compute:
1. Gx
2. Gy
3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python
SOBEL_X = np.array([
    [-1,  0,  1],
    [-2,  0,  2],
    [-1,  0,  1]], dtype=np.float64)

SOBEL_Y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]], dtype=np.float64)
````

````python
def convolve(img, kernel):
    kernel = np.flipud(np.fliplr(kernel))
    k_h, k_w = kernel.shape
    pad_h, pad_w = k_h // 2, k_w // 2
    padded_img = np.pad(
        img,
        ((pad_h, pad_h), (pad_w, pad_w)),
        mode='reflect')
    h, w = img.shape
    out = np.zeros_like(img, dtype=np.float64)

    for i in range(h):
        for j in range(w):
            region = padded_img[i:i+k_h, j:j+k_w]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
def sobel_edges(img_gray):
    Gx = convolve(img_gray, SOBEL_X)
    Gy = convolve(img_gray, SOBEL_Y)
    mag = np.sqrt(Gx**2 + Gy**2)
    mag = mag / mag.max() * 255
    mag = mag.astype(np.uint8)
    return Gx, Gy, mag
````

1. Apply Gaussian blur (from Part 1).
2. Subtract blurred image from original: <code> mask = img - blurred </code>
3. Add scaled mask back: <code>sharp = img + alpha * mask</code> , where you can choose any value of alpha.

````python
def convolve(img, kernel):

    assert img.ndim == 2, "Image must be grayscale (2D)"
    assert kernel.shape[0] == kernel.shape[1], "Kernel must be square"
    assert kernel.shape[0] % 2 == 1, "Kernel size must be odd"

    kernel = np.flipud(np.fliplr(kernel))

    k = kernel.shape[0]
    pad = k // 2

    padded = np.pad(img, ((pad, pad), (pad, pad)), mode='reflect')

    h, w = img.shape
    out = np.zeros((h, w), dtype=np.float64)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out
````

````python
laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  4, -1],
    [ 0, -1,  0]
], dtype=np.float64)
````

````python
def laplacian_sharpen(img, alpha):
    lap = convolve(img, laplacian_kernel)
    sharp = img - alpha * lap
    return lap, sharp
````

````python
def gaussian_kernel(ksize, sigma):
    assert ksize % 2 == 1, "Kernel size must be odd"

    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= kernel.sum()

    return kernel

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)
def unsharp_mask(img, ksize, sigma, alpha):
    blurred = gaussian_blur_gray(img, ksize, sigma)
    mask = img - blurred
    sharp = img + alpha * mask
    return blurred, mask, sharp
````

#### Task - Frequency Analysis (LPF & HPF Verification)
Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
import matplotlib.pyplot as plt
def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude = np.log(1 + np.abs(fshift))
    return magnitude
````

````python

orig = img_gray
blurred = gaussian_blur_gray(img_gray, ksize=9, sigma=2.0)

_, _, sharp_alpha_08 = unsharp_mask(img_gray, ksize=9, sigma=2.0, alpha=0.8)
_, _, sharp_alpha_15 = unsharp_mask(img_gray, ksize=9, sigma=2.0, alpha=1.5)

````

````python
spec_orig = magnitude_spectrum(orig)
spec_blur = magnitude_spectrum(blurred)
spec_sharp_08 = magnitude_spectrum(sharp_alpha_08)
spec_sharp_15 = magnitude_spectrum(sharp_alpha_15)
plt.figure(figsize=(14, 10))

plt.subplot(4, 2, 1)
plt.imshow(orig, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(4, 2, 3)
plt.imshow(blurred, cmap='gray')
plt.title("Gaussian Blurred (LPF)")
plt.axis("off")

plt.subplot(4, 2, 5)
plt.imshow(sharp_alpha_08, cmap='gray')
plt.title("Unsharp Mask (α=0.8)")
plt.axis("off")

plt.subplot(4, 2, 7)
plt.imshow(sharp_alpha_15, cmap='gray')
plt.title("Unsharp Mask (α=1.5)")
plt.axis("off")

plt.subplot(4, 2, 2)
plt.imshow(spec_orig, cmap='gray')
plt.title("Spectrum: Original")
plt.axis("off")

plt.subplot(4, 2, 4)
plt.imshow(spec_blur, cmap='gray')
plt.title("Spectrum: Gaussian Blur")
plt.axis("off")

plt.subplot(4, 2, 6)
plt.imshow(spec_sharp_08, cmap='gray')
plt.title("Spectrum: Sharp (α=0.8)")
plt.axis("off")

plt.subplot(4, 2, 8)
plt.imshow(spec_sharp_15, cmap='gray')
plt.title("Spectrum: Sharp (α=1.5)")
plt.axis("off")

plt.tight_layout()
plt.show()
````

````python
# Center of spectrum → Low frequencies (smooth variations)
# Corners / edges → High frequencies (edges, noise, fine details)
# Bright region → High energy present
# Dark region → Frequency suppressed

# Bright center:
#   → Strong low-frequency components (smooth regions, illumination)
# Visible outer energy:
#   → Presence of edges and textures

# Center becomes brighter and wider:
#   → Low frequencies preserved
# Outer regions become darker:
#   → High frequencies suppressed
#
# Increasing sigma:
#   → Stronger suppression of high frequencies
#   → Even darker corners
#
# Conclusion:
# Gaussian kernel acts as a LOW PASS FILTER (LPF)

# Larger sigma:
#   → Narrower passband
#   → More aggressive low-pass filtering

# Convolution in spatial domain corresponds to multiplication in frequency domain.
# Gaussian kernel suppresses high frequencies → LPF.
# Laplacian / unsharp masking boosts high frequencies → HPF / high-boost filter.
# Frequency spectra visually confirm this behavior.
````

````python
img = img_part5.astype(np.float64)

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
````

````python
yellow_map = R + G - 2 * B
rg_diff = R - G
````

````python
kernel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float64)

kernel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float64)
````

````python
edge_rg_x = convolve(rg_diff, kernel_x)
edge_rg_y = convolve(rg_diff, kernel_y)

edge_yellow_x = convolve(yellow_map, kernel_x)
edge_yellow_y = convolve(yellow_map, kernel_y)
````

````python
edge_rg = np.sqrt(edge_rg_x**2 + edge_rg_y**2)
edge_yellow = np.sqrt(edge_yellow_x**2 + edge_yellow_y**2)

edge_combined = edge_rg + edge_yellow
edge_combined = edge_combined / edge_combined.max() * 255
edge_combined = edge_combined.astype(np.uint8)
binary_edges = np.zeros_like(edge_combined)
binary_edges[edge_combined > 80] = 255
````
