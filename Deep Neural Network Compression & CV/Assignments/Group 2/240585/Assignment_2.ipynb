````python
def RGB_to_HSV(image):

  R = image[:,:,0]
  G = image[:,:,1]
  B = image[:,:,2]

  print("R channel: ",R)
  print("G channel: ",G)
  print("B channel: ",B)

  R_norm = R/255.0
  G_norm = G/255.0
  B_norm = B/255.0

#Finding elementary quantities Cmax, Cmin, delta
  Cmax = np.maximum(np.maximum(R_norm,G_norm),B_norm)
  Cmin = np.minimum(np.minimum(R_norm,G_norm),B_norm)
  delta = Cmax-Cmin
#Initialising the Hue Array
  H = np.zeros_like(Cmax)
#We'll compute Hue only where it is nonzero
  mask_delta_nonzero = delta != 0

  mask_r = (Cmax == R_norm) & mask_delta_nonzero
  mask_g = (Cmax == G_norm) & mask_delta_nonzero
  mask_b = (Cmax == B_norm) & mask_delta_nonzero

#Algorithms for different branch conditions
  H[mask_r] = 60 * ((G_norm - B_norm)[mask_r] / delta[mask_r] % 6)
  H[mask_g] = 60 * (((B_norm - R_norm)[mask_g] / delta[mask_g]) + 2)
  H[mask_b] = 60 * (((R_norm - G_norm)[mask_b] / delta[mask_b]) + 4)

#Computing Saturation
  S = np.zeros_like(Cmax)
  mask_cmax_nonzero = Cmax != 0
  S[mask_cmax_nonzero] = delta[mask_cmax_nonzero] / Cmax[mask_cmax_nonzero]

#Computing Value
  V = Cmax

#Converting to OpenCV convention
  H_OpenCV = np.uint8(H/2)
  S_OpenCV = np.uint8(S*255)
  V_OpenCV = np.uint8(V*255)
  HSV_OpenCV = np.dstack((H_OpenCV,S_OpenCV,V_OpenCV))
  return(HSV_OpenCV)

#Re-conversion
HSV_output = RGB_to_HSV(image)
RGB_test = cv2.cvtColor(HSV_output,cv2.COLOR_HSV2RGB)
plt.imshow(RGB_test)
plt.axis("off")
````

````python
def cyberpunk2077(img):
    #Modifiable Parameters
    alpha = 1.15
    gamma = 0.9
    Shadows = 0.4
    Highlights = 0.6
    sat_index = 1.3

    #1. Conversion of image type to float32 0–1
    img = img.astype(np.float32) / 255.0

    #2. Gamma < 1 brightens the midtones thereby boosting neon signals and reflections, imparting a "moody" night look.
    img_gamma = np.power(img, gamma)

    #3. Boosting Saturation via conversion to HSV form
    hsv = cv2.cvtColor((img_gamma * 255).astype(np.uint8), cv2.COLOR_RGB2HSV)
    h, s, v = cv2.split(hsv)
    s = np.clip(s * sat_index, 0, 255).astype(np.uint8)
    hsv_boost = cv2.merge((h, s, v))
    img_sat = cv2.cvtColor(hsv_boost, cv2.COLOR_HSV2RGB).astype(np.float32) / 255.0

    #4. Luminance-based tint
    L = 0.299 * img_sat[:, :, 0] + 0.587 * img_sat[:, :, 1] + 0.114 * img_sat[:, :, 2]
    teal = np.array([0.0, 0.4, 0.6], dtype=np.float32)
    magenta = np.array([0.6, 0.0, 0.6], dtype=np.float32)
    img_tint = img_sat + (1 - L)[..., None] * teal * Shadows + L[..., None] * magenta * Highlights
    img_tint = np.clip(img_tint, 0, 1)

    #5. Adding Contrast
    img_contrast = np.clip(alpha * (img_tint - 0.5) + 0.5, 0, 1)

    return (img_contrast * 255).astype(np.uint8)


def retrofilm(img):
  #Modifiable Parameters
  sat_index = 0.8
  Shadows = 0.4
  Highlights = 0.9
  alpha = 0.85

  #1. Conversion of image type to float32 0–1
  img = img.astype(np.float32) / 255.0

  #2. Reducing the contrast effect inherent to the picture
  image_faded = np.clip((alpha*(img-0.5)+0.5),0,1)

  #3. Lowering saturation levels to resemble an older camera output
  hsv = cv2.cvtColor((image_faded*255).astype(np.uint8),cv2.COLOR_RGB2HSV)
  h,s,v = cv2.split(hsv)
  s = np.clip(sat_index*s, 0, 255).astype(np.uint8)
  hsv_boost = cv2.merge((h,s,v))
  img_sat = cv2.cvtColor(hsv_boost,cv2.COLOR_HSV2RGB).astype(np.float32) / 255.0

  #4. Luminance-based control of the tint
  L = 0.299*img_sat[:,:,0] + 0.587*img_sat[:,:,1] + 0.114*img_sat[:,:,2]
  teal = np.array([0.0,0.3,0.45], dtype = np.float32)
  yellow = np.array([0.4,0.35,0.0], dtype = np.float32)
  img_tint = np.clip((img_sat + (1-L)[...,None] * teal * Shadows + (L)[...,None] * yellow * Highlights),0,1)

  #5. Adding noise to the picture to resemble that via an old analog camera
  noise = np.random.normal(0.0,0.03,img_tint.shape)
  img_grainy = np.clip(img_tint + noise, 0,1)

  return (img_grainy*255).astype(np.uint8)
````

````python
Value1 = 0.3
Value2 = 0.8

#Utilising the commonly used Gray World Assumption
def gray_world_balance(img):
  b,g,r = cv2.split(img.astype(np.float32))
  b_avg, g_avg, r_avg = np.mean(b),np.mean(g),np.mean(r)
  k = (b_avg + g_avg + r_avg)/3

  b *= k/b_avg
  g *= k/g_avg
  r *= k/r_avg

  final = cv2.merge([np.clip(b,0,255),
                     np.clip(g,0,255),
                     np.clip(r,0,255)])
  return final.astype(np.uint8)

#White Balance
def white_balance(img_rgb, value):
  neutral = gray_world_balance(img_rgb)

  cold_tint = np.array([0.85, 0.92, 1.15])
  warm_tint = np.array([1.15, 1.05, 0.85])

  cold_img = neutral * cold_tint
  warm_img = neutral * warm_tint

  if value < 0.5:
    t = value/0.5
    img = (1-t)*cold_img + t*neutral

  else:
    t = (value - 0.5) / 0.5
    img = (1-t)*neutral + t*warm_img

  return(np.clip(img,0,255).astype(np.uint8))

#Final Output
def show_white_balance(value):

    filtered = white_balance(image,value)

    plt.subplot(1,2,1)
    plt.imshow(image)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(filtered)
    plt.title(f"White Balance = {value:.2f}")
    plt.axis("off")

#When the slider is at value 0.3
show_white_balance(Value1)
````
