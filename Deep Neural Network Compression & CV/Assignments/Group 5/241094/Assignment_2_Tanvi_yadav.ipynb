````python
import numpy as np

def gray_histogram(gray_img):
    hist = np.zeros(256, dtype=np.int32)

    for i in range(gray_img.shape[0]):
        for j in range(gray_img.shape[1]):
            pixel = gray_img[i, j]
            hist[pixel] += 1

    return hist
````

Write a function <code>rgb_to_hsv</code> that takes a 0-255 ( not normalised ) RGB image and returns an HSV image ( also not normalised ).

````python
# Your PART B codes follow here. You can create more cells under this. Write clean and commented code.

import numpy as np
import cv2
import matplotlib.pyplot as plt

def rgb_to_hsv(img_rgb):
  rgb = img_rgb.astype(np.float32) # convert to float for maths operations.

  # split channels
  r = rgb[..., 0]
  g = rgb[..., 1]
  b = rgb[..., 2]


  cmax = np.maximum(np.maximum(r, g), b)   # for each pixel,maximum of R,G,B
  cmin = np.minimum(np.minimum(r, g), b)   # for each pixel,minimum of R,G,B
  delta = cmax - cmin  # delta shows how colorful the pixel is.

  H = np.zeros_like(cmax, dtype=np.float32)
  nz = delta != 0.0   # condition that delta != 0

  mask_r = (cmax == r) & nz
  mask_g = (cmax == g) & nz
  mask_b = (cmax == b) & nz

  val_r = ((g - b) / delta)[mask_r]
  H[mask_r] = 60.0 * (np.mod(val_r, 6.0)) #compute Hue in degrees for pixels where Red is max.

  val_b = ((r - g) / delta)[mask_b]
  H[mask_b] = 60.0 * (val_b + 4.0)
  H[mask_g] = 60.0 * (((b - r)/delta)[mask_g] + 2.0)

  # take Hue in degrees and scale for OpenCV
  H_opencv = H / 2.0
  H_opencv = np.mod(H_opencv, 180.0)

  # saturation calculation
  S = np.zeros_like(cmax, dtype=np.float32)
  nonzero_cmax = cmax != 0.0
  S[nonzero_cmax] = (delta[nonzero_cmax] / cmax[nonzero_cmax]) # scale to 0..255
  S_opencv = S * 255.0 # scale S from 0..1 to OpenCV range 0..255.


  V_opencv = cmax.copy()   # already 0..255 range because we kept R,G,B in 0..255.

  # stack into HSV final image and convert to uint8
  hsv = np.stack([H_opencv, S_opencv, V_opencv], axis=-1)
  hsv_uint8 = np.clip(hsv, 0, 255).astype(np.uint8)

  return hsv_uint8
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt
````

````python
def Yadavsky_enhancer(img_rgb, contrast=1.08, brightness=5, sat_boost=1.15, blue_tint=25, gamma=0.95):
  img = img_rgb.astype(np.float32)
  img = img * contrast + brightness
  img_norm = np.clip(img / 255.0, 0, 1)
  img_gamma = np.power(img_norm, gamma)
  img = img_gamma * 255.0
  hsv = cv2.cvtColor(img.astype(np.uint8), cv2.COLOR_RGB2HSV).astype(np.float32)
  H, S, V = hsv[...,0], hsv[...,1], hsv[...,2]
  S = S * sat_boost
  S = np.clip(S, 0, 255)

  hsv_mod = np.stack([H, S, V], axis=-1)
  img_sat = cv2.cvtColor(hsv_mod.astype(np.uint8), cv2.COLOR_HSV2RGB).astype(np.float32)

  R = img_sat[...,0]
  G = img_sat[...,1]
  B = img_sat[...,2]

  B = B + blue_tint
  R = R * 0.98
  G = G * 0.99

  out = np.stack([R, G, B], axis=-1)
  out = np.clip(out, 0, 255).astype(np.uint8)
  return out

````

````python
import numpy as np
import cv2

def gold_ki_layer(img_rgb,
                  contrast=1.08,
                  brightness=6.0,
                  vibrance=1.12,
                  highlight_warm=10.0,
                  shadow_cool=6.0,
                  unsharp_strength=0.6,   # 0..1
                  vignette_strength=0.35):
   img = img_rgb.astype(np.float32)
   img = img * contrast + brightness
   img = np.clip(img, 0.0, 255.0)  #np.clip keeps values inside valid range 0..255 so nothing overflows.

   hsv = cv2.cvtColor(img.astype(np.uint8), cv2.COLOR_RGB2HSV).astype(np.float32)
   H = hsv[..., 0]; S = hsv[..., 1]; V = hsv[..., 2] # seperating the channels
   sat_factor = 1.0 + (vibrance - 1.0) * (1.0 - (S / 255.0))
   S = S * sat_factor
   S = np.clip(S, 0.0, 255.0)

   hsv2 = np.stack([H, S, V], axis=-1).astype(np.float32)
   img_vib = cv2.cvtColor(hsv2.astype(np.uint8), cv2.COLOR_HSV2RGB).astype(np.float32)

   R = img_vib[..., 0]
   G = img_vib[..., 1]
   B = img_vib[..., 2]

   # luminance
   """ we compute lum as a brightness map normalized to 0 to 1. we use this to decide which pixels are shadow vs which are highlight."""
   lum = (0.299 * R + 0.587 * G + 0.114 * B) / 255.0
   lum = np.clip(lum, 0.0, 1.0)

   shadow_mask = (1.0 - lum) ** 1.4   # strong where lum is low
   highlight_mask = (lum ** 2.0)      # strong where lum is high

   # apply warm gold to highlights
   R = R + highlight_warm * 1.0 * highlight_mask
   G = G + highlight_warm * 0.45 * highlight_mask

   combined = np.stack([R, G, B], axis=-1).astype(np.float32)
   combined = np.clip(combined, 0.0, 255.0)

   h, w = combined.shape[:2]

   # create normalized coordinate grids -1..1
   y = np.linspace(-1.0, 1.0, h)[:, None]
   x = np.linspace(-1.0, 1.0, w)[None, :]
   radius = np.sqrt(x**2 + y**2)

   # vignette_mask mixes the radial falloff with a base value so corners are softly dimmed (not fully black).
   vignette = 1.0 - (radius / np.sqrt(2.0))
   vignette = np.clip(vignette, 0.0, 1.0)
   vignette_mask = (vignette * (1.0 - vignette_strength)) + vignette_strength * 0.7


   # apply per-channel
   combined[..., 0] *= vignette_mask
   combined[..., 1] *= vignette_mask
   combined[..., 2] *= vignette_mask


   out = np.clip(combined, 0.0, 255.0).astype(np.uint8)
   return out
````

Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image.
````python
""" The White Patch method assumes the brightest pixel in each color channel
should represent pure white (255).So, each channel is scaled so its maximum
value becomes 255, removing color casts in the image. """

def white_patch_balance(img_rgb):
    img = img_rgb.astype(np.float32)

    # Separate channels
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]

    # Find max value in each channel
    Rmax = np.max(R)
    Gmax = np.max(G)
    Bmax = np.max(B)

    # to prevent divide-by-zero
    if Rmax == 0: Rmax = 1
    if Gmax == 0: Gmax = 1
    if Bmax == 0: Bmax = 1

    # scale factors
    SR = 255.0 / Rmax
    SG = 255.0 / Gmax
    SB = 255.0 / Bmax

    # Apply scaling
    R_bal = R * SR
    G_bal = G * SG
    B_bal = B * SB

    # Stack back and clip
    out = np.stack([R_bal, G_bal, B_bal], axis=-1)
    out = np.clip(out, 0, 255).astype(np.uint8)

    return out
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def white_balance(img_rgb):
    """Simple white-patch balance helper (not the slider)."""

    img = img_rgb.astype(np.float32)
    Rmax = max(np.max(img[...,0]), 1.0)
    Gmax = max(np.max(img[...,1]), 1.0)
    Bmax = max(np.max(img[...,2]), 1.0)
    SR, SG, SB = 255.0/Rmax, 255.0/Gmax, 255.0/Bmax
    out = np.empty_like(img, dtype=np.float32)
    out[...,0] = img[...,0] * SR
    out[...,1] = img[...,1] * SG
    out[...,2] = img[...,2] * SB
    return np.clip(out,0,255).astype(np.uint8)

def white_balance_slider(img_rgb, value):

    if not (0.0 <= value <= 1.0):
        raise ValueError("value must be in [0,1]")

    # Convert to float for math
    img = img_rgb.astype(np.float32)

    # compute original luminance mean (used later to preserve brightness)
    lum_orig = (0.299*img[...,0] + 0.587*img[...,1] + 0.114*img[...,2])
    mean_lum_orig = np.maximum(np.mean(lum_orig), 1e-6)

    # define symmetric delta (how much channels change from neutral)
    # Positive delta means warm: boost R slightly, small boost to G, lower B.
    delta = np.array([0.12, 0.03, -0.12], dtype=np.float32)

    # cold gains = 1 - delta; warm gains = 1 + delta
    cold_gains = 1.0 - delta
    warm_gains = 1.0 + delta

    # linear interpolation parameter (alpha)
    alpha = np.clip(value, 0.0, 1.0)

    # compute per-channel gains so that alpha=0.5 => gains = 1.0
    gains = (1.0 - alpha) * cold_gains + alpha * warm_gains
    # shape (3,) -> broadcastable to image

    # Apply gains per channel
    out = np.empty_like(img, dtype=np.float32)
    out[...,0] = img[...,0] * gains[0]
    out[...,1] = img[...,1] * gains[1]
    out[...,2] = img[...,2] * gains[2]


    V = lum_orig / 255.0              # 0..1
    highlight_weight = np.clip( (V - 0.4) / 0.6, 0.0, 1.0 )  # smooth ramp 0.4..1.0
    # blend between original pixel and gained pixel according to highlight weight and slider strength
    blend_strength = np.abs(alpha - 0.5) * 2.0  # 0 at 0.5, 1 at 0 or 1
    out = img * (1.0 - (highlight_weight * blend_strength)[...,None]) + \
          out * (highlight_weight * blend_strength)[...,None]

    # Preserve global luminance: scale so mean luminance matches original
    lum_out = (0.299*out[...,0] + 0.587*out[...,1] + 0.114*out[...,2])
    mean_lum_out = np.maximum(np.mean(lum_out), 1e-6)
    lum_scale = mean_lum_orig / mean_lum_out
    out = out * lum_scale

    # Final clip and convert
    out = np.clip(out, 0.0, 255.0).astype(np.uint8)
    return out

# ----------------------------
# Demo: show original, cold, neutral, warm
# ----------------------------
# replace img_np with your loaded image variable name
try:
    img_example = img_np  # your currently loaded image
except NameError:
    # fallback: create a test image
    img_example = cv2.cvtColor(cv2.imread('test.jpg'), cv2.COLOR_BGR2RGB)

# produce three example outputs
out_coldest = white_balance_slider(img_example, 0.0)
out_cold = white_balance_slider(img_example, 0.25)
out_neutral = white_balance_slider(img_example, 0.5)
out_warm = white_balance_slider(img_example, 0.75)
out_warmest = white_balance_slider(img_example, 1.0)

plt.figure(figsize=(16,8))
plt.subplot(2,3,1); plt.imshow(out_coldest); plt.title("0.0 - Coldest"); plt.axis('off')
plt.subplot(2,3,2); plt.imshow(out_cold); plt.title("0.25 - Cool"); plt.axis('off')
plt.subplot(2,3,3); plt.imshow(out_neutral); plt.title("0.5 - Original"); plt.axis('off')
plt.subplot(2,3,4); plt.imshow(out_warm); plt.title("0.75 - Warm"); plt.axis('off')
plt.subplot(2,3,5); plt.imshow(out_warmest); plt.title("1.0 - Warmest"); plt.axis('off')

plt.show()
````

Values below 0.5 shift the image toward blue/cool tones, and values above 0.5 shift it toward warm/orange tones.
Intermediate values (0.25, 0.75) give subtle adjustments, while 0.0 and 1.0 give extreme cold/warm effects.

````python
""" INTERACTIVE SLIDER """

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import ipywidgets as widgets
import cv2
import io
from PIL import Image

# convert image (numpy uint8 RGB) to downloadable bytes (PNG)
def img_to_bytes(img_rgb):
    pil = Image.fromarray(img_rgb)
    buf = io.BytesIO()
    pil.save(buf, format='PNG')
    buf.seek(0)
    return buf

# Create the slider widget and preset buttons
slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.01,
                             description='WB value', continuous_update=True, layout=widgets.Layout(width='70%'))

btn_0 = widgets.Button(description='0.0 (Cold)', layout=widgets.Layout(width='120px'))
btn_25 = widgets.Button(description='0.25', layout=widgets.Layout(width='80px'))
btn_50 = widgets.Button(description='0.5 (Neutral)', layout=widgets.Layout(width='120px'))
btn_75 = widgets.Button(description='0.75', layout=widgets.Layout(width='80px'))
btn_100 = widgets.Button(description='1.0 (Warm)', layout=widgets.Layout(width='120px'))

download_btn = widgets.Button(description='Download Result', button_style='success', layout=widgets.Layout(width='160px'))

# The image variable to use (make sure img_np exists in your notebook)
try:
    img_example = img_np  # your loaded RGB uint8 image
except NameError:
    raise RuntimeError("Please load your image into variable `img_np` (RGB uint8) and define white_balance_slider before running this cell.")

# Display area
out_area = widgets.Output(layout=widgets.Layout(border='1px solid gray'))

def render_and_display(value):
    """Run white balance and draw original + result."""
    # compute processed image
    processed = white_balance_slider(img_example, float(value))

    # show side-by-side using matplotlib inside the output widget
    with out_area:
        clear_output(wait=True)
        fig, axes = plt.subplots(1,2, figsize=(12,6))
        axes[0].imshow(img_example)
        axes[0].set_title("Original")
        axes[0].axis('off')
        axes[1].imshow(processed)
        axes[1].set_title(f"White-balanced (value={value:.2f})")
        axes[1].axis('off')
        plt.tight_layout()
        plt.show()
    return processed

# initial render
_current_result = render_and_display(slider.value)

# slider callback
def on_slider_change(change):
    global _current_result
    if change['name'] == 'value':
        _current_result = render_and_display(change['new'])

slider.observe(on_slider_change)

# preset button callbacks
def set_value_and_render(v):
    slider.value = v  # this triggers the slider observer and updates display

btn_0.on_click(lambda _: set_value_and_render(0.0))
btn_25.on_click(lambda _: set_value_and_render(0.25))
btn_50.on_click(lambda _: set_value_and_render(0.5))
btn_75.on_click(lambda _: set_value_and_render(0.75))
btn_100.on_click(lambda _: set_value_and_render(1.0))

# download callback
def on_download_clicked(_):
    global _current_result
    buf = img_to_bytes(_current_result)
    # create a downloadable link widget
    b = widgets.FileLink(value=buf, filename='wb_result.png', result_html_prefix="Download: ")
    with out_area:
        print("If running in Colab, right-click the image and choose 'Save image as...' to download.")
        display(Image.open(buf))

download_btn.on_click(on_download_clicked)

# Layout widgets
preset_row = widgets.HBox([btn_0, btn_25, btn_50, btn_75, btn_100, download_btn])
ui = widgets.VBox([slider, preset_row, out_area])

display(ui)
````
