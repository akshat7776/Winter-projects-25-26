````python
# Your PART C codes follow here. You can create more cells under this. Write clean and commented code.

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2

def create_lpf_mask(radius):
    mask = np.zeros((rows, cols),np.uint8)

    for i in range(rows):
        for j in range(cols):
            if (i - crow)**2 + (j - ccol)**2 <= radius**2:
                mask[i, j] = 1
    return mask
````

````python
mask10 = create_lpf_mask(10)
mask30 = create_lpf_mask(30)
mask80 = create_lpf_mask(80)

f10 = fshift * mask10
f30 = fshift * mask30
f80 = fshift * mask80

img10 = np.abs(np.fft.ifft2(np.fft.ifftshift(f10)))
img30 = np.abs(np.fft.ifft2(np.fft.ifftshift(f30)))
img80 = np.abs(np.fft.ifft2(np.fft.ifftshift(f80)))

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img10, cmap='gray')
plt.title("LPF radius 10")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img30, cmap='gray')
plt.title("LPF radius 30")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(img80, cmap='gray')
plt.title("LPF radius 80")
plt.axis("off")

plt.show()

# to see circles

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(mask10, cmap='gray')
plt.title("LPF Mask (radius 10)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(mask30, cmap='gray')
plt.title("LPF Mask (radius 30)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(mask80, cmap='gray')
plt.title("LPF Mask (radius 80)")
plt.axis("off")

plt.show()

````

This is low pass filter so radius decides how many frequency we can keep. if we have small radius, it means only very low frequencies so removes most high frequency details like edge or texture and hence image is more blurred.

````python
mask_hpf2 = 1 - mask10
f_hpf2 = fshift * mask_hpf2
img_hpf2 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_hpf2)))

plt.imshow(img_hpf2, cmap='gray')
plt.title("HPF result radius 10")
plt.axis("off")
plt.show()


mask_hpf1 = 1 - mask30
f_hpf1 = fshift * mask_hpf1
img_hpf1 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_hpf1)))

plt.imshow(img_hpf1, cmap='gray')
plt.title("HPF result radius 30")
plt.axis("off")
plt.show()


mask_hpf3 = 1 - mask80
f_hpf3 = fshift * mask_hpf3
img_hpf3 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_hpf3)))

plt.imshow(img_hpf3, cmap='gray')
plt.title("HPF result radius 80")
plt.axis("off")
plt.show()


# to see circles
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(mask_hpf2, cmap='gray')
plt.title("HPF Mask (radius 10)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(mask_hpf1, cmap='gray')
plt.title("HPF Mask (radius 30)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(mask_hpf3, cmap='gray')
plt.title("HPF Mask (radius 80)")
plt.axis("off")

plt.show()


# this is HPF so here centre area becomes 0 means smooth components (low frequency) are removed and outer area becomes one means edges(high frequency) are kept.
# reconstructed images will show only edges and texture, not smooth background.
````

````python
#CREATING THE SQUARE MASK
side_length = 50
rows, cols = gray.shape
crow, ccol = rows//2, cols//2
mask_lpf_square = np.zeros((rows, cols), np.uint8)

# coordinates
row_start = crow - side_length // 2
row_end = crow + side_length // 2
col_start = ccol - side_length // 2
col_end = ccol + side_length // 2

mask_lpf_square[row_start:row_end, col_start:col_end] = 1

plt.figure(figsize=(5,5))
plt.imshow(mask_lpf_square, cmap='gray')
plt.title('Square LPF Mask (Side ' + str(side_length) + ')')
plt.axis('off')
plt.show()

f_lpf_square = fshift * mask_lpf_square

ishift_lpf_square = np.fft.ifftshift(f_lpf_square)
img_back_lpf_square = np.fft.ifft2(ishift_lpf_square)
img_back_lpf_square = np.abs(img_back_lpf_square)

plt.figure(figsize=(5,5))
plt.imshow(img_back_lpf_square, cmap='gray')
plt.title('Image with Square LPF (Side ' + str(side_length) + ')')
plt.axis('off')
plt.show()
````

Explanation:- In frequency domain, FFT splits image into two components:-
1. Magnitude :- which controls intensity, contrast and texture strength.
2. Phase:- which controls edges, shapes etc.

so when we combined magnitude of image 1 and phase of image 2 , we get edges and shape same as imag 2 but with background, texture like image 1 and similarly for other case.

Hence we can say that phase defines structure and magnitude decides texture.
