````python
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np

# PASTE BASE64 STRING HERE
base64_image = """ """
# long base64 string

img_data = base64.b64decode(base64_image)
img = Image.open(io.BytesIO(img_data))

plt.imshow(img)
plt.axis("off")
````

````python
img_rgb=np.array(img)
plt.figure()
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
````

````python
img_rgb.shape
````

````python
img_gray = (0.299 * img_rgb[:, :, 0] + 0.587 * img_rgb[:, :, 1] + 0.114 * img_rgb[:, :, 2])
img_gray = img_gray.astype(np.uint8)
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale (NumPy Weighted Sum)")
plt.axis("off")
plt.show()
````

````python
import cv2
gray_cv2=cv2.cvtColor(img_rgb,cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale CV2")
plt.axis("off")
plt.show()
````

````python
diff = cv2.absdiff(img_gray, gray_cv2)
print("Max absolute difference (NumPy vs cvtColor):", diff.max())
print("Mean absolute difference (NumPy vs cvtColor):", diff.mean())
````
