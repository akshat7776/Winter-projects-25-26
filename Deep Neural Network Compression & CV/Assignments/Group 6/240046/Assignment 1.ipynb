````python
import os

out_dir = '/mnt/data/lpf_hpf_outputs' # Define an output directory
os.makedirs(out_dir, exist_ok=True) # Create the directory if it doesn't exist

gray = gray_numpy
h,w = gray.shape
print("Grayscale:", gray.shape)

#FFT
F = np.fft.fft2(gray_numpy)
fshift = np.fft.fftshift(F)

# mask factory
def circular_mask(shape, radius):
    H, W = shape
    cy, cx = H//2, W//2
    Y, X = np.ogrid[:H, :W]
    dist2 = (Y-cy)**2 + (X-cx)**2
    return (dist2 <= radius**2).astype(np.float32)

# Define radii for LPF masks
radii = [10, 30, 80]

# clip radii to avoid oversized
max_r = int(min(h,w)/2)
radii = [min(r, max_r) for r in radii]


recons = []
for r in radii:
    mask = circular_mask((h,w), r)
    # reason: smaller radius keeps fewer low-frequency components => stronger blur
    masked = Fshift * mask
    inv = np.fft.ifftshift(masked)
    recon = np.fft.ifft2(inv)
    recon = np.real(recon)
    recon_u8 = np.clip(np.round(recon), 0, 255).astype(np.uint8)
    recons.append((r, mask, recon_u8))
    Image.fromarray((mask*255).astype(np.uint8)).save(os.path.join(out_dir, f"lpf_mask_r{r}.png"))
    Image.fromarray(recon_u8).save(os.path.join(out_dir, f"recon_lpf_r{r}.png"))


# display
plt.figure(figsize=(14,4))
plt.subplot(1, len(recons)+1, 1)
plt.imshow(gray, cmap='gray'); plt.axis('off'); plt.title('Original')
for i, (r, mask, recon) in enumerate(recons, start=2):
    plt.subplot(1, len(recons)+1, i)
    plt.imshow(recon, cmap='gray'); plt.axis('off'); plt.title(f'LPF r={r}')
plt.suptitle('LPF reconstructions (smaller radius => stronger blur)')
plt.show()
````

````python
n = len(radii)
plt.figure(figsize=(4*(n+1), 4))
plt_idx = 1

plt.subplot(1, n+1, plt_idx)
plt.imshow(gray, cmap='gray', vmin=0, vmax=255)
plt.axis('off'); plt.title("Original")
plt_idx += 1

for r_in in radii:
    r = int(min(r_in, max_r))
    LPF = circular_mask((h,w), r)
    HPF = 1.0 - LPF

    # Apply HPF in freq domain
    Fmasked = Fshift * HPF

    inv_shift = np.fft.ifftshift(Fmasked)
    recon_complex = np.fft.ifft2(inv_shift)
    recon = np.real(recon_complex)
    recon_u8 = np.clip(np.round(recon), 0, 255).astype(np.uint8)

    plt.subplot(1, n+1, plt_idx)
    plt.imshow(recon_u8, cmap='gray', vmin=0, vmax=255)
    plt.axis('off')
    plt.title(f'HPF (r={r}) ')
    plt_idx += 1
````

````python
def square_mask(shape, side):
    H, W = shape
    cy, cx = H//2, W//2
    half = side//2
    mask = np.zeros((H,W), dtype=np.float32)
    y0 = max(0, cy-half); y1 = min(H, cy+half + (1 if side%2 else 0))
    x0 = max(0, cx-half); x1 = min(W, cx+half + (1 if side%2 else 0))
    mask[y0:y1, x0:x1] = 1.0
    return mask

square_side = 50
mask = square_mask((h,w), square_side)

masked = Fshift * mask
inv = np.fft.ifftshift(masked)
recon = np.fft.ifft2(inv)
recon = np.real(recon)
recon_u8 = np.clip(np.round(recon),0,255).astype(np.uint8)


plt.figure(figsize=(10,4))
plt.subplot(1,2,1); plt.imshow(recon_u8, cmap='gray'); plt.axis('off'); plt.title(f'Square LPF recon (side={square_side})')
plt.subplot(1,2,2); plt.imshow((mask*255).astype(np.uint8), cmap='gray'); plt.axis('off'); plt.title('Square LPF mask (center)')
plt.show()
````

````python
# Magnitude and Phase
mag = np.abs(Fshift)
phase = np.angle(Fshift)
eps = 1e-12


# (A) TRUE PHASE-ONLY REPRESENTATION
# phase_only = Fshift / |Fshift|
#
# IMPORTANT DIFFERENCE FROM np.angle(Fshift):
# - This line produces a COMPLEX array where:
#       magnitude = 1
#       phase     = original phase
#   i.e., phase_only = e^{j * θ}
#
# - np.angle(Fshift) gives ONLY θ (real numbers),
#   NOT a usable complex spectrum.
#   You CANNOT inverse-FFT angles directly.
#
phase_only = Fshift / (mag + eps)

# Inverse FFT
recon_phase = np.real(np.fft.ifft2(np.fft.ifftshift(phase_only)))

# Normalize for display
p = recon_phase - recon_phase.min()
if p.max() != 0:
    p = p / p.max()
recon_phase_u8 = (p * 255).astype(np.uint8)



# (B) RECONSTRUCTION USING np.angle(Fshift)
# np.angle(Fshift) returns ONLY the phase values θ (real array),
# but NOT the complex representation e^{jθ}.
#
# Therefore, reconstruction requires:
#       exp(1j * np.angle(Fshift))
#
# If you incorrectly do IFFT(np.angle(Fshift)), you get garbage,
# because IFFT expects COMPLEX frequencies—not raw angles.
#
phase_exp = np.exp(1j * phase)
recon_from_angle = np.real(np.fft.ifft2(np.fft.ifftshift(phase_exp)))

# Normalize
q = recon_from_angle - recon_from_angle.min()
if q.max() != 0:
    q = q / q.max()
recon_from_angle_u8 = (q * 255).astype(np.uint8)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray'); plt.axis('off'); plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(recon_phase_u8, cmap='gray'); plt.axis('off')
plt.title("Phase-Only Reconstruction\n(fshift / |fshift|)")

plt.subplot(1,3,3)
plt.imshow(recon_from_angle_u8, cmap='gray'); plt.axis('off')
plt.title("Reconstruction using exp(1j * angle)\n(Equivalent to phase-only)")

plt.show()
````

````python
# magnitude only (drop phase)
mag_only = np.abs(Fshift)

# magnitude-only reconstruction (zero-phase)
recon_mag = np.real(np.fft.ifft2(np.fft.ifftshift(mag_only)))

# Magnitude-only reconstructions have extremely low contrast,
# because removing phase destroys almost all spatial structure.
r = recon_mag - recon_mag.min()
if r.max() != 0:
    r = r / r.max()

mag_only_visible = (r ** 0.3 * 255).astype(np.uint8)

# Magnitude Spectrum (log-scaled)
# Log is used because FFT magnitudes have *huge* dynamic range:
# DC term is extremely large, high frequencies are tiny.
# Without logarithm, the image looks black except the center.
mag_log = np.log1p(mag_only)
m = mag_log - mag_log.min()
m = (m / (m.max() + 1e-12) * 255).astype(np.uint8)

# Display
plt.figure(figsize=(14,4))

plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray'); plt.axis('off')
plt.title("Original grayscale")

plt.subplot(1,3,2)
plt.imshow(m, cmap='gray'); plt.axis('off')
plt.title("Magnitude Spectrum (log-scaled)\n(log makes details visible)")

plt.subplot(1,3,3)
plt.imshow(mag_only_visible, cmap='gray'); plt.axis('off')
plt.title("Magnitude-only reconstruction\n(heavily contrast-enhanced)")

plt.show()

# • Magnitude-only reconstruction looks nothing like the original because
#   almost ALL spatial/structural information (edges, shapes, positions)
#   lives in the phase — not in the magnitude.
#
# • Raw magnitude-only output has near-zero contrast, which is why
#   the image originally appeared completely black.
#
# • Logarithm is used in plotting Magnitude Spectrum because the FFT magnitude
#   spans several orders of magnitude. Without log compression,
#   the display becomes dominated by DC and all details vanish.

````

````python
def to_gray_uint8(rgb):
    arr = np.asarray(rgb).astype(np.float32)
    w = np.array([0.2989, 0.5870, 0.1140], dtype=np.float32)
    g = (arr * w.reshape((1,1,3))).sum(axis=2)
    return np.clip(np.round(g), 0, 255).astype(np.uint8)

g1 = to_gray_uint8(img)
g2 = to_gray_uint8(img2)

# resize g2 to g1 if shapes differ
if g2.shape != g1.shape:
    g2 = np.array(Image.fromarray(g2).resize((g1.shape[1], g1.shape[0]), resample=Image.BICUBIC)).astype(np.uint8)

# FFTs (centered)
F1 = np.fft.fftshift(np.fft.fft2(g1))
F2 = np.fft.fftshift(np.fft.fft2(g2))

# magnitudes and phases
mag1 = np.abs(F1)
mag2 = np.abs(F2)
phase1 = np.angle(F1)
phase2 = np.angle(F2)

# swap: |F1|*exp(i*phase2) and |F2|*exp(i*phase1)
F1_swapped = mag1 * np.exp(1j * phase2)
F2_swapped = mag2 * np.exp(1j * phase1)

# inverse and reconstruct
rec1 = np.real(np.fft.ifft2(np.fft.ifftshift(F1_swapped)))
rec2 = np.real(np.fft.ifft2(np.fft.ifftshift(F2_swapped)))

# robust normalization for display
def norm_disp(img, pmin=1, pmax=99, gamma=1.0):
    lo = np.percentile(img, pmin)
    hi = np.percentile(img, pmax)
    out = np.clip((img - lo) / (hi - lo + 1e-12), 0, 1)
    if gamma != 1.0:
        out = out ** (1.0/gamma)
    return (out * 255).astype(np.uint8)

g1_vis = g1
g2_vis = g2
rec1_vis = norm_disp(rec1, pmin=1, pmax=99, gamma=0.9)
rec2_vis = norm_disp(rec2, pmin=1, pmax=99, gamma=0.9)

# display originals and swapped reconstructions
plt.figure(figsize=(12,8))
plt.subplot(2,2,1); plt.imshow(g1_vis, cmap='gray'); plt.axis('off'); plt.title('Original Image (img)')
plt.subplot(2,2,2); plt.imshow(g2_vis, cmap='gray'); plt.axis('off'); plt.title('Original Image (img2)')
plt.subplot(2,2,3); plt.imshow(rec1_vis, cmap='gray'); plt.axis('off'); plt.title('|F(img)| + phase(img2)')
plt.subplot(2,2,4); plt.imshow(rec2_vis, cmap='gray'); plt.axis('off'); plt.title('|F(img2)| + phase(img)')
plt.tight_layout()
plt.show()
````
