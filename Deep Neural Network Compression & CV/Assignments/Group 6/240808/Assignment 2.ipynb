````python
import numpy as np
import matplotlib.pyplot as plt
import cv2
# img_rgb is in RGB mode
# img_rgb_R=img_rgb[:,:,0]
#img_rgb_G=img_rgb[:,:,1]
#img_rgb_B=img_rgb[:,:,2]

def rgb_to_hsv(img_rgb):
    img_rgb=img_rgb.astype(np.float32)
  # img_rgb 3d array spliting into 3 channels of 2d array
    R=img_rgb[:,:,0]
    G=img_rgb[:,:,1]
    B=img_rgb[:,:,2]


    rows,cols,ch=img_rgb.shape       # shape= (H,W,3)
    max_intensity=np.maximum(np.maximum(R,G),B)  # 2d array of maximum intensity per pixel in[0-1]
    min_intensity=np.minimum(np.minimum(R,G),B)   # 2d array of minimum intensity per pixel in [0-1]
    difference=max_intensity-min_intensity  # 2d array of difference per pixel
    # making 2d  empty arrays for H,S,V
    H=np.zeros((rows,cols),dtype=np.float32)
    S=np.zeros((rows,cols),dtype=np.float32)
    V=np.zeros((rows,cols),dtype=np.float32)
    for i in range(rows):
      for j in range(cols):
        V[i,j]=max_intensity[i,j]


        if difference[i,j]==0:
            H[i,j]=0  # Hue is undefined
            S[i,j]=0  # each value in pixel vector is same so color is gray

        else:
           if max_intensity[i,j]==R[i,j]:  # color is in red region
              H[i,j]=60*(((G[i,j]-B[i,j])/difference[i,j])%6) # in 0-60 or 300-360 degree
           elif max_intensity[i,j]==G[i,j]: # color is in green region
              H[i,j]=60*(((B[i,j]-R[i,j])/difference[i,j])+2)  # around 120 degree

           else:   # color is in blue region
              H[i,j]=60*(((R[i,j]-G[i,j])/difference[i,j])+4)  # around 240 degree

       # H[i,j]=H[i,j]*179/360
       # H[i,j]=np.clip(H[i,j],0,179)
        #H[i,j]=H[i,j]%180
       # H[i,j]=np.uint8(H[i,j])

        if max_intensity[i,j]==0:
            S[i,j]=0  # RGB value will be [0,0,0]
        else:
            S[i,j]=(difference[i,j]/max_intensity[i,j])*255 # how R,G,B values varies from each other

    # converting degree in open cv units from 0to 179
    H=H*179/360
     # clipping the values and converting H,S,V to uint8
    H=np.clip(H,0,179).astype(np.uint8)
    S=np.clip(S,0,255).astype(np.uint8)
    V=np.clip(V,0,255).astype(np.uint8)

   # merging back to get back HSV image
    hsv=np.dstack((H,S,V))
    return hsv
# creating hsv image from rgb image
img_hsv=rgb_to_hsv(img_rgb)
# converting hsv image to back rgb image
img_rgb_new=cv2.cvtColor(img_hsv,cv2.COLOR_HSV2RGB).astype(np.uint8)


 # display


plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("img_rgb")


plt.subplot(1,2,2)
plt.imshow(img_rgb_new)
plt.title("img_rgb_new")
plt.axis("off")
plt.show()




````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt
img1_rgb=img1.convert("RGB")
img1_rgb=np.array(img1_rgb)

def white_patch_balance(img_rgb):
    # converting into float
    img_f=img_rgb.astype(np.float32)
    rows,cols,ch=img_rgb.shape
    # seperating channels
    R=img_f[:,:,0]
    G=img_f[:,:,1]
    B=img_f[:,:,2]
    # finding brightest pixel
    Brightness=R+G+B  # 2D array with Brigthness[i,j]=R[i,j]+B[i,j]+G[i,j]
    max_value=-1
    max_rows=0
    max_cols=0
    for i in range(rows):
      for j in range(cols):
         if Brightness[i,j]>max_value:
              max_value=Brightness[i,j]
              max_rows=i
              max_cols=j

    # brightest pixel =[R[max_rows,max_cols],G[max_rows,max_cols],B[max_rows,max_cols]]
    # scaling to make brightest pixel=[255,255,255]
    if R[max_rows,max_cols]==0:
         scale_R=0
    else:
        scale_R=255/R[max_rows,max_cols]
    if G[max_rows,max_cols]==0:
        scale_G=0
    else:
        scale_G=255/G[max_rows,max_cols]
    if B[max_rows,max_cols]==0:
         scale_B=0
    else:
        scale_B=255/B[max_rows,max_cols]
    # adjusting other R,G,B values by scaling
    R_new=R*scale_R
    G_new=G*scale_G
    B_new=B*scale_B
    # merging back to get balanced image
    img_rgb_balanced=np.dstack((R_new,G_new,B_new))
    # clipping to [0-255]
    img_rgb_balanced=np.clip(img_rgb_balanced,[0,0,0],[255,255,255]).astype(np.uint8)
    return img_rgb_balanced


 # balancing a image
img_rgb_balanced =white_patch_balance(img1_rgb)
 # display
plt.subplot(1,2,1)
plt.imshow(img1_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_rgb_balanced)
plt.title("Balanced")
plt.axis("off")
plt.show()



````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt
def white_balance(img_rgb,value):
     #r,g,b =cv2.spilt(img_rgb.astype(np.float32))
     # converting into float
     img_f=img_rgb.astype(np.float32)
     #sepreating channels
     r=img_f[:,:,0]
     g=img_f[:,:,1]
     b=img_f[:,:,2]

     if value>0.5:  # image moves toward warm
        r_new=r+(value-0.5)*((255-r)/0.5)  # r value boosts up so image looks warmer
        # at extreme value =1 r_new will update to 255 making warmest image
        b_new=b+(0.5-value)*((255-r)/0.5)  # b value decreases

     elif value<0.5:  # image moves toward cold
          r_new=r+(value-0.5)*((255-b)/0.5) # r value decreases
          # at extreme value=0 b_new will update to 255 making coldest image
          b_new=b+(0.5-value)*((255-b)/0.5)  # b value boosts up to make image cold
     else:    # value =0.5 image looks original
       r_new=r
       b_new=b
     # merging back to get balanced image
     img_balanced=np.dstack((r_new,g,b_new))
     # clipping to [0-255]
     img_balanced=np.clip(img_balanced,[0,0,0],[255,255,255]).astype(np.uint8)
     return img_balanced
# balancing image on basis of value
img_balanced1=white_balance(img_rgb,0)
img_balanced2=white_balance(img_rgb,0.25)
img_balanced3=white_balance(img_rgb,0.5)
img_balanced4=white_balance(img_rgb,0.75)
img_balanced5=white_balance(img_rgb,1)
# display

plt.figure(figsize=(20,10))
plt.subplot(1,5,1)
plt.imshow(img_balanced1)
plt.title("Coldest image")
plt.axis("off")

plt.subplot(1,5,2)
plt.imshow(img_balanced2)
plt.title("colder image")
plt.axis("off")

plt.subplot(1,5,3)
plt.imshow(img_balanced3)
plt.title("Original image")
plt.axis("off")

plt.subplot(1,5,4)
plt.imshow(img_balanced4)
plt.title("warmer image")
plt.axis("off")

plt.subplot(1,5,5)
plt.imshow(img_balanced5)
plt.title("Warmest image")
plt.axis("off")

plt.show()




````
