````python
def lpf_mask_circle(rows, cols, radius):
    mask = np.zeros((rows, cols), dtype=np.float32)
    crow, ccol = rows // 2, cols // 2   # center

    Y, X = np.ogrid[:rows, :cols]
    dist = (Y - crow)**2 + (X - ccol)**2

    mask[dist <= radius * radius] = 1.0
    return mask

rows, cols = gray.shape

radii = [10, 30, 80]

masks = []
for r in radii:
  masks.append(lpf_mask_circle(rows, cols, r))

img_restored_lpf = []
for i in range(len(radii)):
    r = radii[i]
    mask = masks[i]

    # apply LPF in frequency domain
    fshift_lpf = fshift * mask

    # inverse shift + inverse FFT
    f_ishift = np.fft.ifftshift(fshift_lpf)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    img_restored_lpf.append((r, img_back))


# Plot results
plt.figure(figsize=(14, 4))

plt.subplot(1, 4, 1)
plt.title("Original gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

for i, (r, img_b) in enumerate(img_restored_lpf, start=2):
    plt.subplot(1, 4, i)
    plt.title(f"LPF radius={r}")
    plt.imshow(img_b, cmap='gray')
    plt.axis("off")

plt.show()

# Comment you can add in code:
# Smaller radius → more low frequencies kept, more high frequencies removed → image looks more blurred.
````

````python
def hpf_mask_circle(rows, cols, radius):
    mask = np.ones((rows, cols), dtype=np.float32)
    crow, ccol = rows // 2, cols // 2   # center

    Y, X = np.ogrid[:rows, :cols]
    dist = (Y - crow)**2 + (X - ccol)**2

    mask[dist <= radius * radius] = 0.0
    return mask

rows, cols = gray.shape

radii = [10, 30, 80]

masks = []
for r in radii:
  masks.append(hpf_mask_circle(rows, cols, r))

img_restored_hpf = []
for i in range(len(radii)):
    r = radii[i]
    mask = masks[i]

    # apply LPF in frequency domain
    fshift_hpf = fshift * mask

    # inverse shift + inverse FFT
    f_ishift = np.fft.ifftshift(fshift_hpf)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    img_restored_hpf.append((r, img_back))


# Plot results
plt.figure(figsize=(14, 4))

plt.subplot(1, 4, 1)
plt.title("Original gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

for i, (r, img_b) in enumerate(img_restored_hpf, start=2):
    plt.subplot(1, 4, i)
    plt.title(f"HPF radius={r}")
    plt.imshow(img_b, cmap='gray')
    plt.axis("off")

plt.show()

````

````python
square_side = 50
half = square_side // 2
crow, ccol = rows // 2, cols // 2   # center

# start with all zeros (block all frequencies)
square_mask = np.zeros((rows, cols), dtype=np.float32)

# put a square of 1s at the center → pass those low frequencies
square_mask[crow-half:crow+half, ccol-half:ccol+half] = 1.0

# Apply square LPF
fshift_square = fshift * square_mask

# Reconstruct
f_ishift_square = np.fft.ifftshift(fshift_square)
img_square_lpf = np.fft.ifft2(f_ishift_square)
img_square_lpf = np.abs(img_square_lpf)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Square LPF Mask")
plt.imshow(square_mask, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Square LPF output")
plt.imshow(img_square_lpf, cmap='gray')
plt.axis("off")

plt.show()
````

````python
eps = 1e-8
phase_only = fshift / (np.abs(fshift) + eps)

f_ishift_phase = np.fft.ifftshift(phase_only)
img_phase_only = np.fft.ifft2(f_ishift_phase)
img_phase_only = np.abs(img_phase_only)

# reconstruction from np.angle(fshift) explicitly
angles = np.angle(fshift)               # real-valued phase (radians)
phase_complex = np.exp(1j * angles)     # make unit-magnitude complex spectrum from angles

f_ishift_angle = np.fft.ifftshift(phase_complex)
img_from_angle = np.fft.ifft2(f_ishift_angle)
img_from_angle = np.abs(img_from_angle)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Phase-only (fshift / |fshift|)")
plt.imshow(img_phase_only, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Reconstruction from exp(1j·angle)")
plt.imshow(img_from_angle, cmap='gray')
plt.axis("off")

plt.show()
#increase the Brightness to see the details in phase plot 2 and 3
#both plots are nearly same, both contains important info like edges in the image which indicates that phase is more important than magnitude to determine overall structire of the image
````

````python
mag_only = np.abs(fshift)
mag_only_complex = mag_only * np.exp(1j * 0)

f_ishift_mag = np.fft.ifftshift(mag_only_complex)
img_mag_only = np.fft.ifft2(f_ishift_mag)
img_mag_only = np.abs(img_mag_only)

mag_spectrum = np.log(1 + mag_only)   # log compresses large range

plt.figure(figsize = (15,5))
plt.subplot(1,3,1)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Magnitude Spectrum (log scaled)")
plt.imshow(mag_spectrum, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Magnitude-only Reconstruction")
plt.imshow(img_mag_only, cmap='gray')
plt.axis("off")

plt.show()


# 1. Why log is used in magnitude spectrum?
# Because the raw magnitudes are too big to see properly. Log shrinks them so the spectrum becomes visible.

# 2. Why magnitude-only reconstruction looks different from the image?
# Because magnitude doesnt store shapes or details, it just stores Texture and all other ddetails of the image
````

````python
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np

# PASTE BASE64 STRING HERE
base64_image = """

"""  # long base64 string

img_data = base64.b64decode(base64_image)
L = Image.open(io.BytesIO(img_data))

plt.imshow(img)
plt.axis("off")
````

````python
#converting to Grayscale
kira_np = np.array(kira)
L_np = np.array(L)
gray_kira = cv2.cvtColor(kira_np, cv2.COLOR_RGB2GRAY)
gray_L = cv2.cvtColor(L_np, cv2.COLOR_RGB2GRAY)


if gray_L.shape != gray_kira.shape:
    gray_L = cv2.resize(gray_L, (gray_kira.shape[1], gray_kira.shape[0]))


plt.subplot(1,2,1)
plt.imshow(gray_kira, cmap='gray')
plt.title('Grascale Kira')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gray_L, cmap='gray')
plt.title('Grascale L')
plt.axis('off')
````

````python
gray_kira = gray_kira.astype(np.float32)
gray_L = gray_L.astype(np.float32)

#calculating fft of both images
fft_kira = np.fft.fft2(gray_kira)
fft_L = np.fft.fft2(gray_L)

fft_shifted_kira = np.fft.fftshift(fft_kira)
fft_shifted_L = np.fft.fftshift(fft_L)

#calculating mag and phase of both
mag_kira = np.abs(fft_shifted_kira)
mag_L = np.abs(fft_shifted_L)

phase_kira = np.angle(fft_shifted_kira)
phase_L = np.angle(fft_shifted_L)

phase_kira_mag_L = mag_L * np.exp(1j*phase_kira)
phase_L_mag_kira = mag_kira * np.exp(1j*phase_L)

#Reconstruction
FS1 = np.fft.ifftshift(phase_kira_mag_L)
FS2 = np.fft.ifftshift(phase_L_mag_kira)

reconstructed_kira = np.fft.ifft2(FS1)
reconstructed_kira = np.abs(reconstructed_kira)

reconstructed_L = np.fft.ifft2(FS2)
reconstructed_L = np.abs(reconstructed_L)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.title("Gray kira")
plt.imshow(gray_kira, cmap='gray')
plt.axis("off")

plt.subplot(2, 2, 2)
plt.title("phase of kira + mag of L")
plt.imshow(reconstructed_kira, cmap='gray')
plt.axis("off")

plt.subplot(2, 2, 3)
plt.title("Gray L")
plt.imshow(gray_L, cmap='gray')
plt.axis("off")

plt.subplot(2, 2, 4)
plt.title("phase of L + mag of kira")
plt.imshow(reconstructed_L, cmap='gray')
plt.axis("off")

plt.tight_layout()
plt.show()

#hence proved that phas contains more info of the overall structure of the image than the magnitude
#that is Phase preserves structure while magnitude preserves the texxture of the image
````
