````python
import numpy as np
import matplotlib.pyplot as plt
import requests
from PIL import Image
from io import BytesIO

#task 1
url = "https://picsum.photos/200"

response = requests.get(url)
img = Image.open(BytesIO(response.content)).convert("RGB")

rgb_img = np.array(img)
gray_img = np.array(img.convert("L"))

def convolve(img, kernel):
    img = img.astype(np.float32)
    kernel = kernel.astype(np.float32)

    kH, kW = kernel.shape
    pad_h = kH // 2
    pad_w = kW // 2

    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    # reflect padding mirrors image values at the boundary.
    # This avoids artificial dark borders that zero-padding would introduce
    # and preserves edge continuity during convolution.

    H, W = img.shape
    output = np.zeros((H, W), dtype=np.float32)

    for i in range(H):
        for j in range(W):
            region = padded[i:i+kH, j:j+kW]
            output[i, j] = np.sum(region * kernel)

    # In NumPy, * performs element-wise multiplication.
    # This is valid here because 'region' and 'kernel' have the same shape.
    # NumPy does NOT perform convolution automatically with *.

    return output

#task 2
# Box filter averages local neighborhood values.
# Normalizing by kernel size ensures brightness is preserved.

def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32)
    kernel /= kernel.size
    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    output = np.zeros_like(img, dtype=np.float32)
    for c in range(3):
        output[:, :, c] = average_blur_gray(img[:, :, c], ksize)
    return output

#task 3
# Gaussian kernel weights nearby pixels more than distant ones.
# Kernel is normalized so that sum equals 1, preserving image intensity.
# sigma controls the spread of the Gaussian.
# Larger sigma results in stronger smoothing (lower high-frequency content).

def gaussian_kernel(ksize, sigma):
    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= np.sum(kernel)

    return kernel

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    output = np.zeros_like(img, dtype=np.float32)
    for c in range(3):
        output[:, :, c] = gaussian_blur_gray(img[:, :, c], ksize, sigma)
    return output

# outputs already computed
avg_gray = average_blur_gray(gray_img, ksize=5)
gauss_gray = gaussian_blur_gray(gray_img, ksize=5, sigma=1.0)

plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(gray_img, cmap='gray')
plt.axis('off')

plt.subplot(1,3,2)
plt.title("Average Blur")
plt.imshow(avg_gray, cmap='gray')
plt.axis('off')

plt.subplot(1,3,3)
plt.title("Gaussian Blur")
plt.imshow(gauss_gray, cmap='gray')
plt.axis('off')

plt.show()

# outputs already computed
avg_rgb = average_blur_rgb(rgb_img, ksize=5)
gauss_rgb = gaussian_blur_rgb(rgb_img, ksize=5, sigma=1.0)

plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(np.clip(rgb_img / 255.0, 0, 1))
plt.axis('off')

plt.subplot(1,3,2)
plt.title("Average Blur")
plt.imshow(np.clip(avg_rgb / 255.0, 0, 1))
plt.axis('off')

plt.subplot(1,3,3)
plt.title("Gaussian Blur")
plt.imshow(np.clip(gauss_rgb / 255.0, 0, 1))
plt.axis('off')

plt.show()
````

````python
url = "https://picsum.photos/200"
response = requests.get(url)

img = Image.open(BytesIO(response.content)).convert("L")
gray_img = np.array(img.convert("L"))

sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

sobel_y = np.array([
    [-1, -2, -1],
    [0, 0, 0],
    [1, 2, 1]
], dtype=np.float32)

# Gx: horizontal intensity changes (vertical edges)
Gx = convolve(gray_img, sobel_x)
# Gy: vertical intensity changes (horizontal edges)
Gy = convolve(gray_img, sobel_y)

# Gradient magnitude combines both directions
gradient_magnitude = np.sqrt(Gx ** 2 + Gy ** 2)

# Normalize gradient magnitude to range 0–255 for visualization and saving
grad_norm = gradient_magnitude / gradient_magnitude.max()
grad_norm = (grad_norm * 255).astype(np.uint8)

# Optional thresholding step:
# The normalized gradient magnitude contains both strong edges and weak responses due to noise.
# By applying a threshold, only pixels with gradient values greater than or equal to the threshold
# are retained as edges, while all weaker responses are suppressed.
threshold = 50
edge_map = np.zeros_like(grad_norm)
edge_map[grad_norm >= threshold] = 255

plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(gray_img, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Gradient magnitude")
plt.imshow(grad_norm, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Edge map")
plt.imshow(edge_map, cmap="gray")
plt.axis("off")

plt.show()

Image.fromarray(edge_map).save("sobel_edge_map.png")
````

````python
# make sure images are float
orig = gray.astype(np.float32)

# function to compute magnitude spectrum
def magnitude_spectrum(img):
    fft = np.fft.fft2(img)
    fft_shift = np.fft.fftshift(fft)
    magnitude = np.log(1 + np.abs(fft_shift))
    return magnitude

# compute spectra
spec_orig = magnitude_spectrum(orig)
spec_blur = magnitude_spectrum(blurred)
spec_sharp = magnitude_spectrum(sharp_unsharp)

# normalize for display
spec_orig /= spec_orig.max()
spec_blur /= spec_blur.max()
spec_sharp /= spec_sharp.max()

# display
plt.figure(figsize=(12,6))

plt.subplot(2,3,1)
plt.title("Original image")
plt.imshow(orig, cmap="gray")
plt.axis("off")

plt.subplot(2,3,2)
plt.title("Gaussian blurred")
plt.imshow(blurred, cmap="gray")
plt.axis("off")

plt.subplot(2,3,3)
plt.title("Unsharp masked")
plt.imshow(sharp_unsharp, cmap="gray")
plt.axis("off")

plt.subplot(2,3,4)
plt.title("Spectrum: original")
plt.imshow(spec_orig, cmap="gray")
plt.axis("off")

plt.subplot(2,3,5)
plt.title("Spectrum: blurred")
plt.imshow(spec_blur, cmap="gray")
plt.axis("off")

plt.subplot(2,3,6)
plt.title("Spectrum: unsharp")
plt.imshow(spec_sharp, cmap="gray")
plt.axis("off")

plt.show()
````

````python
img = img_part5.astype(np.float32)

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

# simple edge kernel (Sobel-like but symmetric)
edge_kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

# edge response in Red and Green channels
edge_R = convolve(R, edge_kernel)
edge_G = convolve(G, edge_kernel)

# compute R - G difference image
RG_diff = R - G

# edge response of R-G difference
edge_RG = convolve(RG_diff, edge_kernel)

# combine responses
# yellow edges give strong R, G and R-G response
combined = np.abs(edge_R) + np.abs(edge_G) + 2 * np.abs(edge_RG)

# normalize to 0-255
combined = combined / combined.max()
combined = (combined * 255).astype(np.uint8)

# threshold to get clean outline
threshold = 100
yellow_edges = np.zeros_like(combined)
yellow_edges[combined > threshold] = 255

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img_part5.astype(np.uint8))
plt.title("Original image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(combined, cmap="gray")
plt.title("Combined response")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(yellow_edges, cmap="gray")
plt.title("Yellow leaf edges")
plt.axis("off")

plt.show()

# Observation:
# Yellow regions have high Red and Green values but low Blue.
# Green regions have high Green but low Red and Blue.

# Observation:
# When transitioning from green to yellow, Red increases sharply
# while Green remains high, causing a strong gradient in R and in (R - G).

# Observation:
# The edge response of (R - G) is significantly stronger at yellow boundaries
# compared to pure green boundaries.

# Conclusion:
# By combining edge responses from Red, Green and (R - G),
# we selectively highlight edges where colour transitions into yellow.
# at certain spots of the yellow leave's edge, we get a break because instead of the yellow leaf not transitioning into a green leaf
````

````python
# Convert the RGB image to HSV color space.
# HSV separates color information (Hue) from intensity (Value),
# making color manipulation easier than in RGB.
def rgb_to_hsv(img):
    img = img.astype(np.float32) / 255.0
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]

    Cmax = np.max(img, axis=2)
    Cmin = np.min(img, axis=2)
    delta = Cmax - Cmin

    H = np.zeros_like(Cmax)

    mask = delta != 0
    r_mask = (Cmax == R) & mask
    g_mask = (Cmax == G) & mask
    b_mask = (Cmax == B) & mask

    H[r_mask] = (60 * ((G[r_mask] - B[r_mask]) / delta[r_mask]) + 360) % 360
    H[g_mask] = (60 * ((B[g_mask] - R[g_mask]) / delta[g_mask]) + 120)
    H[b_mask] = (60 * ((R[b_mask] - G[b_mask]) / delta[b_mask]) + 240)

    S = np.zeros_like(Cmax)
    S[Cmax != 0] = delta[Cmax != 0] / Cmax[Cmax != 0]

    V = Cmax

    return np.stack([H, S, V], axis=2)

def hsv_to_rgb(hsv):
    H = hsv[..., 0]
    S = hsv[..., 1]
    V = hsv[..., 2]

    C = V * S
    X = C * (1 - np.abs((H / 60) % 2 - 1))
    m = V - C

    R = np.zeros_like(H)
    G = np.zeros_like(H)
    B = np.zeros_like(H)

    h0 = (0 <= H) & (H < 60)
    h1 = (60 <= H) & (H < 120)
    h2 = (120 <= H) & (H < 180)
    h3 = (180 <= H) & (H < 240)
    h4 = (240 <= H) & (H < 300)
    h5 = (300 <= H) & (H < 360)

    R[h0], G[h0], B[h0] = C[h0], X[h0], 0
    R[h1], G[h1], B[h1] = X[h1], C[h1], 0
    R[h2], G[h2], B[h2] = 0, C[h2], X[h2]
    R[h3], G[h3], B[h3] = 0, X[h3], C[h3]
    R[h4], G[h4], B[h4] = X[h4], 0, C[h4]
    R[h5], G[h5], B[h5] = C[h5], 0, X[h5]

    rgb = np.stack([R + m, G + m, B + m], axis=2)
    return (rgb * 255).astype(np.uint8)

# convert to HSV
hsv = rgb_to_hsv(img_part5)

H = hsv[..., 0]
S = hsv[..., 1]
V = hsv[..., 2]

# Create a mask to isolate yellow pixels.
# Yellow occupies a narrow range of Hue values in HSV.
# Saturation and Value thresholds remove dark or non-yellow regions.
# yellow mask in HSV
yellow_mask = (
    (H > 30) & (H < 80) &     # wider hue range
    (S > 0.25) &              # allow low saturation
    (V > 0.3)                 # allow highlights
)

R = img_part5[..., 0].astype(np.float32) / 255.0
B = img_part5[..., 2].astype(np.float32) / 255.0

yellow_mask = yellow_mask & (R > B)

# Change the Hue of detected yellow pixels to red.
# In this HSV implementation, Hue is measured in degrees [0, 360),
# and Hue = 0 corresponds to red.
# Saturation and Value are slightly scaled (not overwritten)
# to obtain a less bright red while preserving texture.
H[yellow_mask] = 0
S[yellow_mask] = np.clip(S[yellow_mask] * 0.9, 0, 1)
V[yellow_mask] = np.clip(V[yellow_mask] * 0.9, 0, 1)

hsv_modified = np.stack([H, S, V], axis=2)

# Convert the modified HSV image back to RGB for display.
result = hsv_to_rgb(hsv_modified)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img_part5)
plt.title("Original image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(result)
plt.title("Yellow leaf changed to red")
plt.axis("off")

plt.show()
````
