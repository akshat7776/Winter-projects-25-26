````python
#Implementing C1 Part by creating 3 LPF mask as demonstarted in the slides and lecs
row,col = gray_img.shape
crow = row//2
ccol = col//2

radius = 10

mask_lpf1 = np.zeros((row,col),np.uint8)
cv2.circle(mask_lpf1,(ccol,crow),radius,1,-1)

f_lpf1 = fshift * mask_lpf1
ishift_lpf1 = np.fft.ifftshift(f_lpf1)
img_back_lpf1 = np.fft.ifft2(ishift_lpf1)
img_back_lpf1 = np.abs(img_back_lpf1)

radius = 30

mask_lpf2 = np.zeros((row,col),np.uint8)
cv2.circle(mask_lpf2,(ccol,crow),radius,1,-1)

f_lpf2 = fshift * mask_lpf2
ishift_lpf2 = np.fft.ifftshift(f_lpf2)
img_back_lpf2 = np.fft.ifft2(ishift_lpf2)
img_back_lpf2 = np.abs(img_back_lpf2)

radius = 80

mask_lpf3 = np.zeros((row,col),np.uint8)
cv2.circle(mask_lpf3,(ccol,crow),radius,1,-1)

f_lpf3 = fshift * mask_lpf3
ishift_lpf3 = np.fft.ifftshift(f_lpf3)
img_back_lpf3 = np.fft.ifft2(ishift_lpf3)
img_back_lpf3 = np.abs(img_back_lpf3)

plt.figsize=(15,15)

plt.subplot(1,3,1)
plt.imshow(img_back_lpf1,cmap='gray')
plt.title("LPF with radius=10")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img_back_lpf2,cmap='gray')
plt.title("LPF with radius=30")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_back_lpf3,cmap='gray')
plt.title("LPF with radius=80")
plt.axis('off')

## As the radius decreases the area through which high frequency data will be blocked increases thus more high frequency data is blocked ... and as covered in the slides this removes edges and details and thus the images smoothens and blurs that is only low frequency component remains.
````

````python
#Implementing C2 Part by creating HPF Mask as demonstarted in lec and slides

mask_hpf1 = 1 - mask_lpf1
mask_hpf2 = 1 - mask_lpf2
mask_hpf3 = 1 - mask_lpf3

f_hpf1 = fshift * mask_hpf1
ishift_hpf1 = np.fft.ifftshift(f_hpf1)
img_back_hpf1 = np.fft.ifft2(ishift_hpf1)
img_back_hpf1 = np.abs(img_back_hpf1)

f_hpf2 = fshift * mask_hpf2
ishift_hpf2 = np.fft.ifftshift(f_hpf2)
img_back_hpf2 = np.fft.ifft2(ishift_hpf2)
img_back_hpf2 = np.abs(img_back_hpf2)

f_hpf3 = fshift * mask_hpf3
ishift_hpf3 = np.fft.ifftshift(f_hpf3)
img_back_hpf3 = np.fft.ifft2(ishift_hpf3)
img_back_hpf3 = np.abs(img_back_hpf3)

plt.figsize=(15,15)

plt.subplot(1,3,1)
plt.imshow(img_back_hpf1,cmap='gray')
plt.title("HPF with radius=10")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img_back_hpf2,cmap='gray')
plt.title("HPF with radius=30")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_back_hpf3,cmap='gray')
plt.title("HPF with radius=80")
plt.axis('off')
````
