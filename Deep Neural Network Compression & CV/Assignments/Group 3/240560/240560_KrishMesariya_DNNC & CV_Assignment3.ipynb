````python
# Part A

def convolve(image, filt):
    filt = np.asarray(filt)
    k = filt.shape[0]          # assuming square filter
    image = np.asarray(image)

    result = np.zeros_like(image, dtype=float)
    p = k // 2

    padded_img = np.pad(image, ((p, p), (p, p)), mode='reflect')
    height, width = image.shape

    for row in range(height):
        for col in range(width):
            val = 0.0
            for u in range(k):
                for v in range(k):
                    val += padded_img[row + u, col + v] * filt[u, v]
            result[row, col] = val

    return result
````

````python
def convert_rgb_to_hsv(image):
    img = image.astype(np.float32) / 255.0

    r_ch = img[:, :, 0]
    g_ch = img[:, :, 1]
    b_ch = img[:, :, 2]

    max_val = np.maximum.reduce([r_ch, g_ch, b_ch])
    min_val = np.minimum.reduce([r_ch, g_ch, b_ch])
    delta = max_val - min_val

    hue = np.zeros_like(max_val)
    sat = np.zeros_like(max_val)
    val = max_val

    non_zero = delta != 0

    idx = (max_val == r_ch) & non_zero
    hue[idx] = (60 * ((g_ch - b_ch) / delta) % 360)[idx]

    idx = (max_val == g_ch) & non_zero
    hue[idx] = (60 * ((b_ch - r_ch) / delta) + 120)[idx]

    idx = (max_val == b_ch) & non_zero
    hue[idx] = (60 * ((r_ch - g_ch) / delta) + 240)[idx]

    sat[max_val != 0] = delta[max_val != 0] / max_val[max_val != 0]

    hsv_img = np.zeros_like(img)
    hsv_img[:, :, 0] = hue / 2
    hsv_img[:, :, 1] = sat * 255
    hsv_img[:, :, 2] = val * 255

    return hsv_img.astype(np.uint8)


hsv_img = convert_rgb_to_hsv(img_part5)
H_ch, S_ch, V_ch = hsv_img[:, :, 0], hsv_img[:, :, 1], hsv_img[:, :, 2]

yellow_mask = (H_ch >= 15) & (H_ch <= 40) & (S_ch > 60) & (V_ch > 50)

k = float(input(
    "Enter brightness factor (0 → yellow, 1 → red, e.g. 0.3 or 0.5): "
))

hsv_modified = hsv_img.copy()
hsv_modified[:, :, 0][yellow_mask] = np.clip(
    (1 - k) * hsv_modified[:, :, 0][yellow_mask], 0, 255
)

final_img = cv2.cvtColor(hsv_modified, cv2.COLOR_HSV2RGB)

plt.imshow(final_img)
plt.axis('off')
plt.show()
````
