````python

import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files


uploaded = files.upload()
img_name = list(uploaded.keys())[0]

img = cv2.imread(img_name)

if img is None:
    raise ValueError("Image not loaded properly")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

img_rgb = img_rgb.astype(np.float32)
img_gray = img_gray.astype(np.float32)


# Task 1

def convolve(img, kernel):
    k = kernel.shape[0]
    pad = k // 2

    padded = np.pad(img, pad, mode='reflect')
    h, w = img.shape
    out = np.zeros((h, w), dtype=np.float32)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out


# Task 2

def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32)
    kernel = kernel / (ksize * ksize)
    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    out = np.zeros_like(img)
    for c in range(3):
        out[:, :, c] = average_blur_gray(img[:, :, c], ksize)
    return out


# Task 3

def gaussian_kernel(ksize, sigma):
    ax = np.arange(-ksize//2 + 1, ksize//2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)
    return kernel.astype(np.float32)

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    out = np.zeros_like(img)
    for c in range(3):
        out[:, :, c] = gaussian_blur_gray(img[:, :, c], ksize, sigma)
    return out


ksize = 5
sigma = 1.0

avg_gray = average_blur_gray(img_gray, ksize)
avg_rgb  = average_blur_rgb(img_rgb, ksize)

gauss_gray = gaussian_blur_gray(img_gray, ksize, sigma)
gauss_rgb  = gaussian_blur_rgb(img_rgb, ksize, sigma)


plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.title("Original RGB")
plt.imshow(img_rgb.astype(np.uint8))
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title("Average Blur RGB")
plt.imshow(avg_rgb.astype(np.uint8))
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title("Gaussian Blur RGB")
plt.imshow(gauss_rgb.astype(np.uint8))
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title("Original Gray")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title("Average Blur Gray")
plt.imshow(avg_gray, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title("Gaussian Blur Gray")
plt.imshow(gauss_gray, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python


import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files


uploaded = files.upload()
img_name = list(uploaded.keys())[0]

img = cv2.imread(img_name)

if img is None:
    raise ValueError("Image not loaded")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray = gray.astype(np.float32)


def convolve(img, kernel):
    k = kernel.shape[0]
    pad = k // 2

    padded = np.pad(img, pad, mode='reflect')
    h, w = img.shape
    out = np.zeros((h, w), dtype=np.float32)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out


sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]], dtype=np.float32)

sobel_y = np.array([[ 1,  2,  1],
                    [ 0,  0,  0],
                    [-1, -2, -1]], dtype=np.float32)


Gx = convolve(gray, sobel_x)
Gy = convolve(gray, sobel_y)

# Gradient magnitude
grad_mag = np.sqrt(Gx**2 + Gy**2)


def normalize(img):
    img = img - img.min()
    img = img / img.max()
    return (img * 255).astype(np.uint8)

Gx_n = normalize(np.abs(Gx))
Gy_n = normalize(np.abs(Gy))
grad_n = normalize(grad_mag)

plt.figure(figsize=(12, 6))

plt.subplot(1, 4, 1)
plt.title("Original Gray")
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title("Sobel Gx")
plt.imshow(Gx_n, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title("Sobel Gy")
plt.imshow(Gy_n, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title("Gradient Magnitude")
plt.imshow(grad_n, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()


cv2.imwrite("sobel_edge_map.png", grad_n)
````

````python


import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files


uploaded = files.upload()
img_name = list(uploaded.keys())[0]

img = cv2.imread(img_name, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image not loaded")

img = img.astype(np.float32)


def convolve(img, kernel):
    k = kernel.shape[0]
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')

    h, w = img.shape
    out = np.zeros((h, w), dtype=np.float32)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out


# Task 1


# 8-connected Laplacian kernel
laplacian_kernel = np.array([[ 1,  1,  1],
                             [ 1, -8,  1],
                             [ 1,  1,  1]], dtype=np.float32)

# Compute Laplacian
lap = convolve(img, laplacian_kernel)

# Sharpen with alpha in [0.2, 1]
alpha = 0.5
sharp_lap = img - alpha * lap

# Alpha = 1 case
sharp_lap_alpha1 = img - lap

lap_cv = cv2.Laplacian(img, cv2.CV_32F, ksize=3)
sharp_cv = img - lap_cv

# Compare matrices
same_output = np.allclose(sharp_lap_alpha1, sharp_cv)
print("NumPy and OpenCV outputs same (alpha=1):", same_output)


# Task 2


# Gaussian kernel
def gaussian_kernel(ksize, sigma):
    ax = np.arange(-ksize//2 + 1, ksize//2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= np.sum(kernel)
    return kernel.astype(np.float32)

# Gaussian blur
g_kernel = gaussian_kernel(5, 1.0)
blurred = convolve(img, g_kernel)

# Mask
mask = img - blurred

# Unsharp masking
alpha_usm = 1.0
sharp_usm = img + alpha_usm * mask


alpha_negative = -1.0
blurred_from_negative = img + alpha_negative * mask


def normalize(img):
    img = img - img.min()
    img = img / img.max()
    return (img * 255).astype(np.uint8)


plt.figure(figsize=(14, 8))

plt.subplot(2, 3, 1)
plt.title("Original Gray")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.title("Laplacian")
plt.imshow(normalize(lap), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.title("Laplacian Sharpen (alpha=0.5)")
plt.imshow(normalize(sharp_lap), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.title("Gaussian Blurred")
plt.imshow(normalize(blurred), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.title("Unsharp Mask (alpha=1)")
plt.imshow(normalize(sharp_usm), cmap='gray')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.title("Negative Alpha (Blur Effect)")
plt.imshow(normalize(blurred_from_negative), cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python


import numpy as np
import cv2
import matplotlib.pyplot as plt


def magnitude_spectrum(img):
    """
    Computes log magnitude spectrum of an image
    """
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude = np.log(1 + np.abs(fshift))
    return magnitude


orig = img
blur = blurred
sharp = sharp_usm


spec_orig = magnitude_spectrum(orig)
spec_blur = magnitude_spectrum(blur)
spec_sharp = magnitude_spectrum(sharp)

plt.figure(figsize=(12, 8))

plt.subplot(3, 2, 1)
plt.title("Original Image")
plt.imshow(orig, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 2)
plt.title("Magnitude Spectrum (Original)")
plt.imshow(spec_orig, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 3)
plt.title("Gaussian Blurred Image (LPF)")
plt.imshow(blur, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 4)
plt.title("Magnitude Spectrum (Gaussian Blur)")
plt.imshow(spec_blur, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 5)
plt.title("Unsharp Masked Image (HPF)")
plt.imshow(sharp, cmap='gray')
plt.axis('off')

plt.subplot(3, 2, 6)
plt.title("Magnitude Spectrum (Unsharp Mask)")
plt.imshow(spec_sharp, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
# Bright center = low frequencies (smooth regions)
# Dark outer areas = high frequencies (edges and details)
# Gaussian blur reduces high frequencies (LPF)
# Unsharp masking enhances high frequencies (HPF)
````

````python


import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files


uploaded = files.upload()
img_name = list(uploaded.keys())[0]

img = cv2.imread(img_name)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = img.astype(np.float32)

# Split channels
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

def convolve(img, kernel):
    k = kernel.shape[0]
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')

    h, w = img.shape
    out = np.zeros((h, w), dtype=np.float32)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+k, j:j+k]
            out[i, j] = np.sum(region * kernel)

    return out


edge_kernel = np.array([[-1,  0,  1],
                        [-2,  0,  2],
                        [-1,  0,  1]], dtype=np.float32)



edge_R = convolve(R, edge_kernel)

edge_G = convolve(G, edge_kernel)

edge_B = convolve(B, edge_kernel)


RG_diff = R - G
edge_RG = convolve(RG_diff, edge_kernel)

combined = np.abs(edge_R) + np.abs(edge_RG) - 0.5 * np.abs(edge_B)

combined -= combined.min()
combined /= combined.max()
combined = (combined * 255).astype(np.uint8)

_, edge_map = cv2.threshold(combined, 60, 255, cv2.THRESH_BINARY)


plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.imshow(img.astype(np.uint8))
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Combined Edge Response")
plt.imshow(combined, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Detected Yellow Leaf Edges")
plt.imshow(edge_map, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
# Explanation:
# Yellow regions have high Red and Green values but low Blue,
# whereas green regions have high Green and low Red.
# Hence, the transition from green to yellow causes a strong
# increase in the Red channel and in the (R - G) difference.
#
# By detecting edges separately in RGB channels and emphasizing
# Red and (R - G) responses while suppressing Blue, the custom
# kernel highlights boundaries where the colour transitions
# into a yellowish hue, giving a clean outline around the leaf.
````
