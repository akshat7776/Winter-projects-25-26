````python
import numpy as np
def convolve(img,kernel):
  row,col=img.shape
  rowk,colk=kernel.shape
  pad=rowk//2
  imgfinal = np.zeros((row, col), dtype=np.float32)
  img_pad=np.pad(img,((pad,pad),(pad,pad)),mode='reflect')
  kernel = np.flipud(np.fliplr(kernel))
  for i in range(row):
    for j in range(col):
      imgfinal[i,j]=np.sum(img_pad[i:i+rowk,j:j+colk]*kernel)
  return imgfinal

````

````python
def gaussian_kernel(ksize, sigma):
    if ksize % 2 == 0:
        raise ValueError("ksize must be odd")

    ax = np.arange(-ksize // 2 + 1, ksize // 2 + 1, dtype=np.float32)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2.0 * sigma**2))
    kernel /= np.sum(kernel)
    return kernel
````

````python
def convolve_same(img, kernel):
    row, col = img.shape
    rowk, colk = kernel.shape
    pad = rowk // 2
    img_padded = np.pad(img, ((pad, pad), (pad, pad)), mode='reflect')
    kernel = np.flipud(np.fliplr(kernel))
    imgfinal = np.zeros((row, col), np.float32)
    for i in range(row):
        for j in range(col):
            imgfinal[i,j] = np.sum(img_padded[i:i+rowk, j:j+colk] * kernel)
    return imgfinal
````

````python
def magnitude_spectrum(img):
    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude = np.log(1 + np.abs(fshift))
    return magnitude
````

````python
def rgb_to_hsv(arr):
  rows,columns,color=arr.shape
  hsv_arr = np.zeros((rows, columns, 3), dtype=np.float64)
  for i in range(rows):
    for j in range(columns):
      r_norm=arr[i,j,0]/255
      g_norm=arr[i,j,1]/255
      b_norm=arr[i,j,2]/255
      cmax=max(r_norm,g_norm,b_norm)
      cmin=min(r_norm,g_norm,b_norm)
      delta=cmax-cmin
      if delta==0:
        hue=0
      elif delta>0:
        if cmax==r_norm:
          hue=60*(((g_norm-b_norm)/delta)%6)
        elif cmax==g_norm:
          hue=60*(((b_norm-r_norm)/delta)+2)
        elif cmax==b_norm:
          hue=60*(((r_norm-g_norm)/delta)+4)

      if cmax==0:
        sat=0
      else:
        sat_norm=delta/cmax
        sat=sat_norm*255

      v=cmax*255

      hsv_arr[i,j,0]=hue/2.0
      hsv_arr[i,j,1]=sat
      hsv_arr[i,j,2]=v

  hsv_arr=np.clip(hsv_arr,0,255).astype(np.uint8)
  return hsv_arr
````
