````python
#TASK 1
def convolve(img, kernel):

    x, y = kernel.shape
    img_with_padding = np.pad(
        img,((x//2, x//2), (y//2, y//2)),mode='reflect'
    )

    output = np.zeros(img.shape)
    a, b = img.shape

    for i in range(a):
        for j in range(b):
            output[i, j] = np.sum(
                img_with_padding[i:i+x, j:j+y] * kernel
            )

    return output
````
