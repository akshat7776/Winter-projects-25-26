Now, you will need to create a function <code>white_balance(img_rgb, value)</code>

It takes the img_rgb ( 0-255 RGB image ) and value ( 0-1 decimal number ).
1. value = 0 outputs the coldest version of the image
2. value = 0.5 is the original image itself
3. value = 1 is the warmest version of the image

The output should be the processed 0-255 RGB image. Display the original and processed image.

Try to make it look as good as possible. I dont expect you guys to copy my phone and create industry level outputs, but work hard and experiment. You can include even failed tries or what you felt was not satisfactory ( that will show how much effort you put in )

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt


bgr = cv2.imread("EEA pic2.jpg")
img_rgb = bgr[:, :, ::-1].astype(np.float32)

def white_balance(img_rgb, value):

    img = img_rgb.copy()

    t = (value - 0.5) * 2.0
    gain_r = 1.0 + t
    gain_b = 1.0 - t

    img[:, :, 0] = img[:, :, 0] * gain_r
    img[:, :, 1] = img[:, :, 1] * 1.0
    img[:, :, 2] = img[:, :, 2] * gain_b
    np.clip(img, 0, 255, out=img)
    return img.astype(np.uint8)

plt.figure(figsize=(5,5))


plt.subplot(3,2,1)
plt.imshow(white_balance(img_rgb, 0.0))
plt.axis('off')
plt.title("Cold,value=0.0")

plt.subplot(3,2,2)
plt.imshow(white_balance(img_rgb, 0.25))
plt.axis('off')
plt.title("Cool, value=0.25")

plt.subplot(3,2,3)
plt.imshow(white_balance(img_rgb,0.5))
plt.axis('off')
plt.title("Originalvalue=0.5")

plt.subplot(3,2,4)
plt.imshow(white_balance(img_rgb, 0.75))
plt.axis('off')
plt.title("Warm, value=0.75")

plt.subplot(3,2,5)
plt.imshow(white_balance(img_rgb, 1.0))
plt.axis('off')
plt.title("Hot, value=1.0")
````
