````python
# Part A

def convolve(img, kernel):
  kernel = np.array(kernel)
  a = kernel.shape[0] # Assuming a sqaure kernel matrix
  img = np.array(img)
  img_convolved = np.zeros(img.shape)
  pad = int(a/2)
  img_pad = np.pad(img, ((pad,pad), (pad,pad)), mode = 'reflect')
  H, W = img.shape
  for x in range(H):
    for y in range (W):
      for i in range(-pad, pad+1):
        for j in range (-pad, pad+1):
          img_convolved[x,y] += img_pad[x+i+pad, y+j+pad] * kernel[i+pad, j+pad]
  return img_convolved
````

````python
f_original = np.fft.fft2(gray_img)
fshift_original = np.fft.fftshift(f_original)
magnitude_original = np.abs(fshift_original)
magnitude_spectrum_original = np.log(1 + magnitude_original)

plt.imshow(magnitude_spectrum_original, cmap='gray')
plt.title("Original Magnitude Spectrum")
plt.axis('off')
plt.show()
````

````python
def Gaussian_blur(kernel_size) :
  blur = cv2.GaussianBlur(gray_img, (kernel_size,kernel_size), 0)
  f_gaussian = np.fft.fft2(blur)
  fshift_gaussian = np.fft.fftshift(f_gaussian)
  magnitude_gaussian = np.abs(fshift_gaussian)
  magnitude_spectrum_gaussian = np.log(1 + magnitude_gaussian)

  plt.imshow(magnitude_spectrum_gaussian, cmap='gray')
  plt.title(f'Magnitude Spectrum - Gaussian Blurred, with kernel size = {kernel_size}')
  plt.axis('off')
  plt.show()

blur = cv2.GaussianBlur(gray_img, (3,3), 0)

def unsharp_masking(k):
  unsharp_cv = cv2.addWeighted(gray_img, 1 + k, blur, -k , 0)
  f_unsharp = np.fft.fft2(unsharp_cv)
  fshift_unsharp = np.fft.fftshift(f_unsharp)
  magnitude_unsharp = np.abs(fshift_unsharp)
  magnitude_spectrum_unsharp = np.log(1 + magnitude_unsharp)

  plt.imshow(magnitude_spectrum_unsharp, cmap='gray')
  plt.title(f'Magnitude Spectrum - Unsharp masked, witk k = {k}')
  plt.axis('off')
  plt.show()
````

````python
Gaussian_blur(3)
Gaussian_blur(5)
# We can see that on increasing kernel size, the magnitude spectrum becomes more darker on the outer regions, its because blurring filters out high frequencies and the
# outer regions signify m=higher frequencies in the magnitude spectrum
````

````python
unsharp_masking(0.5)
unsharp_masking(1.0)
unsharp_masking(2.0)
# We can see as we increase k, rhe outer regions of the image become more and more brighter which is expected as Unsharp Masking leads to sharpening which means enhancing
# the intensities of high frequencies
````

````python
def rgb_to_hsv(img):
    img = img.astype(np.float32) / 255.0
    r, g, b = img[:,:,0], img[:,:,1], img[:,:,2]
    mx = np.maximum.reduce([r, g, b])
    mn = np.minimum.reduce([r, g, b])
    diff = mx - mn
    h = np.zeros_like(mx)
    s = np.zeros_like(mx)
    v = mx
    mask = diff != 0
    h[(mx == r) & mask] = (60 * ((g - b) / diff) % 360)[(mx == r) & mask]
    h[(mx == g) & mask] = (60 * ((b - r) / diff) + 120)[(mx == g) & mask]
    h[(mx == b) & mask] = (60 * ((r - g) / diff) + 240)[(mx == b) & mask]
    s[mx != 0] = diff[mx != 0] / mx[mx != 0]
    hsv = np.zeros_like(img)
    hsv[:,:,0] = h / 2
    hsv[:,:,1] = s * 255
    hsv[:,:,2] = v * 255
    return hsv.astype(np.uint8)

hsv = rgb_to_hsv(img_part5)
H, S, V = hsv[:,:,0], hsv[:,:,1], hsv[:,:,2]
mask = (H >= 15) & (H <= 40) & (S > 60) & (V > 50) #  Got this data from google to identify yellow pixels
k = float(input("Enter brightness factor (e.g. 0.3 or 0.5), where 0 is yellow and 1 is red: ")) # Made this input for fetching the need of the user
hsv_new = hsv.copy()
hsv_new[:,:,0][mask] = np.clip((1-k)*hsv_new[:,:,0][mask], 0, 255)

result = cv2.cvtColor(hsv_new, cv2.COLOR_HSV2RGB)
plt.imshow(result)
plt.show()
````
