````python
def rgb_to_hsv(arr):
    arr = arr.astype(np.float32)/255.0
    r, g, b = arr[:, :, 0], arr[:, :, 1], arr[:, :, 2]
    max = np.maximum(np.maximum(r, g), b)
    min = np.minimum(np.minimum(r, g), b)
    delta = max - min
    h = np.zeros_like(r)
    s = np.zeros_like(r)
    v = np.zeros_like(r)
    v = max
    s = np.where(max == 0, 0, delta / max)
    # Cannot comapre arrays directly that is why created boolean masks
    mask_r = (max == r)
    mask_g = (max == g)
    mask_b = (max == b)
    h[mask_r] = (60 * ((g[mask_r] - b[mask_r]) / delta[mask_r]) + 360) % 360
    h[mask_g] = (60 * ((b[mask_g] - r[mask_g]) / delta[mask_g]) + 120) % 360
    h[mask_b] = (60 * ((r[mask_b] - g[mask_b]) / delta[mask_b]) + 240) % 360
    # Setting the hue to be 0 where delta is zero
    h[delta == 0] = 0
    hsv = np.zeros(arr.shape, dtype=np.uint8)
    h = h/2 # For making the hue in range of 0 to 179 for making it OpenCV applicable
    s = s*255 # For making the s values in range of 0 to 255 for making it OpenCV applicable
    v = v*255 # For making the v values in range of 0 to 255 for making it OpenCV applicable
    hsv[:,:,0] = h
    hsv[:,:,1] = s
    hsv[:,:,2] = v
    return hsv
````

````python
# Use the value of bright from 0 to 1

def mood_swing_machine(image, bright, saturation, contrast):
    adjusted = image.astype(np.float32)/255.0
    adjusted = adjusted * contrast + bright
    adjusted = np.clip(adjusted, 0.0, 1.0)

    # Converting to uint8 for OpenCV HSV conversion
    adjusted = (adjusted * 255.0).round().astype(np.uint8)

    # Converting RGB -> HSV (For changing saturation)
    hsv = cv2.cvtColor(adjusted, cv2.COLOR_RGB2HSV).astype(np.float32)

    # Modifying saturation channel
    hsv[:, :, 1] = np.clip(hsv[:, :, 1] * saturation, 0, 255)

    # Converting back to RGB and ensuring uint8
    final = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB)

    # Original
    plt.subplot(1,2,1)
    orig_disp = image
    if orig_disp.dtype != np.uint8:
        orig_disp = np.clip(orig_disp, 0.0, 1.0)
        orig_disp = (orig_disp * 255.0).round().astype(np.uint8)
    plt.imshow(orig_disp)
    plt.title('Original Image')
    plt.axis('off')

    # Processed
    plt.subplot(1,2,2)
    plt.imshow(final)
    plt.title('Mood Swings')
    plt.axis('off')

    plt.tight_layout()
    plt.show()

img2 = np.array(img2)
mood_swing_machine(img2, 0.006, 1.2, 3.0)

# As you can see in the output the mood or the vibe of the image is completely changed, you can use this filter and play with different values of brightness, Saturation
# and contrast to enjoy one image in different moods and vibes.
````

````python
def gamma_for_all(arr, gamma):
    if arr.dtype != np.uint8:
        if arr.max() <= 1.0:
            arr_uint8 = (arr * 255).astype(np.uint8)
        else:
            arr_uint8 = arr.astype(np.uint8)
    else:
        arr_uint8 = arr
    table = np.array([(i/255.0)**gamma*255 for i in range(256)]).astype(np.uint8)
    final_u8 = cv2.LUT(arr_uint8, table)

    # Original
    plt.subplot(1,2,1)
    orig_disp = arr
    if orig_disp.dtype != np.uint8:
        orig_disp = np.clip(orig_disp, 0.0, 1.0)
        orig_disp = (orig_disp * 255.0).round().astype(np.uint8)
    plt.imshow(orig_disp)
    plt.title('Original Image')
    plt.axis('off')

    # Processed
    plt.subplot(1,2,2)
    plt.imshow(final_u8)
    plt.title(f'Gamma = {gamma}')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
````

````python

from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np

# PASTE BASE64 STRING HERE
base64_image_white2 = """""" # long base64 string

img_data_white2 = base64.b64decode(base64_image_white2)
img_white2 = Image.open(io.BytesIO(img_data_white2))

plt.imshow(img_white2)
plt.axis("off")

img_white2 = np.array(img_white2)
````

````python
def white_patch_balance(img_rgb):
    new = img_rgb.astype(np.float32)
    if new.max() > 1.0:
        new /= 255.0

    # Splitting channels
    r, g, b = new[:,:, 0], new[:,:, 1], new[:,:, 2]

    # Finding the brightest pixel in each channel
    r_max = np.max(r)
    g_max = np.max(g)
    b_max = np.max(b)

    # For avoiding division by zero in the scaling
    r_max = max(r_max, 1)
    g_max = max(g_max, 1)
    b_max = max(b_max, 1)

    # Scaling each channel=
    r_scaled = r / r_max
    g_scaled = g / g_max
    b_scaled = b / b_max

    # Merging and clipping
    out = np.stack([r_scaled, g_scaled, b_scaled], axis=-1)
    out = np.clip(out, 0.0, 1.0)

    # Converting back to uint8
    out = (out * 255).astype(np.uint8)

    return out

````

````python
# First image

output_white = white_patch_balance(img_white)
plt.subplot(1,2,1)
orig_disp = img_white
if orig_disp.dtype != np.uint8:
        orig_disp = np.clip(orig_disp, 0.0, 1.0)
        orig_disp = (orig_disp * 255.0).round().astype(np.uint8)
plt.imshow(orig_disp)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(output_white)
plt.title('White patch balanced image')
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
def white_balance(img, value):
    v = max(0.0, min(1.0, float(value)))
    a = img.astype(np.float32)
    if a.max() <= 1.0:
        a *= 255.0

    pr = a[:,:,0].max()
    pg = a[:,:,1].max()
    pb = a[:,:,2].max()
    P = np.array([pr, pg, pb], dtype=np.float32)
    P = np.maximum(P, 1e-3)

    base = P.mean() # Used the means of maximums for scaling later
    cold = np.array([0.6, 0.95, 1.8], dtype=np.float32) # Enhanced blue more, decreased red and decreased green a bit
    warm = np.array([1.8, 1.05, 0.6], dtype=np.float32) # Enhanced red more, decreased blue and increased green slightly

    gc = (base * cold) / P # Cold gains
    gw = (base * warm) / P # warm gains

    # Implementing the gains
    if v <= 0.5:
        t = 1.0 - (v / 0.5)
        gains = gc * t + (1.0 - t) * np.array([1.0,1.0,1.0],dtype=np.float32)
    else:
        t = (v - 0.5) / 0.5
        gains = np.array([1.0,1.0,1.0],dtype=np.float32) * (1.0 - t) + gw * t

    gains = np.clip(gains, 0.5, 2.5) # For avoiding tremendous or ver low gain due to outlier pixels
    out = a * gains.reshape(1,1,3)
    return np.clip(out, 0, 255).astype(np.uint8)

output = white_balance(img_gamma1, 1) # For producing the warmest image

# Plots
plt.subplot(1,2,1)
orig_disp = img_gamma1
if orig_disp.dtype != np.uint8:
        orig_disp = np.clip(orig_disp, 0.0, 1.0)
        orig_disp = (orig_disp * 255.0).round().astype(np.uint8)
plt.imshow(orig_disp)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(output)
plt.title('White balanced')
plt.axis('off')

plt.tight_layout()
plt.show()
````
