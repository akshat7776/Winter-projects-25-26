````python
#Creating a function to apply filter and reconstruct image and compare results

def apply_filter_lpf(fft_shift,radius):

  #Creating Low-Pass Mask

  rows, cols = gray_scale_cv.shape
  crows, ccols = rows//2, cols//2

  mask_lpf = np.zeros((rows,cols), np.uint8)
  cv2.circle(mask_lpf,(ccols,crows), radius, 1, -1)

  #Applying masks using convolutions

  frequencies_lpf = fft_shift*mask_lpf

  #Inverting shift to convert from frequencies to images

  inverse_shift_lpf = np.fft.ifftshift(frequencies_lpf)
  inverse_image_lpf = np.fft.ifft2(inverse_shift_lpf)
  inverse_image_lpf = np.abs(inverse_image_lpf)

  return inverse_image_lpf.astype(np.float32)

def compare(image_1,image_2,image_3):
  plt.figure(figsize=(20,10))
  plt.subplot(1,3,1)
  plt.imshow(image_1,cmap="grey")
  plt.title("Radius 10")

  plt.subplot(1,3,2)
  plt.imshow(image_2,cmap="grey")
  plt.title("Radius 30")

  plt.subplot(1,3,3)
  plt.imshow(image_3,cmap="grey")
  plt.title("Radius 80")

compare(apply_filter_lpf(fft_shift,10),apply_filter_lpf(fft_shift,50),apply_filter_lpf(fft_shift,80))


````

````python
#Creating a function to apply filter and reconstruct image and compare results

def apply_filter_hpf(fft_shift,radius):

  #Creating High-Pass Mask

  rows, cols = gray_scale_cv.shape
  crows, ccols = rows//2, cols//2

  mask_hpf = np.ones((rows,cols), np.uint8)
  cv2.circle(mask_hpf,(ccols,crows), radius, 0, -1)

  #Applying masks using convolutions

  frequencies_hpf = fft_shift*mask_hpf

  #Inverting shift to convert from frequencies to images

  inverse_shift_hpf = np.fft.ifftshift(frequencies_hpf)
  inverse_image_hpf = np.fft.ifft2(inverse_shift_hpf)
  inverse_image_hpf = np.abs(inverse_image_hpf)

  return inverse_image_hpf.astype(np.float32)

compare(apply_filter_hpf(fft_shift,10),apply_filter_hpf(fft_shift,50),apply_filter_hpf(fft_shift,80))


````

````python
#Creating Square LPF

rows,cols = gray_scale_cv.shape
crows,ccols = rows//2,cols//2

side=50

#We need zeroes outside the 50x50 sqaure mask and 1's inside it to convolve over the image

square_lpf = np.zeros((rows,cols), np.float32)

half_length = side//2

square_lpf[crows-half_length:crows+half_length,ccols-half_length:ccols+half_length] = 1
````

````python
#Checking if we have a correct mask

plt.imshow(square_lpf, cmap="gray")
plt.title("Square LPF Mask (50x50)")
````

````python
#We do indeed!!!
#Convolving over image

frequencies_square_lpf = fft_shift*square_lpf

img_sq_lpf = np.fft.ifft2(np.fft.ifftshift(frequencies_square_lpf))
img_sq_lpf = np.abs(img_sq_lpf)

def compare_lpf(image_1,image_2,image_3):
  plt.figure(figsize=(20,10))
  plt.subplot(1,3,1)
  plt.imshow(image_1,cmap="grey")
  plt.title("Original Image")

  plt.subplot(1,3,2)
  plt.imshow(image_2,cmap="grey")
  plt.title("Circular mask")

  plt.subplot(1,3,3)
  plt.imshow(image_3,cmap="grey")
  plt.title("Square Mask")

compare_lpf(gray_scale_cv,apply_filter_lpf(fft_shift,50),img_sq_lpf)
````

````python
phase_only = fft_shift/np.abs(fft_shift)

#Undoing shift

inverse_shift = np.fft.ifftshift(phase_only)

#Inverting shift

img_phase = np.fft.ifft2(inverse_shift)
img_phase = np.abs(img_phase)

plt.figure(figsize=(10,10))
plt.imshow(img_phase,cmap="gray")
plt.title("Phase Only Reconstruction")
````

The main difference between np.angle(f_shift) and phase_only is that the former returns phase(Φ) as a 2d array in [F = |F|e^(iΦ)] and the latter returns complex frequency coefficients e^(iΦ) i.e it maintains structure.

````python
#Reconstructing from np.angle(f_shift) by obtaining complex coefficients

complex_coefficients = np.exp(1j*np.angle(fft_shift))

#Repeating similar procedure

inverse = np.fft.ifftshift(complex_coefficients)
inverse_img = np.fft.ifft2(inverse)
inverse_img = np.abs(inverse_img)

plt.figure(figsize=(10,10))
plt.imshow(inverse_img,cmap="gray")
plt.title("Phase Reconstruction")
````

````python
#Magnitude only reconstruction

mag_only = np.abs(fft_shift)

#Undoing shift

inverse_mag = np.fft.ifftshift(mag_only)
mag_img = np.fft.ifft2(inverse_mag)
mag_img = np.abs(mag_img)

plt.figure(figsize=(10,10))
plt.imshow(mag_img,cmap="gray")
plt.title("Magnitude Reconstruction")
````

It shows a solid black block as my image has very small variations in intensity.

The magnitudes of magnitudes conatins very large values and variation. It has high values near the center where frequency is low and low value near edges where frequency is high. Taking log scales values and variations appropriately.

Here we used the raw magnitudes for reconstruction, in the magnitude spectrum we used their logarythms. That's why it looks different

````python
def difference(original_img,magnitude,mag_reconstruction):
  plt.figure(figsize=(20,10))
  plt.subplot(1,3,1)
  plt.imshow(original_img)
  plt.title("Original Image")
  plt.axis("off")

  plt.subplot(1,3,2)
  plt.imshow(magnitude,cmap="grey")
  plt.title("Magnitude Spectrum")
  plt.axis("off")

  plt.subplot(1,3,3)
  plt.imshow(mag_reconstruction,cmap="grey")
  plt.title("Magnitude reconstruction")
  plt.axis("off")

difference(img,magnitude,mag_img)
````
