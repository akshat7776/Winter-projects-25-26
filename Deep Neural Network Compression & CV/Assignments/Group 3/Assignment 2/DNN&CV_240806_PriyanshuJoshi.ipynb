````python
def gray_histogram(gray_img):
    # Create array of 256 bins (0–255)
    hist = np.zeros(256, dtype=int)

    rows, cols = gray_img.shape
    for i in range(rows):
        for j in range(cols):
            pixel = gray_img[i, j]
            hist[pixel] += 1

    return hist

hist = gray_histogram(gray_img)

plt.figure(figsize=(10, 4))
plt.bar(np.arange(256), hist, width=1, color='black')
plt.title("Grayscale Histogram")
plt.xlabel("Intensity (0–255)")
plt.ylabel("Pixel Count")
plt.show()
````

Write a function <code>rgb_to_hsv</code> that takes a 0-255 ( not normalised ) RGB image and returns an HSV image ( also not normalised ).

````python
import numpy as np

def rgb_to_hsv(img_rgb):

    # Convert to float for calculation
    img = img_rgb.astype(np.float32)
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]

    # Max and min per pixel
    maxc = np.max(img, axis=-1)
    minc = np.min(img, axis=-1)
    delta = maxc - minc

    # Hue calculation
    H = np.zeros_like(maxc)

    # Avoid division by zero
    mask = delta != 0

    # When max = R
    idx = (maxc == R) & mask
    H[idx] = (60 * ((G[idx] - B[idx]) / delta[idx]) + 360) % 360

    # When max = G
    idx = (maxc == G) & mask
    H[idx] = (60 * ((B[idx] - R[idx]) / delta[idx]) + 120) % 360

    # When max = B
    idx = (maxc == B) & mask
    H[idx] = (60 * ((R[idx] - G[idx]) / delta[idx]) + 240) % 360

    # OpenCV scales H from 0–360 → 0–179
    H = (H / 2).astype(np.uint8)

    # ---- Saturation ----
    S = np.zeros_like(maxc)
    nonzero_mask = maxc != 0
    S[nonzero_mask] = (delta[nonzero_mask] / maxc[nonzero_mask]) * 255
    S = S.astype(np.uint8)

    # ---- Value ----
    V = maxc.astype(np.uint8)

    # Combine into HSV image (uint8)
    hsv = np.stack([H, S, V], axis=-1)
    return hsv
````

````python
# img_rgb is original RGB image (0–255)
hsv_custom = rgb_to_hsv(img_rgb)

# Converting back to RGB using OpenCV
rgb_back = cv2.cvtColor(hsv_custom, cv2.COLOR_HSV2RGB)

#Comparison
difference=cv2.absdiff(img_rgb,rgb_back)

# Show both images
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(rgb_back)
plt.title("RGB after HSV→RGB using cv2")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(difference)
plt.title("Comparing results")
plt.axis("off")

plt.show()
````

````python
def clamp_uint8(img):
    """Clip to 0-255 and convert to uint8."""
    return np.clip(img, 0, 255).astype(np.uint8)

def adjust_gamma(img, gamma=1.0):
    """Gamma correction on image in float [0,255]."""
    inv = 1.0 / gamma
    lut = np.array([((i / 255.0) ** inv) * 255 for i in np.arange(256)]).astype("float32")
    # apply LUT channel-wise
    out = cv2.LUT(img.astype(np.uint8), lut.astype(np.uint8))
    return out.astype(np.float32)


# Filter 1: FrostySunrise
# Name: "FrostySunrise" (icy, warm highlight boost)
# Description:
#   - Contrast boost (alpha), small brightness lift (beta)
#   - Split-toning: cool-blue tint to shadows, warm-orange tint to highlights
#   - Vibrance-like saturation increase (boost low-sat pixels more)
#   - Small gamma tweak for midtones and clarity via unsharp mask

def FrostySunrise(img_rgb):
    # Work on float copy
    img = img_rgb.astype(np.float32)

    # 1) Contrast & brightness (linear): out = alpha*img + beta
    # (implemented as contrast boost to make highlights pop a bit)
    alpha = 1.08  # contrast
    beta = 6.0    # brightness shift
    # --> Contrast modification
    img = img * alpha + beta

    # 2) Split-toning (shadows -> bluish, highlights -> warm)
    img_hsv = cv2.cvtColor(clamp_uint8(img), cv2.COLOR_RGB2HSV).astype(np.float32)
    H = img_hsv[..., 0]
    S = img_hsv[..., 1]
    V = img_hsv[..., 2]

    # Masks for shadows and highlights
    shadow_mask = (V < 90).astype(np.float32)  # shadows
    mid_mask = ((V >= 90) & (V < 160)).astype(np.float32)
    highlight_mask = (V >= 160).astype(np.float32)

    # Converting original rgb to float for tinting
    img_t = img.copy()

    # Cool tint for shadows: add to B channel, reduce R a bit
    # --> Blue boost in shadows
    img_t[..., 2] += 18 * shadow_mask  # add blue in shadows (B channel idx 2 in RGB)
    img_t[..., 0] -= 6 * shadow_mask   # slightly reduce R in shadows

    # Warm tint for highlights: boost R and G in highlights
    img_t[..., 0] += 18 * highlight_mask  # R
    img_t[..., 1] += 10 * highlight_mask  # G

    # 3) Vibrance-like saturation: increase saturation more for low-saturation pixels
    hsv2 = cv2.cvtColor(clamp_uint8(img_t), cv2.COLOR_RGB2HSV).astype(np.float32)
    H2 = hsv2[..., 0]
    S2 = hsv2[..., 1]
    V2 = hsv2[..., 2]
    # low-sat -> stronger boost
    sat_boost = (1.0 - (S2 / 255.0)) * 0.55
    S2 = S2 + (sat_boost * 255.0)
    S2 = np.clip(S2, 0, 255)
    hsv2[..., 1] = S2

    # 4) Small gamma correction to lift midtones slightly (gamma <1 -> brighten midtones)
    rgb_after_sat = cv2.cvtColor(hsv2.astype(np.uint8), cv2.COLOR_HSV2RGB).astype(np.float32)
    # gamma correction
    rgb_gamma = adjust_gamma(rgb_after_sat.astype(np.uint8), gamma=0.95).astype(np.float32)

    # 5) Unsharp mask / clarity: add a small high-pass to sharpen subtle detail
    blurred = cv2.GaussianBlur(rgb_gamma, (0,0), sigmaX=2.0, sigmaY=2.0)
    highpass = rgb_gamma - blurred
    rgb_sharp = rgb_gamma + 0.8 * highpass

    # Final tone mapping clamp
    out = clamp_uint8(rgb_sharp)
    return out

# Filter 2: TropicalPunch
# Name: "TropicalPunch" (vivid, warm, punchy)
# Description:
#   - Strong saturation increase
#   - Hue rotation (slight warm shift)
#   - Channel mixing (push reds/oranges, reduce blue a bit)
#   - Contrast (CLAHE on V channel) + gamma correction

def TropicalPunch(img_rgb):
    img = img_rgb.astype(np.float32)

    # 1) Channel mixing (pre-tone): boost R and G slightly from other channels
    # Implement a simple mixing matrix (RGB_out = M @ RGB_in)
    M = np.array([[1.12, 0.02, -0.02],   # more red
                  [0.02, 1.05, 0.00],    # slightly more green
                  [-0.03, 0.01, 0.95]])  # reduce blue slightly
    # Apply per-pixel
    shaped = img.reshape((-1,3)).dot(M.T).reshape(img.shape)

    # 2) Converting to HSV
    hsv = cv2.cvtColor(clamp_uint8(shaped), cv2.COLOR_RGB2HSV).astype(np.float32)
    H = hsv[..., 0]
    S = hsv[..., 1]
    V = hsv[..., 2]

    # 3) Increase saturation strongly
    S = S * 1.25
    S = np.clip(S, 0, 255)

    # 4) Hue rotation: shift hue towards warmer tones (+8 degrees)
    hue_shift_degrees = 8
    hue_shift_units = hue_shift_degrees / 2.0  # scale to 0-179
    H = (H + hue_shift_units) % 180

    hsv[..., 0] = H
    hsv[..., 1] = S
    hsv[..., 2] = V

    # 5) Stronger contrast via CLAHE on V channel
    V_uint8 = clamp_uint8(hsv[..., 2])
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    V_clahe = clahe.apply(V_uint8)
    hsv[..., 2] = V_clahe.astype(np.float32)

    # 6) Gamma correction
    rgb_after = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB).astype(np.float32)
    rgb_gamma = adjust_gamma(rgb_after.astype(np.uint8), gamma=1.08).astype(np.float32)

    out = clamp_uint8(rgb_gamma)
    return out


img_path = "Image.jpeg"
img_pil = cv2.imread(img_path)
if img_pil is None:
    raise FileNotFoundError(f"Image not found at {img_path}")
img_rgb = cv2.cvtColor(img_pil, cv2.COLOR_BGR2RGB)

# Applying filters
out1 = FrostySunrise(img_rgb)
out2 = TropicalPunch(img_rgb)

# -------------------------
# Displaying original + outputs
# -------------------------
plt.figure(figsize=(18,8))
plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Original (RGB)")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(out1)
plt.title("FrostySunrise")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(out2)
plt.title("TropicalPunch")
plt.axis('off')

plt.tight_layout()
plt.show()

# -------------------------
# Short notes
# -------------------------
note1 = (
    "FrostySunrise: I boosted contrast and added a cool tint to shadows while warming highlights "
    "to preserve the cold mood but make sunlight-like areas pop—this suits snow/ice dawn scenes."
)
note2 = (
    "TropicalPunch: I used channel mixing, strong saturation, hue shift and CLAHE-based contrast "
    "to create a bold, warm, punchy look—good for beach/sunset/foliage images."
)
print("Notes:\n1)", note1, "\n\n2)", note2)
````

Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image.
````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def white_patch_balance(img_rgb):
    img = img_rgb.astype(np.float32)

    # Find per-channel maxima
    max_per_chan = np.max(img, axis=(0,1)).astype(np.float32)

    # Avoid division by zero (if a channel is all zeros)
    eps = 1e-8
    scales = np.where(max_per_chan > eps, 255.0 / max_per_chan, 1.0)

    # Apply per-channel scaling
    balanced = img * scales.reshape((1,1,3))

    # Clip & convert to uint8
    balanced_uint8 = np.clip(balanced, 0, 255).astype(np.uint8)

    return balanced_uint8, max_per_chan, scales

img_bgr = cv2.imread("Image.jpeg")
if img_bgr is None:
    raise FileNotFoundError("Could not find /mnt/data/Image.jpeg. Make sure the file is present.")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

balanced_img, max_vals, scales = white_patch_balance(img_rgb)

print("Per-channel maxima (R, G, B):", max_vals)
print("Per-channel scales (R, G, B):", scales)

# Displaying side-by-side
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original (RGB)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(balanced_img)
plt.title("White-Patch Balanced (RGB)")
plt.axis("off")

plt.tight_layout()
plt.show()
````

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def clamp_uint8(img):
    return np.clip(img, 0, 255).astype(np.uint8)

def white_balance(img_rgb, value):
    assert 0.0 <= value <= 1.0

    # Converting to float
    img = img_rgb.astype(np.float32)

    # factor: -1 (cold) ... 0 (original) ... +1 (warm)
    factor = (value - 0.5) * 2.0

    # Base warm_delta: how much to change channels at factor=+1 (warmest)
    # Positive -> multiply channel more, Negative -> multiply channel less
    # (Order: R, G, B)
    warm_delta = np.array([0.55, 0.18, -0.45], dtype=np.float32)
    # At factor=-1, we get the inverse (cooling) effect

    # Computing per-channel scale factors
    # scale = 1 + factor * warm_delta
    scales = 1.0 + factor * warm_delta
    # Prevent negative or zero scales
    scales = np.clip(scales, 0.01, 3.0)

    # Applying channel-wise scaling
    out = img * scales.reshape((1,1,3))

    # Converting to HSV to tweak saturation/value for a smoother effect
    out_uint8 = clamp_uint8(out)
    hsv = cv2.cvtColor(out_uint8, cv2.COLOR_RGB2HSV).astype(np.float32)
    H = hsv[..., 0]
    S = hsv[..., 1]
    V = hsv[..., 2]

    # Saturation tweak: warm -> slightly more saturation, cool -> slightly less
    # small factor so it doesn't oversaturate
    sat_scale = 1.0 + factor * 0.22   # range roughly 0.78 .. 1.22
    S = S * sat_scale
    S = np.clip(S, 0, 255)

    # Brightness tweak: warm -> brighten midtones a touch, cold -> slight darken
    val_scale = 1.0 + factor * 0.04   # gentle change
    V = V * val_scale
    V = np.clip(V, 0, 255)

    hsv[..., 1] = S
    hsv[..., 2] = V

    # Convert back to RGB
    rgb_out = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB).astype(np.float32)

    # Applying a subtle global curve to preserve contrast (light contrast boost)
    alpha = 1.02  # mild contrast
    beta = 0.0
    rgb_out = rgb_out * alpha + beta

    return clamp_uint8(rgb_out)


img_bgr = cv2.imread("Image.jpeg")
if img_bgr is None:
    raise FileNotFoundError("/mnt/data/Image.jpeg not found in container.")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

vals = [0.0, 0.5, 1.0]   # coldest, original (identity), warmest
titles = ["Cold (value=0.0)", "Original (value=0.5)", "Warm (value=1.0)"]
outs = [white_balance(img_rgb, v) for v in vals]

plt.figure(figsize=(15,6))
plt.subplot(1,4,1)
plt.imshow(img_rgb)
plt.title("Original (input)")
plt.axis("off")

for i, out in enumerate(outs, start=2):
    plt.subplot(1,4,i)
    plt.imshow(out)
    plt.title(titles[i-2])
    plt.axis("off")

plt.tight_layout()
plt.show()
````
