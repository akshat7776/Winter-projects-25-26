````python
#Converting into the grayscale image
img = np.array(img)
red = img[:,:,0]
green = img[:,:,1]
blue = img[:,:,2]
gray_img = (red*0.299) + (green*0.587) + (blue*0.114)
gray_img = gray_img.astype(np.uint8)
plt.figure(figsize=(5,5))
plt.title("Grayscale Using Maths")
plt.imshow(gray_img,cmap='gray')
plt.axis("off")
````

````python
#Making the histogram function

def gray_histogram(gray_img):
  counter = np.zeros(256,dtype=int)
  row,col = gray_img.shape

  for i in range(row):
    for j in range(col):
      counter[gray_img[i,j]] += 1

  return counter

#Plotting the histogram of the gray image

gray_histo = gray_histogram(gray_img)

plt.figure(figsize=(5,5))
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Values")
plt.ylabel("Number of Pixels")
plt.bar(range(256),gray_histo,color='gray',width=2)
plt.show()
````

````python
#We will make a RGb to HSV function
#The value(V) in the HSV is defined as brightness/lightness of the pixel so we basically have to see the max value out of R,G,B for that pixel
#The saturation is how vivid/pure a colour is for this we will have to define delta = max-min of the rgb values and then divide it by value, and if value=0 then saturaion is also 0
#Hue is the purest form of the color and is measured as angle moved from red color. for this we will see which of r,g,b is max and then based of that we will asign angles. if red is max we will have 60*(g-b)/delta obv mod 360 and then similarily if green is max 60*(b-r/delta + 2) and if blue is max 60*(r-g/delta + 4).
#To make it opencv compatible we will have to divide by 2 as open cv uses 8 bit integer to store and 360 cant be store in just 8 bit inrtgere so they make it 0-179 thus finally we will divide by 2.

#I originally made a function using for loops but the for loop method was too slow so instead used vectorization
"""def rgb_to_hsv(img):
  row,col,channel = img.shape
  hsv_img = np.zeros_like(img)
  for i in range(row):
    for j in range(col):
      r = img[i,j,0]/255.0
      g = img[i,j,1]/255.0
      b = img[i,j,2]/255.0

      maxi = max(r,g,b)
      mini = min(r,g,b)
      delta = maxi-mini

      #Hue
      h= 0.0
      if delta==0.0:
        h=0
      elif maxi==r:
        h=60*(g-b)/delta
        h %= 360
      elif maxi==g:
        h = 60*(b-r)/delta + 120
      elif maxi==b:
        h = 60*(r-g)/delta + 240

      #Saturation
      s=0.0
      if maxi==0.0:
        s=0.0
      else:
        s = delta/maxi

      #VAlue
      v = maxi

      #Final changes and scaling (hue is halved due to reason explained above (360 cant be store d in 8-bit))
      h = int(h/2)
      s = int(s*255)
      v = int(v*255)


      hsv_img[i,j] = [h,s,v]

  return hsv_img
"""
import cv2

def rgb_to_hsv(img):
    img = img.astype(np.float32) / 255.0

    r = img[:, :, 0]
    g = img[:, :, 1]
    b = img[:, :, 2]

    maxi = np.max(img, axis=2)
    mini = np.min(img, axis=2)
    delta = maxi-mini

    #Value
    v = maxi
    #saturation
    s = np.zeros_like(v)
    non_zero_v = maxi != 0
    s[non_zero_v] = delta[non_zero_v] / maxi[non_zero_v]
    #Hue
    h = np.zeros_like(v)
    delta_nz = delta != 0

    mask_r = (maxi == r) & delta_nz
    h[mask_r] = ( (g[mask_r] - b[mask_r]) / delta[mask_r] ) % 6

    mask_g = (maxi == g) & delta_nz
    h[mask_g] = ( (b[mask_g] - r[mask_g]) / delta[mask_g] ) + 2

    mask_b = (maxi == b) & delta_nz
    h[mask_b] = ( (r[mask_b] - g[mask_b]) / delta[mask_b] ) + 4


    h = h * 60
    h = h/2
    s = s * 255
    v = v * 255
    hsv_out = np.dstack((h, s, v))

    return hsv_out.astype(np.uint8)

hsv_image = rgb_to_hsv(img)

reconvertingtorgb = cv2.cvtColor(hsv_image,cv2.COLOR_HSV2RGB)

plt.figure(figsize=(20,20))
plt.subplot(1, 3, 1)
plt.title("Original RGB")
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("HSV image using my function")
plt.imshow(hsv_image)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("RGB to HSV using my function then reconverted to RGB using cv2")
plt.imshow(reconvertingtorgb)
plt.axis('off')
plt.show()
````

````python
#This is a sabrina carpenter's espresso music video inspired espresso filter, it makes the image very warm, add yeloow tint by adding to the red and green while subtracting from the blue and it also increases contrast and increases saturation.
#ESPRESSO FILTER(Warm,Sunny Vibes)
def espresso_filter(img):
  output = img.copy()

  hsv_img = rgb_to_hsv(output)
  h = hsv_img[:,:,0]
  s = hsv_img[:,:,1]
  v = hsv_img[:,:,2]

  s_float = s.astype(np.float32)
  s_float *= 1.25        # THis is a saturation boost
  s = np.clip(s_float,0,255).astype(np.uint8)

  hsv_changed = np.dstack((h,s,v))
  output = cv2.cvtColor(hsv_changed,cv2.COLOR_HSV2RGB)
  output = output.astype(np.float32)

  output[:,:,0] += 60.0
  output[:,:,1] += 20.0           #These three lines make the image warmer by changing the tints pushing the red and green tints and supressing the bluish(cold) tints
  output[:,:,2] -= 40.0
  output = np.clip(output,0,255)

  output *= 1.10           #this increases brightness thus giving even more warmer/sunny vibes

  output = np.clip(output,0,255).astype(np.uint8)
  return output

espresso_img = espresso_filter(img)

plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Espresso Filter")
plt.imshow(espresso_img)
plt.axis('off')
````

````python
#This filter takes the complete opposite approach to my previous filter this makes the whole image more darkish, glooomy, cyberpunkesque and gothic vibe. It uses gamma corection and we tone down the warm saturations and tone down the red (warm)values while increasing the green and blue values.
#GOTHIC FILTER (Gloomy,Dark Vibes)
def gothic_filter(img):
  output = img.copy()
  output = output.astype(np.float32)

  output = (output-128.0)*1.25 + 128.0    # stretches difference between light and dark pixels that is increases contrast
  output = np.clip(output,0,255)

  output = 255.0*np.power(output/255.0,1.8)      # this is a gamma correction, which is non linear brightness adjustment
  output = np.clip(output,0,255)

  output[:,:,0] -= 60
  output[:,:,1] += 10     # these three lines shifts the colour balances that is the tint, it reduces red and increases green a bit and blue very much
  output[:,:,2] += 20

  output = np.clip(output,0,255).astype(np.uint8)
  return output

gothic_img = gothic_filter(img)

plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Gothic Filter")
plt.imshow(gothic_img)
plt.axis('off')
````

#### White Balance Function
Write a function <code>white_patch_balance(img_rgb)</code> that takes a 0-255 RGB image and then returns a White Patch Method Balanced 0-255 RBG image. Use Numpy only to keep track of the brightest pixel and then scale the entire image with the factor you find. Display the original and output image.

````python
#In white correction our rule is that the brightest pixel in the wjole image should be pure white and then we rescale all the other pixels accordingly.

def white_patch_balance(img):
  img_float = img.astype(np.float32)

  maxi = np.max(img_float,axis=(0,1))

  scaling_factor = 255.0 / (maxi + 1e-6)  #1e-6 is used to handle the case where the image contains a pixel which is pure black(0) and we cant divide by 0

  output = img_float*scaling_factor
  output = np.clip(output,0,255).astype(np.uint8)

  return output

white_balanced_img = white_patch_balance(img)

plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("White Patch Balanced Image")
plt.imshow(white_balanced_img)
plt.axis('off')
````

````python
##Here will try to recreate the white balance slider
#We try to give the slider effect by choosing 5 point from 0 to 1. i.e 0,0.25,0.5,0.75 and 1. And we go from cold to cool to neutral to warm to hot signifying the changes from very bluish to very warmish image.

def white_balance(img,value):
  img_float = img.astype(np.float32)

  deviation = value - 0.5       # sees how far from the middle/neutral value is the given input value
  strength = 0.5       # we can change teh strength ourselves depending on our liking

  r_scale = 1.0
  b_scale = 1.0

  if deviation < 0:
    multipl = np.abs(deviation)*2
    b_scale += (multipl*strength)
    r_scale -= (multipl*strength)

  elif deviation > 0:
    multipl = deviation *2
    b_scale -= (multipl*strength)
    r_scale += (multipl*strength)

  img_float[:,:,0] *= r_scale
  img_float[:,:,2] *= b_scale

  image = np.clip(img_float,0,255).astype(np.uint8)
  return image

cold_img = white_balance(img,0.0)
cool_img = white_balance(img,0.25)
neutral_img = white_balance(img,0.50)
warm_img = white_balance(img,0.75)
hot_img = white_balance(img,1.0)

plt.figure(figsize=(50,50))
plt.subplot(1, 5, 1)
plt.title("Cold Image, Value = 0")
plt.imshow(cold_img)
plt.axis('off')

plt.subplot(1, 5, 2)
plt.title("Cool Image,Value =0.25")
plt.imshow(cool_img)
plt.axis('off')

plt.subplot(1, 5, 3)
plt.title("Neutral/Original Image, Value = 0.50")
plt.imshow(neutral_img)
plt.axis('off')

plt.subplot(1, 5, 4)
plt.title("warm Image,Value = 0.75")
plt.imshow(warm_img)
plt.axis('off')

plt.subplot(1, 5, 5)
plt.title("Hot Image,Value = 1.0")
plt.imshow(hot_img)
plt.axis('off')

plt.show()
````
