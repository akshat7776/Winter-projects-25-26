C1

````python
import numpy as np
import matplotlib.pyplot as plt

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2  # center of frequency domain

# Radii for Low-Pass Filters
radii = [10, 30, 80]

lp_results = []

for r in radii:
    # Create coordinate grid
    Y, X = np.ogrid[:rows, :cols]

    # Distance of each point from the center
    dist_from_center = np.sqrt((X - ccol)**2 + (Y - crow)**2)

    # LPF mask:
    mask = dist_from_center <= r   

    # Apply mask to shifted FFT
    fshift_filtered = fshift * mask

    # Inverse shift and inverse FFT to reconstruct spatial image
    f_ishift = np.fft.ifftshift(fshift_filtered)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    lp_results.append((r, img_back))

plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

for i, (r, img_back) in enumerate(lp_results, start=2):
    plt.subplot(2, 2, i)
    plt.imshow(img_back, cmap='gray')
    plt.title(f"LPF radius = {r}")
    plt.axis('off')

plt.tight_layout()
plt.show()

''' • As the radius decreases, the filter keeps *fewer* high-frequency components.
    • High frequencies correspond to edges and fine details.
    • So smaller radius -> more high frequencies removed -> image becomes *more blurred*. '''
````

C2

````python
import numpy as np
import matplotlib.pyplot as plt

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2  # center

# Radius for HPF
r = 30

# Create coordinate grid
Y, X = np.ogrid[:rows, :cols]

# Distance from center
dist_from_center = np.sqrt((X - ccol)**2 + (Y - crow)**2)

# High-pass filter mask:
# 1 outside the circle (high frequencies), 0 inside (low frequencies removed)
hpf_mask = dist_from_center >= r

# Apply mask to shifted FFT
fshift_hpf = fshift * hpf_mask

# Inverse shift + inverse FFT to reconstruct
f_ishift_hpf = np.fft.ifftshift(fshift_hpf)
img_hpf = np.fft.ifft2(f_ishift_hpf)
img_hpf = np.abs(img_hpf)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(hpf_mask, cmap='gray')
plt.title("HPF Mask")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_hpf, cmap='gray')
plt.title("After HPF (edges enhanced)")
plt.axis('off')

plt.tight_layout()
plt.show()

'''Reason: 
        •HPF keeps high frequencies (edges, fine details) and removes low frequencies (smooth areas).
        •So the output looks sharper and more edge-emphasized, sometimes with halo/outline effects. '''
````

````python
import numpy as np
import matplotlib.pyplot as plt

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2  # center of frequency domain

# Side length of the square LPF
side = 50  
half = side // 2

# Create an all-zero mask
square_lpf_mask = np.zeros((rows, cols), dtype=np.uint8)


# Defining square region at the center and set it to 1 (pass low frequencies)
square_lpf_mask[crow - half : crow + half, ccol - half : ccol + half] = 1

# Apply square LPF mask
fshift_square = fshift * square_lpf_mask

# Reconstruct image: inverse shift -> inverse FFT
f_ishift_square = np.fft.ifftshift(fshift_square)
img_square_lpf = np.fft.ifft2(f_ishift_square)
img_square_lpf = np.abs(img_square_lpf)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(square_lpf_mask, cmap='gray')
plt.title("Square LPF Mask (50×50)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_square_lpf, cmap='gray')
plt.title("After Square LPF")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python

# 1) Creating a safe phase-only spectrum:
magnitude = np.abs(fshift)
eps = 1e-12
safe_magnitude = np.where(magnitude == 0, eps, magnitude)

phase_only = fshift / np.abs(fshift) 

# Reconstructing from phase-only:
f_ishift_phase = np.fft.ifftshift(phase_only)
img_phase_only = np.fft.ifft2(f_ishift_phase)
img_phase_only = np.abs(img_phase_only)  # final spatial-domain image


#2) Showing the phase-only reconstruction
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(1,3,2)

# Magnitude spectrum (log scale)
mag_spectrum = np.log(np.abs(fshift) + 1)
plt.imshow(mag_spectrum, cmap='gray')
plt.title("Magnitude Spectrum (log)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_phase_only, cmap='gray')
plt.title("Reconstruction from phase-only")
plt.axis('off')
plt.tight_layout()
plt.show()
````

````python
# 3) Difference between 'fshift / |fshift|' and 'np.angle(fshift)'
# - fshift / |fshift| returns a complex array with unit magnitude and the phase encoded
#   as a complex exponential, i.e. e^{i*theta}. Its dtype is complex.
# - np.angle(fshift) returns the phase *angle values* (real numbers in radians).
# So:
#   phase_only_complex == fshift / |fshift|         # complex unit spectrum
#   phase_angles_real  == np.angle(fshift)          # real-valued array of angles
#
# The complex unit spectrum is ready for inverse FFT (after ifftshift). The angle array is
# NOT directly suitable for inverse FFT unless we convert it back to complex exponentials:
#   complex = np.exp(1j * angles)

angles = np.angle(fshift)               # real-valued phase array (radians)
unit_from_angles = np.exp(1j * angles)  # convert back to complex unit-magnitude spectrum

# Confirming equivalence
diff_unit = np.max(np.abs(unit_from_angles - phase_only))
print("Max absolute difference between phase-only (division) and exp(1j*angle):", diff_unit)
````

````python
try:
    bad_recon = np.fft.ifft2(np.fft.ifftshift(angles))
    bad_recon = np.abs(bad_recon)
except Exception as e:
    bad_recon = None
    print("Direct ifft2 on angles raised:", e)

# Proper reconstruction from angles
proper_complex = np.exp(1j * angles)
f_ishift_from_angles = np.fft.ifftshift(proper_complex)
img_from_angles = np.fft.ifft2(f_ishift_from_angles)
img_from_angles = np.abs(img_from_angles)

# 5) Displaying results for comparison
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.imshow(img_phase_only, cmap='gray')
plt.title("Reconstructed (phase_only = fshift/|fshift|)")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(img_from_angles, cmap='gray')
plt.title("Reconstructed (exp(1j * np.angle(fshift)))")
plt.axis('off')

plt.subplot(2,2,3)
if bad_recon is not None:
    plt.imshow(bad_recon, cmap='gray')
    plt.title("Incorrect: ifft2(ifftshift(angles))  # nonsense")
else:
    plt.text(0.1, 0.5, "Direct ifft2 on angles failed (see printed exception)", fontsize=12)
    plt.title("Incorrect reconstruction")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(np.abs(img_from_angles - img_phase_only), cmap='hot')
plt.title("Absolute difference between the two correct methods")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
# 6) Explanation:
# - `phase_only = fshift / np.abs(fshift)`:
#     * Produces a complex-valued array where each frequency coefficient has magnitude 1
#       and the original phase (angle). This can be inverted via ifft2 (after ifftshift).
#
# - `np.angle(fshift)`:
#     * Returns the phase angles in radians as a **real-valued** array.
#     * The angles alone are not valid Fourier coefficients. To use angles for reconstruction,
#       we should convert them back to a complex unit-magnitude spectrum using `np.exp(1j * angles)`.
#
# - Attempting to directly call `ifft2` on `angles` (real radians) is incorrect because iFFT
#   expects complex frequency samples (both magnitude and phase encoded as complex numbers).
#   The result of such a direct attempt is meaningless / noisy.
#
# - The two *correct* approaches are equivalent:
#       fshift / |fshift|    <-> gives complex unit spectrum directly
#       np.exp(1j * np.angle(fshift))  <-> reconstructs same complex unit spectrum
#
# - Reconstruction from phase-only often preserves object structure and edges
#   but loses contrast/texture information that was encoded in the magnitude.
````

````python
# Magnitude-only spectrum
mag_only = np.abs(fshift)            # real, non-negative array

# For reconstruction we need complex-valued frequency samples.
# If we set phase to zero, that corresponds to multiplying mag by exp(1j*0) = 1.
mag_complex = mag_only.astype(np.complex128)  # real part=mag, imag part=0 -> zero phase

# Inverse shift + inverse FFT to reconstruct spatial-domain image
f_ishift_mag = np.fft.ifftshift(mag_complex)
img_mag_only = np.fft.ifft2(f_ishift_mag)
img_mag_only = np.abs(img_mag_only) 
````

````python
#     * The Fourier transform encodes both **magnitude** (how strong each frequency is)
#       and **phase** (relative alignment of sinusoids). Phase contains the positional
#       information and is crucial for maintaining object structure (edges, shapes).
#     * When we set all phases to zero, we keep energy at each frequency but lose
#       the alignment that places those sinusoids to form recognizable objects.
#       The result often looks like a symmetric, "texture-y" or blurred blob — sometimes
#       circular ripples centered in the image — but typically not a faithful image of
#       the original scene.
#
# - Role of logarithm in Magnitude Spectrum visualization:
#     * The raw magnitude values span a very large dynamic range (low-freq components
#       can be orders of magnitude larger than high-frequency components).
#     * Plotting the **linear** magnitude usually results in the center (low frequencies)
#       dominating the visualization; almost everything else looks black.
#     * Applying log (e.g., np.log(mag + 1)) **compresses the dynamic range**, making
#       small magnitude values visible and revealing structure in the spectrum.
#
# - Why magnitude-only image lacks structure while phase-only preserves structure:
#     * Phase dictates how different frequency components add up spatially. Without phase,
#       frequency components are all in-phase (zero phase) and produce symmetric patterns
#       rather than the specific arrangement of edges and shapes found in the original.
````

````python
# Loading second image 
second_path = "Image2.jpeg"
img2 = Image.open(second_path).convert("L")   # convert to grayscale

gray1=gray.copy()

# Resize img2 to match gray1
if img2.size[::-1] != gray1.shape:  # PIL size = (W,H) ; gray shape = (H,W)
    img2 = img2.resize((gray1.shape[1], gray1.shape[0]),)

gray2 = np.array(img2).astype(np.float64)

# Computing FFTs
F1 = np.fft.fft2(gray1)
F2 = np.fft.fft2(gray2)

f1 = np.fft.fftshift(F1)
f2 = np.fft.fftshift(F2)

# Extracting magnitude and phase 
mag1 = np.abs(f1)
mag2 = np.abs(f2)

phase1 = np.angle(f1)   
phase2 = np.angle(f2)

# Swapping
F1_swapped_complex = mag1 * np.exp(1j * phase2)
F2_swapped_complex = mag2 * np.exp(1j * phase1)


f1_ishift = np.fft.ifftshift(F1_swapped_complex)
f2_ishift = np.fft.ifftshift(F2_swapped_complex)

img1_from_swap = np.fft.ifft2(f1_ishift)
img2_from_swap = np.fft.ifft2(f2_ishift)


img1_from_swap = np.abs(img1_from_swap)
img2_from_swap = np.abs(img2_from_swap)

# Normalizing images for display (0..1)
def norm01(x):
    x = x - x.min()
    rng = np.ptp(x)
    if rng == 0:
        return x
    return x / (rng + 1e-12)

# Displaying originals and swapped reconstructions 
plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.imshow(norm01(gray1), cmap='gray')
plt.title("Original Image 1 (gray1)")
plt.axis('off')

plt.subplot(2,3,2)
plt.imshow(np.log(mag1 + 1), cmap='gray')
plt.title("Mag1 (log)")
plt.axis('off')

plt.subplot(2,3,3)
plt.imshow(phase1, cmap='gray')
plt.title("Phase1")
plt.axis('off')

plt.subplot(2,3,4)
plt.imshow(norm01(gray2), cmap='gray')
plt.title("Original Image 2 (gray2)")
plt.axis('off')

plt.subplot(2,3,5)
plt.imshow(np.log(mag2 + 1), cmap='gray')
plt.title("Mag2 (log)")
plt.axis('off')

plt.subplot(2,3,6)
plt.imshow(phase2, cmap='gray')
plt.title("Phase2")
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(norm01(img1_from_swap), cmap='gray')
plt.title("Recons: mag1 + phase2  (|F1| e^{iθ2})")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(norm01(img2_from_swap), cmap='gray')
plt.title("Recons: mag2 + phase1  (|F2| e^{iθ1})")
plt.axis('off')

plt.tight_layout()
plt.show()

# Explanation:
# - Construction:
#     F_swapped = mag_source * exp(i * phase_target)
#   produces a complex Fourier spectrum that uses the amplitude 
#   distribution of one image and the positional/structural phase information of the other.
#
# - Observed effect:
#     * The reconstructed image that uses **phase from image A** typically shows the
#       **structure** (shapes, edges, object positions) of image A, even if its magnitudes
#       come from image B. Textural/contrast details (fine grain, overall energy distribution)
#       come from the magnitude.
#     * In short: **phase describes structure / geometry / object placement**, while
#       **magnitude describes texture / relative energy & contrast across frequencies**.
#
# - Practical note:
#     * Because phase controls how sinusoids align in space, swapping phase transfers
#       the recognizable arrangement of features between images. Magnitude mainly
#       affects the perceived contrast and texture details.

````
