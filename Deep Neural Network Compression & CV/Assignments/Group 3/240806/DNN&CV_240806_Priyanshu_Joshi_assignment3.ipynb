````python
import numpy as np
import matplotlib.pyplot as plt
import cv2
````

````python
def convolve(img, kernel):
    """
    img: 2D numpy array (grayscale image)
    kernel: 2D numpy array (odd dimensions)
    """
    kh, kw = kernel.shape
    pad_h = kh // 2
    pad_w = kw // 2

    # Reflect padding
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')

    output = np.zeros_like(img, dtype=np.float64)

    # Flip kernel for true convolution
    kernel_flipped = np.flipud(np.fliplr(kernel))

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel_flipped)

    return output
````

````python
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float64)
    kernel /= kernel.size
    return convolve(img, kernel)

def average_blur_rgb(img, ksize):
    blurred = np.zeros_like(img, dtype=np.float64)

    for c in range(3):
        blurred[:, :, c] = average_blur_gray(img[:, :, c], ksize)

    return blurred.astype(np.uint8)
````

````python
def gaussian_kernel(ksize, sigma):
    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel /= np.sum(kernel)

    return kernel

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    blurred = np.zeros_like(img, dtype=np.float64)
    kernel = gaussian_kernel(ksize, sigma)

    for c in range(3):
        blurred[:, :, c] = convolve(img[:, :, c], kernel)

    return blurred.astype(np.uint8)
````

````python
def magnitude_spectrum(img):
    fft = np.fft.fft2(img)
    fft_shift = np.fft.fftshift(fft)
    magnitude = np.log(1 + np.abs(fft_shift))
    return magnitude
````

````python
spec_original = magnitude_spectrum(gray)
spec_blurred = magnitude_spectrum(blurred)
spec_sharp = magnitude_spectrum(sharp_unsharp)
````

````python
import numpy as np

def rgb_to_hsv(img_rgb):

    # Convert to float for calculation
    img = img_rgb.astype(np.float32)
    R = img[..., 0]
    G = img[..., 1]
    B = img[..., 2]

    # Max and min per pixel
    maxc = np.max(img, axis=-1)
    minc = np.min(img, axis=-1)
    delta = maxc - minc

    # Hue calculation 
    H = np.zeros_like(maxc)

    # Avoid division by zero
    mask = delta != 0

    # When max = R
    idx = (maxc == R) & mask
    H[idx] = (60 * ((G[idx] - B[idx]) / delta[idx]) + 360) % 360

    # When max = G
    idx = (maxc == G) & mask
    H[idx] = (60 * ((B[idx] - R[idx]) / delta[idx]) + 120) % 360

    # When max = B
    idx = (maxc == B) & mask
    H[idx] = (60 * ((R[idx] - G[idx]) / delta[idx]) + 240) % 360

    # OpenCV scales H from 0–360 → 0–179
    H = (H / 2).astype(np.uint8)

    # ---- Saturation ----
    S = np.zeros_like(maxc)
    nonzero_mask = maxc != 0
    S[nonzero_mask] = (delta[nonzero_mask] / maxc[nonzero_mask]) * 255
    S = S.astype(np.uint8)

    # ---- Value ----
    V = maxc.astype(np.uint8)

    # Combine into HSV image (uint8)
    hsv = np.stack([H, S, V], axis=-1)
    return hsv
````

````python
hsv = rgb_to_hsv(img_part5)

H = hsv[:, :, 0]
S = hsv[:, :, 1]
V = hsv[:, :, 2]

yellow_mask = (
    (H >= 20) & (H <= 35) &
    (S >= 100) &
    (V >= 100)
)

H_new = H.copy()
H_new[yellow_mask] = 0    # red hue

hsv_new = np.stack([H_new, S, V], axis=2)

img_red_leaf = cv2.cvtColor(hsv_new, cv2.COLOR_HSV2RGB)

img_red_leaf = np.clip(img_red_leaf, 0, 255).astype(np.uint8)


````
