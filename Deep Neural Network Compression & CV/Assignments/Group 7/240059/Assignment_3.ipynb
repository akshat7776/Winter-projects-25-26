````python
# Task 1
def convolve(img, kernel):
  out_img = np.zeros(img.shape)

  # Getting shapes for iteration and padding
  img_h, img_w = img.shape[:2]
  k_h, k_w = kernel.shape[:2]

  # Padding
  pad_h = k_h // 2
  pad_w = k_w // 2
  pad_img = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode = 'reflect')

  # Convolution (* just does element wise multiplication)
  for i in range(img_h):
    for j in range(img_w):
      out_img[i,j] = np.sum(pad_img[i:i+k_h, j:j+k_w] * kernel) # np.sum() computes convolution sum at a pixel

  return out_img
````

````python
# Task 2
def average_blur_gray(img, ksize):
  kernel = np.ones((ksize, ksize)) / (ksize * ksize)  # Defining the average blur kernel
  return convolve(img, kernel)  # Using the pre-defined function in Task 1
````

````python
# Custom function for average blur on RGB Image
def average_blur_rgb(img, ksize):
  # Getting channels seperated from RGB Image
  img_r = img[:,:,0]
  img_g = img[:,:,1]
  img_b = img[:,:,2]

  # Applying blur on individual channels
  avg_r = average_blur_gray(img_r, ksize)
  avg_g = average_blur_gray(img_g, ksize)
  avg_b = average_blur_gray(img_b, ksize)

  # Final Blurred RGB Image
  out_img = np.zeros_like(img)
  out_img[:,:,0] = avg_r
  out_img[:,:,1] = avg_g
  out_img[:,:,2] = avg_b

  return out_img
````

````python
# Task 3
def gaussian_kernel(ksize, sigma):
  x = np.linspace(-(ksize//2), ksize//2, ksize)
  y = np.linspace(-(ksize//2), ksize//2, ksize)

  # Defining the kernel
  g = np.zeros((ksize,ksize))
  for i in range(ksize):
    for j in range(ksize):
      g[i,j] = (1/(2*np.pi*(sigma**2)))*np.exp(-((x[i]**2)+(y[j]**2))/(2*(sigma**2))) # exponential term in formula is for decay with distance

  return g/np.sum(g)
````

````python
k = gaussian_kernel(3, 1.0)
print("Gaussian Kernel (k = 3): \n", k)
print("\nSum of Gaussian Kernel: ",k.sum())  # Sum of Gaussian Kernel should be 1.0
````

````python
# Custom function for Gaussian blur on Grayscale Image
def gaussian_blur_gray(img, ksize, sigma):
  kernel = gaussian_kernel(ksize, sigma)
  return convolve(img, kernel)
````

````python
# Custom function to get Magnitude Spectrum of any Grayscale Image
def mag_spectrum(img_gray):
  f = np.fft.fft2(img_gray)
  fshift = np.fft.fftshift(f)
  return np.log(np.abs(fshift))
````

````python
# Original Image's Magnitude Spectrum
orig_spectrum = mag_spectrum(img_gray)
# Gaussian Blur Spectrum with sigma = 1
Gblur_spectrum = mag_spectrum(gaussian_blur_gray(img_gray, 7, 1.0))
# Gaussian Blur Spectrum with sigma = 4 (Higher Sigma => More agressive blurring)
Gblur_spectrum2 = mag_spectrum(gaussian_blur_gray(img_gray, 7, 4.0))
````

````python
# Display
plt.figure(figsize=(14,8))

plt.subplot(1,3,1)
plt.imshow(orig_spectrum, cmap='gray')
plt.title("Original Image Spectrum")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(Gblur_spectrum, cmap='gray')
plt.title("Gaussian Blurred Spectrum (Sigma = 1.0)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(Gblur_spectrum2, cmap='gray')
plt.title("Gaussian Blurred Spectrum (Sigma = 4.0)")
plt.axis("off")

plt.show()
````

````python
# Gaussian blur suppresses high frequencies.
# Outer regions (high frequencies) become darker.
# Low frequencies near the center remain bright.

# Increasing sigma or kernel size:
# - Larger dark region
# - More aggressive low-pass filtering

# So, Gaussian blur is a low-pass filter.
````

````python
def yellow_edge(img):
  r,g,b = img[:,:,0], img[:,:,1], img[:,:,2]

  # Using Sobel Edge Detection on R, G channel
  rx = convolve(r, sobel_x)
  ry = convolve(r, sobel_y)
  gx = convolve(g, sobel_x)
  gy = convolve(g, sobel_y)

  # Doing the same on R-G
  rg = r - g
  rgx = convolve(rg, sobel_x)
  rgy = convolve(rg, sobel_y)

  # Computing Magnitude
  r_mag = np.sqrt(rx**2 + ry**2)
  g_mag = np.sqrt(gx**2 + gy**2)
  rg_mag = np.sqrt(rgx**2 + rgy**2)

  edge = rg_mag * 0.6 + r_mag * 0.3 + g_mag * 0.1 # Weighted Sum

  # Normalization
  edge_norm = (edge / edge.max()) * 255
  edge_norm = edge_norm.astype(np.uint8)

  # Thresholding
  edge_map = np.zeros_like(edge_norm)
  edge_map[edge_norm >= 30] = 255

  return edge_map
````

````python
def rgb_to_hsv(img): # Takes a 0-255 (not normalised) RGB image
  # Normalizing the RGB value and change of dtype to ensure the same
  r,g,b = img[:,:,0].astype("float")/255, img[:,:,1].astype("float")/255, img[:,:,2].astype("float")/255

  # Calculating per pixel cmax and cmin
  cmax = np.maximum(np.maximum(r, g), b)
  cmin = np.minimum(np.minimum(r, g), b)
  delta = cmax - cmin

  h = np.zeros_like(cmax)
  s = np.zeros_like(cmax)
  v = cmax  # Value

  # Saturation
  for i in range(cmax.shape[0]):
    for j in range(cmax.shape[1]):
      if cmax[i][j] != 0:
        s[i][j] = delta[i][j] / cmax[i][j]
  # Hue
  for i in range(cmax.shape[0]):
    for j in range(cmax.shape[1]):
      if delta[i][j] != 0:
        if cmax[i][j] == r[i][j]:
          h[i][j] = ((g[i][j] - b[i][j]) / delta[i][j])
        elif cmax[i][j] == g[i][j]:
          h[i][j] = ((b[i][j] - r[i][j]) / delta[i][j]) + 2
        elif cmax[i][j] == b[i][j]:
          h[i][j] = ((r[i][j] - g[i][j]) / delta[i][j]) + 4
        h[i][j] = h[i][j] % 6.0
        h[i][j] = h[i][j] * 60  # Theoretical 0-360 range

  # Scale S,V
  s = (s * 255).astype(np.uint8)
  v = (v * 255).astype(np.uint8)
  # Returns an HSV image (also not normalised, compatible with OpenCV cvtColor)
  hsv = np.zeros_like(img, dtype=np.float32)
  hsv[:,:,0] = h/2  # Since HSV range for OpenCV is 0-179 and not the theoretical 0-360
  hsv[:,:,1] = s
  hsv[:,:,2] = v
  hsv = hsv.astype(np.uint8)
  return hsv
````
