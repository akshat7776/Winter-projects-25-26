````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread("stranger thing.jpeg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.axis("off")
plt.show()
````

````python
def convolve(img, kernel):
    kernel = np.flipud(np.fliplr(kernel))

    kH, kW = kernel.shape
    pad_h = kH // 2
    pad_w = kW // 2

    padded_img = np.pad(
        img,
        pad_width=((pad_h, pad_h), (pad_w, pad_w)),
        mode="reflect"
    )

    H, W = img.shape
    output = np.zeros((H, W), dtype=np.float32)

    for i in range(H):
        for j in range(W):
            region = padded_img[i:i+kH, j:j+kW]
            output[i, j] = np.sum(region * kernel)

    return output

````

````python
#Task 2
def average_blur_gray(img, ksize):
  kernel = np.ones((ksize, ksize)) / (ksize * ksize)
  return convolve(img, kernel)
img_gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
plt.figure(figsize=(14,8))

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(average_blur_gray(img_gray,8), cmap='gray')
plt.title("Avg. Blurred Grayscale Image (k = 8)")
plt.axis("off")

plt.show()
````

````python
def average_blur_rgb(img, ksize):
    out_img = np.zeros_like(img)

    for c in range(3):  # R, G, B
        out_img[:, :, c] = average_blur_gray(img[:, :, c], ksize)

    return out_img
blur_rgb = average_blur_rgb(image, 8)
````

````python
#task3
def gaussian_kernel(ksize, sigma):
    k = ksize // 2
    x, y = np.meshgrid(
        np.arange(-k, k+1),
        np.arange(-k, k+1)
    )

    kernel = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)

    return kernel
````

````python
k = gaussian_kernel(9, 2.0)
print("Gaussian Kernel (k = 9): \n", k)
print("\nSum of Gaussian Kernel: ",k.sum())
````

````python
def gaussian_blur_gray(img, ksize, sigma):
  kernel = gaussian_kernel(ksize, sigma)
  return convolve(img, kernel)
````

````python
def magnitude_spectrum(img):

    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)
    magnitude = np.log(1 + np.abs(fshift))
    return magnitude
````

````python
mag_original = magnitude_spectrum(img_gray)
mag_gaussian = magnitude_spectrum(gaussian_blur_gray(img_gray, 7, 1.0))        # Gaussian blurred image
mag_unsharp = magnitude_spectrum(gaussian_blur_gray(img_gray, 7, 5.0))   # Unsharp masking output
````

````python
mag_original_n = normalize_255(mag_original)
mag_gaussian_n = normalize_255(mag_gaussian)
mag_unsharp_n = normalize_255(mag_unsharp)
````

````python
# OBSERVATIONS:
# 1. Bright center in the magnitude spectrum represents low-frequency components.
# 2. Gaussian blur (LPF) concentrates energy near the center,
#    darkening outer regions (high frequencies are suppressed).
# 3. Increasing Gaussian kernel size or sigma makes the center brighter
#    and the periphery darker, confirming stronger low-pass filtering.
# 4. Unsharp masking increases brightness in outer regions of the spectrum,
#    indicating amplification of high-frequency components (edges).
# 5. Higher alpha in unsharp masking further brightens high-frequency regions,
#    confirming its high-pass behavior.
# 6. This verifies that spatial convolution filters frequencies
#    depending on the kernel used.
````

````python
def color_edge_detection(img, sobel_x, sobel_y):
    # Split channels
    r, g, b = img[:, :, 0], img[:, :, 1], img[:, :, 2]

    # Sobel on R channel
    rx = convolve(r, sobel_x)
    ry = convolve(r, sobel_y)
    r_mag = np.sqrt(rx**2 + ry**2)

    # Sobel on G channel
    gx = convolve(g, sobel_x)
    gy = convolve(g, sobel_y)
    g_mag = np.sqrt(gx**2 + gy**2)

    # Sobel on R-G difference
    rg = r - g
    rgx = convolve(rg, sobel_x)
    rgy = convolve(rg, sobel_y)
    rg_mag = np.sqrt(rgx**2 + rgy**2)

    # Weighted combination
    edge = 0.6 * rg_mag + 0.3 * r_mag + 0.1 * g_mag

    # Normalize to 0–255
    edge_norm = (edge / edge.max()) * 255
    edge_norm = edge_norm.astype(np.uint8)

    # Thresholding
    edge_map = np.zeros_like(edge_norm)
    edge_map[edge_norm >= 30] = 255

    return edge_map
````

````python
def rgb_to_hsv(img):
    # Normalize RGB to [0,1]
    r = img[:, :, 0].astype(np.float32) / 255.0
    g = img[:, :, 1].astype(np.float32) / 255.0
    b = img[:, :, 2].astype(np.float32) / 255.0

    # Compute cmax, cmin, delta
    cmax = np.maximum(np.maximum(r, g), b)
    cmin = np.minimum(np.minimum(r, g), b)
    delta = cmax - cmin

    h = np.zeros_like(cmax, dtype=np.float32)
    s = np.zeros_like(cmax, dtype=np.float32)
    v = cmax.copy()

    H, W = cmax.shape

    # Saturation
    for i in range(H):
        for j in range(W):
            if cmax[i, j] != 0:
                s[i, j] = delta[i, j] / cmax[i, j]

    # Hue
    for i in range(H):
        for j in range(W):
            if delta[i, j] != 0:
                if cmax[i, j] == r[i, j]:
                    h[i, j] = (g[i, j] - b[i, j]) / delta[i, j]
                elif cmax[i, j] == g[i, j]:
                    h[i, j] = (b[i, j] - r[i, j]) / delta[i, j] + 2
                else:  # cmax == b
                    h[i, j] = (r[i, j] - g[i, j]) / delta[i, j] + 4

                h[i, j] = (h[i, j] % 6) * 60  # 0–360 range

    # Scale S and V to OpenCV range
    s = (s * 255).astype(np.uint8)
    v = (v * 255).astype(np.uint8)

    # Construct HSV image (OpenCV compatible)
    hsv = np.zeros_like(img, dtype=np.uint8)
    hsv[:, :, 0] = (h / 2).astype(np.uint8)  # 0–179
    hsv[:, :, 1] = s
    hsv[:, :, 2] = v

    return hsv
````
