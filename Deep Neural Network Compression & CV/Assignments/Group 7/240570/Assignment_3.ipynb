````python
# Task-1
def convolve(img, kernel):
    img = img.astype(np.float32)
    kernel = kernel.astype(np.float32)

    # heigth and width of image and kernel
    h, w = img.shape
    kh, kw = kernel.shape
    pad_h, pad_w = kh // 2, kw // 2

    # Implement padding and it should be done by to make img of dimension
    # (h+kh)x(w+kw), in order to define convolution at edges
    img_padded = np.pad(
        img,
        ((pad_h, pad_h), (pad_w, pad_w)),
        mode="reflect"
    )

    img_convolved = np.zeros((h, w), dtype=np.float32)

    for x in range(h):
        for y in range(w):
            region = img_padded[x:x+kh, y:y+kw]
            img_convolved[x, y] = np.sum(region * kernel)

    return img_convolved

````

````python
# Task-2
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize**2) # Normalized
    return np.clip(convolve(img, kernel),0,255)


def average_blur_rgb(img, ksize):

    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize**2)

    # Apply convolution to each channel independently
    r = convolve(img[:,:,0], kernel)
    g = convolve(img[:,:,1], kernel)
    b = convolve(img[:,:,2], kernel)

    blur= np.stack((r, g, b), axis=-1)
    return np.clip(blur,0,255).astype(np.uint8)
````

````python
# Task -3 Part-1
def gaussian_kernel(ksize, sigma):
    kernel = np.zeros((ksize, ksize), dtype=np.float32)
    center = ksize // 2

    for x in range(ksize):
        for y in range(ksize):
          dx=x-center
          dy=y-center
          kernel[x, y] = np.exp(-(dx**2 + dy**2) / (2 * sigma**2))
    kernel/=(2*np.pi*sigma**2)
    kernel /= kernel.sum() # Normalization
    return kernel

````

````python

# Task-3 Part-2

def gaussian_blur_gray(img, ksize, sigma):

  kernel = gaussian_kernel(ksize, sigma)
  return convolve(img, kernel)


def gaussian_blur_rgb(img, ksize, sigma):

  kernel = gaussian_kernel(ksize, sigma)

  r = convolve(img[:, :, 0], kernel)
  g = convolve(img[:, :, 1], kernel)
  b = convolve(img[:, :, 2], kernel)


  return np.stack((r, g, b), axis=-1)


# Displaying images
plt.figure(figsize=(36, 6))


plt.subplot(1, 6, 1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Gray")
plt.axis('off')


plt.subplot(1, 6, 2)
plt.imshow(average_blur_gray(img_gray, 3).astype(np.uint8), cmap='gray')
plt.title("Average Blur (Gray)")
plt.axis('off')


plt.subplot(1, 6, 3)
plt.imshow(gaussian_blur_gray(img_gray, 3, 1).astype(np.uint8), cmap='gray')
plt.title("Gaussian Blur (Gray)")
plt.axis('off')


plt.subplot(1, 6, 4)
plt.imshow(img_rgb)
plt.title("Original RGB")
plt.axis('off')


plt.subplot(1, 6, 5)
plt.imshow(average_blur_rgb(img_rgb, 3).astype(np.uint8))
plt.title("Average Blur (RGB)")
plt.axis('off')


plt.subplot(1, 6, 6)
plt.imshow(gaussian_blur_rgb(img_rgb, 3, 1).astype(np.uint8))
plt.title("Gaussian Blur (RGB)")
plt.axis('off')


plt.show()
````

Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python
def Gx(img_gray):
  kernel = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
  return convolve(img_gray, kernel)/convolve(img_gray, kernel).max()*255 # Normalized return value
def Gy(img_gray):
  kernel = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
  return convolve(img_gray, kernel)/convolve(img_gray, kernel).max()*255 # Normalized return value
def Gradient_magnitude(img_gray):
  return (np.sqrt(Gx(img_gray)**2 + Gy(img_gray)**2))/(np.sqrt(Gx(img_gray)**2 + Gy(img_gray)**2)).max()*255 # Normalized return value
````

````python
# Displaying filtered images
Sobel_X=Gx(img_gray)
Sobel_Y=Gy(img_gray)
Sobel_mag=Gradient_magnitude(img_gray).astype(np.uint8)
plt.figure(figsize=(24,6))
plt.subplot(1,4,1)
plt.imshow(img_gray,cmap="gray")
plt.title('Original')
plt.axis('off')
plt.subplot(1,4,2)
plt.imshow(Sobel_X,cmap="gray")
plt.title('Sobel X Filter')
plt.axis('off')
plt.subplot(1,4,3)
plt.imshow(Sobel_Y,cmap="gray")
plt.title('Sobel Y Filter')
plt.axis('off')
plt.subplot(1,4,4)
plt.imshow(Sobel_mag,cmap="gray")
plt.title('Sobel Magnitude')
plt.axis('off')
plt.show()
````

1. ORIGINAL IMAGE SPECTRUM:
   - Bright center: High intensity at low frequencies (smooth regions, uniform colors)
   - Dark outer regions: Low intensity at high frequencies (edges, textures, noise)
   - Pattern reflects image content: Regular patterns create bright spots at specific frequencies

2. GAUSSIAN BLUR EFFECTS (LOW-PASS FILTERING):
   - Darker high frequencies: Gaussian blur attenuates high frequencies
   - Larger kernel/sigma → More attenuation: Spectrum gets darker at edges
   - Center remains bright: Low frequencies (smooth areas) preserved
   - Gaussian blurring acts as LOW-PASS FILTER: Allows low frequencies, blocks high frequencies

3. UNSHARP MASKING EFFECTS:
   
   POSITIVE ALPHA (Sharpening):
   - alpha = 0.5: Mild high-frequency boost → slightly brighter outer regions
   - alpha = 1.5: Strong high-frequency boost → much brighter high frequencies
   - alpha = 3.0: Over-sharpening → very bright high frequencies, possible ringing
   
   NEGATIVE ALPHA (More blur):
   - alpha = -1.5: Even darker high frequencies than original blur
   - Acts as stronger low-pass filter
   
   Unsharp masking formula: I_sharp = I + alpha*(I - I_blur)
   - (I - I_blur) = high-frequency components
   - Positive α amplifies high frequencies → BRIGHTER high-frequency regions
   - Negative α reduces high frequencies → DARKER high-frequency regions

4. CONVOLUTION AS FREQUENCY FILTERING:
   - Convolution in spatial domain = Multiplication in frequency domain
   - Kernel's Fourier Transform determines which frequencies pass/attenuate
   

5. DARK vs BRIGHT AREAS INTERPRETATION:
   - BRIGHT areas: Strong frequency components present
   - DARK areas: Weak or absent frequency components
   - CENTER (low frequencies): Always bright
   - EDGES (high frequencies): Bright for sharp images, dark for blurred
   
6. PRACTICAL VERIFICATION OF CONVOLUTION THEOREM:
   - Multiplication in frequency domain = Convolution in spatial domain

````python
def rgb_to_hsv(img_rgb):

  # Segregate into Red,Green and Blue Components
  r=img_rgb[:,:,0].astype(float)
  g=img_rgb[:,:,1].astype(float)
  b=img_rgb[:,:,2].astype(float)

  # Initializing h,s,v arrays
  h=np.zeros(img_rgb.shape[0:2])
  s=np.zeros(img_rgb.shape[0:2])
  v=np.zeros(img_rgb.shape[0:2])

  # Compute h,s,v values over each iteration
  for i in range (img_rgb.shape[0]):
    for j in range (img_rgb.shape[1]):
      cmax=max(r[i,j],g[i,j],b[i,j])
      cmin=min(r[i,j],g[i,j],b[i,j])
      delta=cmax-cmin

      #Hue
      if delta==0:
        h[i,j]=0
      elif cmax==r[i,j]:
        h[i,j]=60*((g[i,j]-b[i,j])/delta)
      elif cmax==g[i,j]:
        h[i,j]=60*((b[i,j]-r[i,j])/delta + 2)
      else:
        h[i,j]=60*((r[i,j]-g[i,j])/delta +4)

      if(h[i,j]<0): # Add 360 to hue if it is negative
        h[i,j]+=360

      h[i,j]/=2 # Compatible with Open-CV versions ( as 0<=hue<180)

      # Saturation
      if cmax==0:
        s[i,j]=0
      else:
        s[i,j]=(delta/cmax)*255  # Not normalized

      # Value
      v[i,j]=cmax

  h=h.astype(np.uint8)
  s=s.astype(np.uint8)
  v=v.astype(np.uint8)
  img_hsv=np.stack((h,s,v),axis=-1)

  return img_hsv


````
