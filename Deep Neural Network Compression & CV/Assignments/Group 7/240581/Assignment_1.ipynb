### B1. Compute the FFT of the grayscale image ( use NumPy )

````python
# Your PART B codes follow here. You can create more cells under this. Write clean and commented code.
````

````python
gray_float=gray.astype(np.float32)
#fft
f= np.fft.fft2(gray_float)
f_shift=np.fft.fftshift(f)
````

````python
magnitude=20*np.log(np.abs(f_shift)+1)
phase =np.angle(f_shift)

plt.figure(figsize=(20,6))
plt.subplot(1,3,1)
plt.imshow(wt_frm, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(magnitude, cmap='gray')
plt.title('Magnitude spec')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(phase, cmap='twilight')
plt.title('Phase')
plt.axis('off')
plt.show()
````

````python
#Low pass mask
rows, cols = gray.shape
crow,ccol = rows//2 , cols//2

radius =10
mask_lpf = np.zeros((rows , cols ), np.uint8)
cv2.circle(mask_lpf, (ccol,crow),radius , 1 ,-1)
f_lpf=f_shift*mask_lpf
ishift_lpf=np.fft.ifftshift(f_lpf)
img_back_lpf=np.fft.ifft2(ishift_lpf)
img_back_lpf=np.abs(img_back_lpf)
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.axis("off")
plt.title("original")

plt.subplot(1,2,2)
plt.imshow(img_back_lpf, cmap='gray')
plt.axis("off")
plt.title("radius=10")
````

````python

radius =30
mask_lpf = np.zeros((rows , cols ), np.uint8)
cv2.circle(mask_lpf, (ccol,crow),radius , 1 ,-1)
f_lpf=f_shift*mask_lpf
ishift_lpf=np.fft.ifftshift(f_lpf)
img_back_lpf=np.fft.ifft2(ishift_lpf)
img_back_lpf=np.abs(img_back_lpf)
plt.figure(figsize=(15,6))

plt.subplot(1,2,1)
plt.imshow(img_back_lpf, cmap='gray')
plt.axis("off")
plt.title("radius=30")



radius =80
mask_lpf = np.zeros((rows , cols ), np.uint8)
cv2.circle(mask_lpf, (ccol,crow),radius , 1 ,-1)
f_lpf=f_shift*mask_lpf
ishift_lpf=np.fft.ifftshift(f_lpf)
img_back_lpf=np.fft.ifft2(ishift_lpf)
img_back_lpf=np.abs(img_back_lpf)

plt.subplot(1,2,2)
plt.imshow(img_back_lpf, cmap='gray')
plt.axis("off")
plt.title("radius=80")
#Blur increases as the radius decreases because low frequencies are located at the center of the FFT.
#smaller radius contain mostly low frequencies and low frequencies make  more blurred image
````

````python
#High frequency mask
rows, cols = gray.shape
crow,ccol = rows//2 , cols//2

radius =100
mask_lpf = np.zeros((rows , cols ), np.uint8)
cv2.circle(mask_lpf, (ccol,crow),radius , 1 ,-1)
mask_hpf = 1 - mask_lpf
f_hpf=f_shift*mask_hpf
ishift_hpf=np.fft.ifftshift(f_hpf)
img_back_hpf=np.fft.ifft2(ishift_hpf)
img_back_hpf=np.abs(img_back_hpf)
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
plt.imshow(img_back_hpf, cmap='gray')
plt.axis("off")
plt.title("hpf r=100")

radius =30
mask_lpf = np.zeros((rows , cols ), np.uint8)
cv2.circle(mask_lpf, (ccol,crow),radius , 1 ,-1)
mask_hpf = 1 - mask_lpf
f_hpf=f_shift*mask_hpf
ishift_hpf=np.fft.ifftshift(f_hpf)
img_back_hpf=np.fft.ifft2(ishift_hpf)
img_back_hpf=np.abs(img_back_hpf)

plt.subplot(1,2,2)
plt.imshow(img_back_hpf, cmap='gray')
plt.axis("off")
plt.title("hpf r=30")
````

````python
#LPF square
rows, cols = gray.shape
crow,ccol = rows//2 , cols//2
side=50
mask_lpf_sq=np.zeros((rows,cols),np.uint8)
cv2.rectangle(mask_lpf_sq, (ccol,crow),(side ,side), 1 ,-1)
f_lpf_sq=f_shift*mask_lpf_sq
ishift_sq= np.fft.ifftshift(f_lpf_sq)
img_back = np.fft.ifft2(ishift_sq)
img_sq = np.abs(img_back)
plt.figure(figsize=(15,6))
plt.imshow(img_sq , cmap='gray')
plt.axis("off")
plt.title("LPF Square")
plt.show()
````

````python
gray_float=gray.astype(np.float32)
#fft
f= np.fft.fft2(gray_float)
f_shift=np.fft.fftshift(f)
phase_only = f_shift/np.abs(f_shift)

ishift_re=np.fft.ifftshift(phase_only)
img_back_re=np.fft.ifft2(ishift_re)
img_back_re=np.abs(img_back_re)

plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.imshow(img_back_re, cmap='twilight')
plt.title('Phase only')
plt.axis("off")
plt.show()

phase =np.angle(f_shift)
plt.subplot(1,2,2)
plt.imshow(phase, cmap='twilight')
plt.title('Phase')
plt.axis("off")
plt.show()
````

````python
#Magnitude only reconstruction
magnitude_only=np.abs(f_shift)
ishift_mag_re=np.fft.ifftshift(magnitude_only)
img_back_mag_re=np.fft.ifft2(ishift_re)
img_back_mag_re=np.abs(img_back_re)

plt.figure(figsize=(15,6))
plt.imshow(img_back_mag_re,cmap='gray')
plt.title("magnitude only reconstruction")
plt.axis("off")
plt.show()
````
