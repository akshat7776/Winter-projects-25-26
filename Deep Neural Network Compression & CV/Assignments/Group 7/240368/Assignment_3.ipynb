````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def average_blur_gray(img, ksize):

  kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)
  return convolve(img, kernel)


def average_blur_rgb(img, ksize):

    r = img[:,:,0]
    g = img[:,:,1]
    b = img[:,:,2]

    r_blur = average_blur_gray(r, ksize)
    g_blur = average_blur_gray(g, ksize)
    b_blur = average_blur_gray(b, ksize)

    rgb_blur = np.dstack((r_blur, g_blur, b_blur))

    return np.clip(rgb_blur, 0, 255).astype(np.uint8)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

name = 'rcrc.jpg'
img = Image.open(name)


def convolve(img, kernel):
    img_float = img.astype(np.float32)
    kernel_float = kernel.astype(np.float32)

    img_h, img_w = img_float.shape
    k_h, k_w = kernel_float.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    padded_img = np.pad(img_float, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img_float)

    for i in range(img_h):
        for j in range(img_w):

            region = padded_img[i : i + k_h, j : j + k_w]
            output[i, j] = np.sum(region * kernel_float)

    return output

def gaussian_kernel(ksize, sigma):

    center = ksize // 2
    x = np.linspace(-center, center, ksize)
    y = np.linspace(-center, center, ksize)
    xx, yy = np.meshgrid(x, y)


    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))

    kernel /= kernel.sum()

    return kernel

def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    return convolve(img, kernel)

def gaussian_blur_rgb(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)

    r = convolve(img[:, :, 0], kernel)
    g = convolve(img[:, :, 1], kernel)
    b = convolve(img[:, :, 2], kernel)

    return np.stack((r, g, b), axis=-1)

img_part5 = np.array(img)
img_gray = np.mean(img_part5, axis=2)

KSIZE = 15
SIGMA = 5

blur_gray = gaussian_blur_gray(img_gray, KSIZE, SIGMA)
blur_rgb = gaussian_blur_rgb(img_part5, KSIZE, SIGMA)

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(np.clip(blur_gray, 0, 255).astype(np.uint8), cmap='gray')
plt.title(f"Gaussian Blur Gray (k={KSIZE}, sigma={SIGMA})")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_part5)
plt.title("Original RGB")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(np.clip(blur_rgb, 0, 255).astype(np.uint8))
plt.title(f"Gaussian Blur RGB (k={KSIZE}, sigma={SIGMA})")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

def convolve(img, kernel):
    # Here we are converting to float 32 to prevent overflowing and negative values.
    img_float = img.astype(np.float32)
    kernel_float = kernel.astype(np.float32)

    img_h, img_w = img_float.shape
    k_h, k_w = kernel_float.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    # Pad image (Reflect mode handles borders smoothly)
    padded_img = np.pad(img_float, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img_float)

    for i in range(img_h):
        for j in range(img_w):
            region = padded_img[i : i + k_h, j : j + k_w]
            output[i, j] = np.sum(region * kernel_float)

    return output

def sobel_edge_detection(img):


    Gx_kernel = np.array([[-1, 0, 1],
                          [-2, 0, 2],
                          [-1, 0, 1]], dtype=np.float32)

    Gy_kernel = np.array([[-1, -2, -1],
                          [0,  0,  0],
                          [1,  2,  1]], dtype=np.float32)

    # Compute Gradients
    gx = convolve(img, Gx_kernel)
    gy = convolve(img, Gy_kernel)

    # Compute Gradient Magnitude
    # Formula: sqrt(gx^2 + gy^2)
    magnitude = np.sqrt(gx**2 + gy**2)

    norm_magnitude = (magnitude / np.max(magnitude)) * 255

    # Optional: Thresholding
    threshold_value = 50
    _,binary_edge_map = cv2.threshold(norm_magnitude.astype(np.uint8), threshold_value, 255, cv2.THRESH_BINARY)
    return gx, gy, binary_edge_map

    return gx, gy, norm_magnitude.astype(np.uint8)
````

````python
import numpy as np
import matplotlib.pyplot as plt

def get_magnitude_spectrum(img):

    img_float = img.astype(np.float32)
    f = np.fft.fft2(img_float)
    fshift = np.fft.fftshift(f)
    magnitude = 20 * np.log(np.abs(fshift) + 1)

    return magnitude

def unsharp_masking(img, ksize, sigma, alpha):

    blurred = gaussian_blur_gray(img, ksize, sigma)
    mask = img.astype(np.float32) - blurred
    sharpened = img.astype(np.float32) + (alpha * mask)
    return np.clip(sharpened, 0, 255).astype(np.uint8), blurred

KSIZE = 15
SIGMA = 5
ALPHA = 2.0

sharp_img, blur_img = unsharp_masking(img_gray, KSIZE, SIGMA, ALPHA)

original_spectrum = get_magnitude_spectrum(img_gray)
blur_spectrum = get_magnitude_spectrum(blur_img)
sharp_spectrum = get_magnitude_spectrum(sharp_img)

plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1), plt.imshow(img_gray, cmap='gray'), plt.title("Original Image")
plt.axis('off')
plt.subplot(2, 3, 2), plt.imshow(blur_img, cmap='gray'), plt.title(f"Gaussian Blur\n(LPF, k={KSIZE})")
plt.axis('off')
plt.subplot(2, 3, 3), plt.imshow(sharp_img, cmap='gray'), plt.title(f"Unsharp Masking\n(HPF Boost, alpha={ALPHA})")
plt.axis('off')

plt.subplot(2, 3, 4), plt.imshow(original_spectrum, cmap='gray'), plt.title("Original Spectrum")
plt.axis('off')

plt.subplot(2, 3, 5), plt.imshow(blur_spectrum, cmap='gray'), plt.title("Blurred Spectrum (LPF)")
plt.axis('off')

plt.subplot(2, 3, 6), plt.imshow(sharp_spectrum, cmap='gray'), plt.title("Sharpened Spectrum (HPF Boost)")
plt.axis('off')

plt.tight_layout()
plt.show()
````
