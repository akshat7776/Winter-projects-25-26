````python
# Task 1 Convolution

import numpy as np

def convolve(img, kernel):

    # Image and kernel shapes
    H, W = img.shape
    kH, kW = kernel.shape

    # Kernel center offsets
    pad_h = kH // 2
    pad_w = kW // 2

    # Reflect padding
    padded_img = np.pad(
        img,
        pad_width=((pad_h, pad_h), (pad_w, pad_w)),
        mode='reflect'
    )

    # Output image
    output = np.zeros_like(img, dtype=np.float32)

    # Convolution operation
    for i in range(H):
        for j in range(W):
            # Extract local region
            window = padded_img[i:i+kH, j:j+kW]

            # Element-wise multiplication (* is NOT matrix multiplication)
            output[i, j] = np.sum(window * kernel)

    return output
````

````python
# Task 2 — Average Blur (using custom convolve)

def average_blur_gray(img, ksize):

    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)
    blurred = convolve(img, kernel)
    return np.clip(blurred, 0, 255).astype(np.uint8)


def average_blur_rgb(img, ksize):

    kernel = np.ones((ksize, ksize), dtype=np.float32) / (ksize * ksize)

    r = convolve(img[:, :, 0], kernel)
    g = convolve(img[:, :, 1], kernel)
    b = convolve(img[:, :, 2], kernel)

    blur = np.stack((r, g, b), axis=-1)
    return np.clip(blur, 0, 255).astype(np.uint8)
````

````python
# Task 3 Gaussian Blur
import numpy as np

def gaussian_kernel(ksize, sigma):

    ax = np.arange(-(ksize // 2), ksize // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)   # normalize

    return kernel.astype(np.float32)
````

````python
def gaussian_blur_gray(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    blurred = convolve(img, kernel)
    return np.clip(blurred, 0, 255).astype(np.uint8)

````

````python
def gaussian_blur_rgb(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)

    r = convolve(img[:, :, 0], kernel)
    g = convolve(img[:, :, 1], kernel)
    b = convolve(img[:, :, 2], kernel)

    blur = np.stack((r, g, b), axis=-1)
    return np.clip(blur, 0, 255).astype(np.uint8)
````

Compute the magnitude spectrum of the original image, gaussian blurred image and the unsharp masking image. Now, play with the alpha value in sharpening and gaussian kernel and notice how the magntude spectrum changes.
Then, in code comments, explain your observations of dark and bright areas in the magnitude spectrum. In nutshell, verify how convolution is actually filtering frequencies depending on the kernel.

````python
import numpy as np
import matplotlib.pyplot as plt

# Magnitude Spectrum Function

def magnitude_spectrum(img):
    img = img.astype(np.float32)
    F = np.fft.fft2(img)
    F = np.fft.fftshift(F)
    return 20 * np.log(np.abs(F) + 1)

# Unsharp Masking (Grayscale)

def unsharp_masking(img, alpha, ksize=7, sigma=2.0):
    blur = gaussian_blur_gray(img, ksize, sigma)
    mask = img.astype(np.float32) - blur.astype(np.float32)
    sharp = img.astype(np.float32) + alpha * mask
    return np.clip(sharp, 0, 255).astype(np.uint8)

# Compute Magnitude Spectra

orig_mag = magnitude_spectrum(gray_img)

gauss_mag = magnitude_spectrum(
    gaussian_blur_gray(gray_img, ksize=3, sigma=1.0)
)

unsharp_neg_mag = magnitude_spectrum(
    unsharp_masking(gray_img, alpha=-1.5)
)

unsharp_pos_mag = magnitude_spectrum(
    unsharp_masking(gray_img, alpha=2.0)
)

unsharp_high_mag = magnitude_spectrum(
    unsharp_masking(gray_img, alpha=10.0)
)


# Display Results

plt.figure(figsize=(30,6))

plt.subplot(1,5,1)
plt.imshow(orig_mag, cmap='gray')
plt.title("Original Magnitude Spectrum")
plt.axis("off")

plt.subplot(1,5,2)
plt.imshow(gauss_mag, cmap='gray')
plt.title("Gaussian Blur Spectrum")
plt.axis("off")

plt.subplot(1,5,3)
plt.imshow(unsharp_neg_mag, cmap='gray')
plt.title("Unsharp Masking (α < 0)")
plt.axis("off")

plt.subplot(1,5,4)
plt.imshow(unsharp_pos_mag, cmap='gray')
plt.title("Unsharp Masking (α = 2)")
plt.axis("off")

plt.subplot(1,5,5)
plt.imshow(unsharp_high_mag, cmap='gray')
plt.title("Unsharp Masking (α = 10)")
plt.axis("off")

plt.show()

# OBSERVATIONS:
# - The original image shows strong low-frequency components at the center.
# - Gaussian blur suppresses high frequencies, making outer regions darker.
# - Negative alpha in unsharp masking further attenuates high frequencies.
# - Positive alpha boosts high frequencies, enhancing edges.
# - Very large alpha values amplify noise due to excessive high-frequency gain.
#
# This verifies that spatial-domain convolution corresponds directly
# to frequency-domain filtering behavior.

````

````python
def yellow_edge(img):

    # Split channels
    r = img[:, :, 0].astype(np.float32)
    g = img[:, :, 1].astype(np.float32)
    b = img[:, :, 2].astype(np.float32)

    # Sobel on Red channel
    rx = convolve(r, sobel_x)
    ry = convolve(r, sobel_y)

    # Sobel on Green channel
    gx = convolve(g, sobel_x)
    gy = convolve(g, sobel_y)

    # Sobel on R-G difference
    rg = r - g
    rgx = convolve(rg, sobel_x)
    rgy = convolve(rg, sobel_y)

    # Gradient magnitudes
    r_mag  = np.sqrt(rx**2 + ry**2)
    g_mag  = np.sqrt(gx**2 + gy**2)
    rg_mag = np.sqrt(rgx**2 + rgy**2)

    # Weighted combination
    edge = 0.6 * rg_mag + 0.3 * r_mag + 0.1 * g_mag

    # Normalize to 0–255
    edge_norm = (edge / edge.max()) * 255
    edge_norm = edge_norm.astype(np.uint8)

    # Thresholding for clean outline
    edge_map = np.zeros_like(edge_norm)
    edge_map[edge_norm >= 30] = 255

    return edge_map
````

````python
def rgb_to_hsv(img_rgb):

    r = img_rgb[:,:,0].astype(float)
    g = img_rgb[:,:,1].astype(float)
    b = img_rgb[:,:,2].astype(float)

    h = np.zeros(img_rgb.shape[0:2])
    s = np.zeros(img_rgb.shape[0:2])
    v = np.zeros(img_rgb.shape[0:2])

    for i in range(img_rgb.shape[0]):
        for j in range(img_rgb.shape[1]):

            cmax = max(r[i,j], g[i,j], b[i,j])
            cmin = min(r[i,j], g[i,j], b[i,j])
            delta = cmax - cmin

            # Hue
            if delta == 0:
                h[i,j] = 0
            elif cmax == r[i,j]:
                h[i,j] = 60 * ((g[i,j] - b[i,j]) / delta)
            elif cmax == g[i,j]:
                h[i,j] = 60 * ((b[i,j] - r[i,j]) / delta + 2)
            else:
                h[i,j] = 60 * ((r[i,j] - g[i,j]) / delta + 4)

            if h[i,j] < 0:
                h[i,j] += 360

            h[i,j] /= 2   # (0–179)

            # Saturation
            if cmax == 0:
                s[i,j] = 0
            else:
                s[i,j] = (delta / cmax) * 255

            # Value
            v[i,j] = cmax

    img_hsv = np.stack(
        (h.astype(np.uint8),
         s.astype(np.uint8),
         v.astype(np.uint8)),
        axis=-1
    )

    return img_hsv
````

````python
def yellow_to_red(img_rgb):

    # Convert to HSV
    img_hsv = rgb_to_hsv(img_rgb)
    img_hsv_mod = img_hsv.copy()

    h = img_hsv_mod[:,:,0]

    # Yellow hue range (OpenCV scale)
    yellow_mask = (h >= 15) & (h <= 30)

    # Visualize mask
    mask_vis = np.zeros_like(img_rgb)
    mask_vis[yellow_mask] = [255, 255, 255]

    plt.figure(figsize=(6,6))
    plt.imshow(mask_vis)
    plt.title("Yellow Region Mask")
    plt.axis("off")
    plt.show()

    # Convert yellow → red
    img_hsv_mod[:,:,0][yellow_mask] = 0   # Red hue

    # Convert back to RGB
    img_rgb_out = cv2.cvtColor(img_hsv_mod, cv2.COLOR_HSV2RGB)

    return img_rgb_out
````
