### C3. Create a square LPF mask
Keep side of square as 50. The square will be at the center of the filter.
Apply, Recontruct and show the output.

````python
# C1 LPF masks

import numpy as np
import cv2
import matplotlib.pyplot as plt

plt.figure(figsize=(18, 6))

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2   # center of the frequency image

radii = [10, 30, 80]

for i, radius in enumerate(radii, start=1):

    # Create low-pass mask

    mask_lpf = np.zeros((rows, cols), np.uint8)
    cv2.circle(mask_lpf, (ccol, crow), radius, 1, -1)

    # Apply LPF in frequency domain
    f_lpf = fshift * mask_lpf

    # Reconstruct image: inverse shift + inverse FFT
    f_ishift_lpf = np.fft.ifftshift(f_lpf)
    img_back_lpf = np.fft.ifft2(f_ishift_lpf)
    img_back_lpf = np.abs(img_back_lpf)

    # Smaller radius  -> fewer low frequencies kept -> stronger smoothing -> more blur
    # Larger radius   -> more low (and some mid) frequencies kept -> less blur

    # Plot result
    plt.subplot(1, 3, i)
    plt.imshow(img_back_lpf, cmap='gray')
    plt.title(f'LPF radius = {radius}')
    plt.axis('off')

plt.show()
````

````python
# C2. High Pass Filter


import numpy as np
import cv2
import matplotlib.pyplot as plt

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2    # center of frequency domain

radius = 30

# Create HPF mask:
mask_hpf = np.ones((rows, cols), np.uint8)
cv2.circle(mask_hpf, (ccol, crow), radius, 0, -1)  # 0 inside radius

# Apply HPF in frequency domain
f_hpf = fshift * mask_hpf

# Inverse FFT to reconstruct image
f_ishift_hpf = np.fft.ifftshift(f_hpf)
img_back_hpf = np.fft.ifft2(f_ishift_hpf)
img_back_hpf = np.abs(img_back_hpf)

# Show result
plt.figure(figsize=(6, 6))
plt.imshow(img_back_hpf, cmap='gray')
plt.title('High Pass Filter ')
plt.axis('off')
plt.show()
````

````python
# C3. Square Low-Pass Filter (LPF)

import numpy as np
import matplotlib.pyplot as plt

rows, cols = gray.shape
crow, ccol = rows // 2, cols // 2      # center of frequency domain

side = 50    # length of the square LPF
half = side // 2

# 1. Create Square LPF Mask

mask_lpf = np.zeros((rows, cols), np.uint8)

mask_lpf[crow - half : crow + half,
         ccol - half : ccol + half] = 1

# 2. Apply LPF
f_lpf = fshift * mask_lpf

# 3. Reconstruct using IFFT
f_ishift_lpf = np.fft.ifftshift(f_lpf)
img_back_lpf = np.fft.ifft2(f_ishift_lpf)
img_back_lpf = np.abs(img_back_lpf)


# 4. Show Output
plt.figure(figsize=(6, 6))
plt.imshow(img_back_lpf, cmap='gray')
plt.title("Square Low-Pass Filter (side = 50)")
plt.axis("off")
plt.show()


````

````python
# Your PART D codes follow here. You can create more cells under this. Write clean and commented code.
# D1. Phase-Only Reconstruction

import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 8))

phase_only = fshift / (np.abs(fshift))

# Inverse FFT
f_ishift_phase = np.fft.ifftshift(phase_only)
img_phase_only = np.fft.ifft2(f_ishift_phase)
img_phase_only = np.abs(img_phase_only)

plt.subplot(1, 2, 1)
plt.imshow(img_phase_only, cmap='gray')
plt.title("Phase-Only Reconstruction")
plt.axis("off")

# 2. Using np.angle(fshift)

theta = np.angle(fshift)

plt.subplot(1, 2, 2)
plt.imshow(theta, cmap='twilight')
plt.title(" Phase ")
plt.axis("off")

plt.show()

# Difference between two approaches
# fshift/abs(fshift) gives complex unit phasors (e^{iθ}) usable for reconstruction,
# whereas np.angle(fshift) gives only real angles θ, which cannot reconstruct the image.



````

````python
# D2 – Magnitude Only Reconstruction

import numpy as np
import matplotlib.pyplot as plt

mag_only = np.abs(fshift)

# Inverse FFT (phase = 0 everywhere)
f_ishift_mag = np.fft.ifftshift(mag_only)
img_back_mag = np.fft.ifft2(f_ishift_mag)
img_back_mag = np.abs(img_back_mag)

# Magnitude spectrum for visualization (log scaling)
magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.title('Magnitude Only Reconstruction')
plt.imshow(img_back_mag, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Magnitude Spectrum ')
plt.imshow(magnitude_spectrum, cmap='gray')
plt.axis('off')

plt.show()


# Notes:
# 1) Magnitude Spectrum uses log() only for visualization.
#    Frequency values vary a lot: low freq are huge, high freq are tiny.
#    log compresses this range so we can SEE both on the same image.
#
# 2) Magnitude-only reconstruction removes phase.
#    Without phase, the image loses structure and becomes blurry.

````
