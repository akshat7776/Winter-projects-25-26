````python
#Task 1
def convolve(img, kernel):
    img = img.astype(float)
    kernel = kernel.astype(float)

    # Flip kernel for true convolution
    kernel = np.flipud(np.fliplr(kernel))

    img_h, img_w = img.shape[:2]
    k_h, k_w = kernel.shape

    pad_h = k_h // 2
    pad_w = k_w // 2

    # Handle grayscale vs RGB
    if img.ndim == 2:
        img = img[:, :, None]

    pad_img = np.pad(
        img,
        ((pad_h, pad_h), (pad_w, pad_w), (0, 0)),
        mode='reflect'
    )

    out_img = np.zeros((img_h, img_w, img.shape[2]))

    for i in range(img_h):
        for j in range(img_w):
            region = pad_img[i:i+k_h, j:j+k_w, :]
            out_img[i, j] = np.sum(region * kernel[:, :, None], axis=(0, 1))

    if out_img.shape[2] == 1:
        return out_img[:, :, 0]

    return out_img
````

````python
def convolve(img, kernel):
    img = img.astype(float)
    kernel = kernel.astype(float)

    # Flip kernel (true convolution)
    kernel = np.flipud(np.fliplr(kernel))

    if img.ndim == 2:
        img = img[:, :, None]

    H, W, C = img.shape
    kH, kW = kernel.shape

    pad_h = kH // 2
    pad_w = kW // 2

    padded = np.pad(
        img,
        ((pad_h, pad_h), (pad_w, pad_w), (0, 0)),
        mode="reflect"
    )

    output = np.zeros((H, W, C))

    for i in range(H):
        for j in range(W):
            region = padded[i:i+kH, j:j+kW, :]
            output[i, j] = np.sum(region * kernel[:, :, None], axis=(0, 1))

    if output.shape[2] == 1:
        return output[:, :, 0]

    return output
````

````python
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize * ksize)
    return convolve(img, kernel)
````

````python
blur_gray_3 = average_blur_gray(gray, 3)
blur_gray_7 = average_blur_gray(gray, 7)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray, cmap="gray")
plt.title("Original Gray")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blur_gray_3, cmap="gray")
plt.title("3x3 Average Blur")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blur_gray_7, cmap="gray")
plt.title("7x7 Average Blur")
plt.axis("off")

plt.show()
````

````python
def average_blur_rgb(img, ksize):
    kernel = np.ones((ksize, ksize)) / (ksize * ksize)

    blurred = np.zeros_like(img, dtype=float)

    for c in range(3):
        blurred[:, :, c] = convolve(img[:, :, c], kernel)

    return blurred
````

````python
blur_rgb_3 = average_blur_rgb(img, 3)
blur_rgb_7 = average_blur_rgb(img, 7)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blur_rgb_3.astype(np.uint8))
plt.title("3x3 Average Blur RGB")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blur_rgb_7.astype(np.uint8))
plt.title("7x7 Average Blur RGB")
plt.axis("off")

plt.show()
````

````python
#Task 3
def gaussian_kernel(ksize, sigma):
    kernel = np.zeros((ksize, ksize))
    center = ksize // 2

    for i in range(ksize):
        for j in range(ksize):
            x = i - center
            y = j - center
            kernel[i, j] = np.exp(-(x**2 + y**2) / (2 * sigma**2))

    kernel /= np.sum(kernel)  # normalize
    return kernel
````

````python
kernel_5 = gaussian_kernel(5, 1.0)

plt.imshow(kernel_5, cmap="gray")
plt.colorbar()
plt.title("5x5 Gaussian Kernel (σ = 1.0)")
plt.show()

print("Kernel sum:", kernel_5.sum())
````

````python
gaussian_gray_5 = convolve(gray, gaussian_kernel(5, 1.0))
gaussian_gray_9 = convolve(gray, gaussian_kernel(9, 2.0))

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gaussian_gray_5, cmap="gray")
plt.title("Gaussian Blur 5x5 (σ=1)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gaussian_gray_9, cmap="gray")
plt.title("Gaussian Blur 9x9 (σ=2)")
plt.axis("off")

plt.show()
````

````python
def gaussian_blur_rgb(img, ksize, sigma):
    kernel = gaussian_kernel(ksize, sigma)
    blurred = np.zeros_like(img, dtype=float)

    for c in range(3):
        blurred[:, :, c] = convolve(img[:, :, c], kernel)

    return blurred
````

````python
gaussian_rgb_5 = gaussian_blur_rgb(img, 5, 1.0)
gaussian_rgb_9 = gaussian_blur_rgb(img, 9, 2.0)

plt.figure(figsize=(18,6))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gaussian_rgb_5.astype(np.uint8))
plt.title("Gaussian Blur RGB 5x5 (σ=1)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gaussian_rgb_9.astype(np.uint8))
plt.title("Gaussian Blur RGB 9x9 (σ=2)")
plt.axis("off")

plt.show()
````

````python
def magnitude_spectrum(img):
    # FFT
    F = np.fft.fft2(img)
    F_shift = np.fft.fftshift(F)

    # Magnitude spectrum (log for visibility)
    mag = np.log(1 + np.abs(F_shift))
    return mag
````

````python
mag_original = magnitude_spectrum(gray)

plt.imshow(mag_original, cmap="gray")
plt.axis("off")
plt.title("Magnitude Spectrum - Original Image")
plt.show()
````

````python
# Bright center indicates strong low-frequency components
# Outer darker regions represent weaker high-frequency content
# Natural images are dominated by low frequencies
````

````python
# Gaussian blur with moderate kernel
gauss_blur = convolve(gray, gaussian_kernel(9, 2.0))

mag_gauss = magnitude_spectrum(gauss_blur)

plt.imshow(mag_gauss, cmap="gray")
plt.axis("off")
plt.title("Magnitude Spectrum - Gaussian Blurred Image")
plt.show()
````

````python
# Gaussian blur suppresses high-frequency components
# Outer regions of spectrum become darker
# Center becomes relatively brighter
# Confirms Gaussian kernel acts as a LOW-PASS FILTER (LPF)
````

````python
alpha = 1.5
mask = gray - gauss_blur
unsharp = gray + alpha * mask
unsharp = np.clip(unsharp, 0, 255)

mag_unsharp = magnitude_spectrum(unsharp)

plt.imshow(mag_unsharp, cmap="gray")
plt.axis("off")
plt.title("Magnitude Spectrum - Unsharp Masked Image")
plt.show()
````

````python
# High-frequency components are amplified
# Outer regions become brighter compared to original
# Center intensity slightly reduces relative to edges
# Confirms unsharp masking enhances HIGH FREQUENCIES (HPF behavior)
````

````python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(mag_original, cmap="gray")
plt.title("Original Spectrum")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(mag_gauss, cmap="gray")
plt.title("Gaussian Blur Spectrum (LPF)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(mag_unsharp, cmap="gray")
plt.title("Unsharp Masking Spectrum (HPF)")
plt.axis("off")

plt.show()
````

````python
# Try different Gaussian sigmas and alpha values
sigmas = [0.8, 2.0, 4.0]
alphas = [0.5, 1.0, 2.0]

for sigma in sigmas:
    blur = convolve(gray, gaussian_kernel(9, sigma))
    mag = magnitude_spectrum(blur)

    plt.imshow(mag, cmap="gray")
    plt.title(f"Gaussian Spectrum (sigma={sigma})")
    plt.axis("off")
    plt.show()
````

````python
for alpha in alphas:
    unsharp = gray + alpha * (gray - gauss_blur)
    unsharp = np.clip(unsharp, 0, 255)

    mag = magnitude_spectrum(unsharp)
    plt.imshow(mag, cmap="gray")
    plt.title(f"Unsharp Spectrum (alpha={alpha})")
    plt.axis("off")
    plt.show()
````

````python
def rgb_to_hsv(img): # Takes a 0-255 (not normalised) RGB image
  # Normalizing the RGB value and change of dtype to ensure the same
  r,g,b = img[:,:,0].astype("float")/255, img[:,:,1].astype("float")/255, img[:,:,2].astype("float")/255

  # Calculating per pixel cmax and cmin
  cmax = np.maximum(np.maximum(r, g), b)
  cmin = np.minimum(np.minimum(r, g), b)
  delta = cmax - cmin

  h = np.zeros_like(cmax)
  s = np.zeros_like(cmax)
  v = cmax  # Value

  # Saturation
  for i in range(cmax.shape[0]):
    for j in range(cmax.shape[1]):
      if cmax[i][j] != 0:
        s[i][j] = delta[i][j] / cmax[i][j]
  # Hue
  for i in range(cmax.shape[0]):
    for j in range(cmax.shape[1]):
      if delta[i][j] != 0:
        if cmax[i][j] == r[i][j]:
          h[i][j] = ((g[i][j] - b[i][j]) / delta[i][j])
        elif cmax[i][j] == g[i][j]:
          h[i][j] = ((b[i][j] - r[i][j]) / delta[i][j]) + 2
        elif cmax[i][j] == b[i][j]:
          h[i][j] = ((r[i][j] - g[i][j]) / delta[i][j]) + 4
        h[i][j] = h[i][j] % 6.0
        h[i][j] = h[i][j] * 60  # Theoretical 0-360 range

  # Scale S,V
  s = (s * 255).astype(np.uint8)
  v = (v * 255).astype(np.uint8)
  # Returns an HSV image (also not normalised, compatible with OpenCV cvtColor)
  hsv = np.zeros_like(img, dtype=np.float32)
  hsv[:,:,0] = h/2  # Since HSV range for OpenCV is 0-179 and not the theoretical 0-360
  hsv[:,:,1] = s
  hsv[:,:,2] = v
  hsv = hsv.astype(np.uint8)
  return hsv
````
