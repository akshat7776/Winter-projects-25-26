````python
#A1
# Function gray_histogram : Function implementation using nupy and matplotlib only
def gray_histogram(gray_img):
  hist=np.zeros(256) # Initialization
  # pixel values iterates over the size of image(900 x 650)
  for i in range (gray_img.shape[0]):
    for j in range (gray_img.shape[1]):
      hist[gray_img[i,j]]+=1
      # increment by 1 over every iteration to get the count of pixels count at
      # a given intensity
  plt.figure(figsize=(18,6))

  # grayscale image
  plt.subplot(1,3,1)
  plt.imshow(gray_img,cmap="gray")
  plt.axis('off')

 # Histogram obtained using only numpy loops
  plt.subplot(1,3,2)
  plt.plot(range(256),hist)
  plt.xlabel('Intensity')
  plt.ylabel('Pixel Count')
  plt.title('Grayscale Histogram(Numpy Loops)')


  # Histogram obtained using matplotlib functions(just to compare the plots)
  plt.subplot(1,3,3)
  plt.hist(gray_img.ravel(),bins=256,range=(0,256))
  plt.xlabel('Intensity')
  plt.ylabel('Pixel Count')
  plt.title('Grayscale Histogram(Matplotlib)')

  plt.show()

gray_histogram(gray)
````

````python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# ---------- Safety checks ----------
# ensure img_rgb exists and is uint8 RGB (H x W x 3)
assert 'img_rgb' in globals(), "img_rgb not defined — decode/load the image first"
assert img_rgb.ndim == 3 and img_rgb.shape[2] == 3, "img_rgb must be HxWx3"
if img_rgb.dtype != np.uint8:
    # scale/convert if float in [0,1]
    if np.issubdtype(img_rgb.dtype, np.floating):
        img_rgb = (img_rgb * 255).astype(np.uint8)
    else:
        img_rgb = img_rgb.astype(np.uint8)

# ---------- Vectorized RGB -> HSV matching OpenCV ranges ----------
def rgb_to_hsv_vectorized(img_rgb_uint8):
    # convert to float for computation
    arr = img_rgb_uint8.astype(np.float32)
    r = arr[..., 0]
    g = arr[..., 1]
    b = arr[..., 2]

    cmax = np.maximum(np.maximum(r, g), b)
    cmin = np.minimum(np.minimum(r, g), b)
    delta = cmax - cmin

    # Hue calculation
    h = np.zeros_like(cmax, dtype=np.float32)

    # Avoid divisions by zero: where delta == 0, hue stays 0
    nonzero = delta > 1e-6

    # Masks for which channel is max
    mask_r = (cmax == r) & nonzero
    mask_g = (cmax == g) & nonzero
    mask_b = (cmax == b) & nonzero

    # compute per-case
    h[mask_r] = 60.0 * ((g[mask_r] - b[mask_r]) / delta[mask_r])       # 0..360
    h[mask_g] = 60.0 * (((b[mask_g] - r[mask_g]) / delta[mask_g]) + 2)
    h[mask_b] = 60.0 * (((r[mask_b] - g[mask_b]) / delta[mask_b]) + 4)

    # normalize negative hues
    h = (h + 360.0) % 360.0

    # OpenCV stores H in [0,179] (i.e. H/2)
    h = (h / 2.0).astype(np.uint8)    # 0..179 → stored in uint8

    # Saturation: if cmax == 0 → s = 0; else s = delta/cmax (0..1) scaled to 0..255
    s = np.zeros_like(cmax, dtype=np.float32)
    mask_cmax_nonzero = cmax > 1e-6
    s[mask_cmax_nonzero] = (delta[mask_cmax_nonzero] / cmax[mask_cmax_nonzero]) * 255.0
    s = np.clip(s, 0, 255).astype(np.uint8)

    # Value is simply cmax scaled to 0..255 (already 0..255)
    v = np.clip(cmax, 0, 255).astype(np.uint8)

    hsv = np.stack([h, s, v], axis=-1)
    return hsv

img_hsv_manual = rgb_to_hsv_vectorized(img_rgb)

# ---------- Compare with OpenCV ----------
img_hsv_opencv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

# Show if they are identical or where they differ
equal_map = (img_hsv_manual == img_hsv_opencv)
print("Per-channel equality percentages:",
      (equal_map[...,0].mean()*100, equal_map[...,1].mean()*100, equal_map[...,2].mean()*100))

# If small differences exist, inspect statistics of difference
diff = img_hsv_manual.astype(int) - img_hsv_opencv.astype(int)
print("Max absolute differences per channel:", np.max(np.abs(diff), axis=(0,1)))

# ---------- Visualization ----------
# Convert HSV (manual & OpenCV) back to RGB for display
rgb_from_manual = cv2.cvtColor(img_hsv_manual, cv2.COLOR_HSV2RGB)
rgb_from_opencv = cv2.cvtColor(img_hsv_opencv, cv2.COLOR_HSV2RGB)

plt.figure(figsize=(18,6))
plt.subplot(1,4,1)
plt.title("Original RGB")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(1,4,2)
plt.title("Manual HSV -> back to RGB")
plt.imshow(rgb_from_manual)
plt.axis("off")

plt.subplot(1,4,3)
plt.title("OpenCV HSV -> back to RGB")
plt.imshow(rgb_from_opencv)
plt.axis("off")

plt.subplot(1,4,4)
plt.title("Manual HSV (displayed as HSV->RGB) difference")
# show abs difference (visual)
plt.imshow(np.clip(np.abs(rgb_from_manual.astype(int)-rgb_from_opencv.astype(int)),0,255).astype(np.uint8))
plt.axis("off")

plt.show()
````

````python
def white_patch_balance(img_rgb):
  img_rgb_=img_rgb.astype(np.float32) # converting to float 32 type
  r,g,b=img_rgb_[:,:,0],img_rgb_[:,:,1],img_rgb_[:,:,2]

  # Brightest Pixel
  V=np.max(img_rgb_,axis=-1)
  idx=np.unravel_index(V.argmax(),V.shape)
  brightest_pixel=img_rgb_[idx]
  r_white,g_white,b_white=brightest_pixel[0],brightest_pixel[1],brightest_pixel[2]

  # Rescaling and Clipping
  scale_r,scale_g,scale_b=255/r_white,255/g_white,255/b_white
  r_new,g_new,b_new=r*scale_r,g*scale_g,b*scale_b
  img_rgb_=np.stack((r_new,g_new,b_new),axis=-1)
  img_wb=np.clip(img_rgb_, 0, 255).astype(np.uint8)

  #Obtaining the display for original and white balanced images
  plt.figure(figsize=(12,6))
  plt.subplot(1,2,1)
  plt.title('Original RGB image')
  plt.imshow(img_rgb)
  plt.axis('off')
  plt.subplot(1,2,2)
  plt.title('White Patch Balanced Image')
  plt.imshow(img_wb)
  plt.axis('off')
  plt.show()

  print(np.where(img_rgb!= img_wb)) # Just to compare where all the image pixels differ

white_patch_balance(img_rgb)
````
